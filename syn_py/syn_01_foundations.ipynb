## Prerequisite bridge: why a laser needs threshold and feedback

An excited atom can emit spontaneously with an uncontrolled phase, or it can undergo stimulated emission into an already occupied electromagnetic mode. **Population inversion** means that the pumped gain medium has more atoms available for stimulated downward transitions than for absorption. A resonator repeatedly returns selected field modes through the medium, but mirrors and the medium also produce losses.

Below threshold, losses dominate and phase-incoherent spontaneous emission remains small. At threshold, gain balances loss for one mode. Above threshold, the field stimulates further emission into its own phase and spatial pattern, while depletion of the inversion limits growth. Coherence is therefore produced by nonlinear field–matter feedback. The Maxwell–Bloch variables used below are collective summaries: field amplitude $E$, material polarization $P$ (phase-coherent dipole response), and inversion $D$ (available gain).


## The laser as Haken's prototype

Haken's synergetics grew from laser theory. A pumped gain medium contains many microscopic emitters whose spontaneous phases are initially incoherent. In a resonator, stimulated emission and field–matter feedback create a threshold: one cavity mode can acquire a stable macroscopic amplitude and phase relation. The result is not imposed phase-by-phase from outside; it is a collective instability of an open, driven, dissipative system.

In a real-variable, single-mode semiclassical reduction, let $E$ denote the cavity-field amplitude, $P$ the collective material polarization, and $D$ the population inversion:

$$\dot E=-\kappa E+gP,$$
$$\dot P=-\gamma_\perp P+gED,$$
$$\dot D=\gamma_\parallel(D_0-D)-2gEP.$$

Here $D_0$ is set by the pump, while $\kappa$, $\gamma_\perp$, and $\gamma_\parallel$ are field, polarization, and inversion relaxation rates. Linearizing the $(E,P)$ subsystem about the nonlasing state $(E,P,D)=(0,0,D_0)$ gives the threshold

$$D_c=\frac{\kappa\gamma_\perp}{g^2}.$$

Below $D_c$, field fluctuations decay. Above it, the nonlasing state is unstable and nonlinear depletion of $D$ saturates the coherent field.


### From quantum emitters to collective laser variables

The three equations above summarize several physical approximations. We now derive them before using them numerically. The purpose is to see how microscopic emitters and a common cavity field give rise to the collective variables of synergetics. No fast-variable elimination is needed yet; slaving is a further reduction discussed later.

Consider $N$ identical effective two-level emitters, with upper and lower laser levels $|e\rangle$ and $|g\rangle$. Define

$$\sigma_j^-=|g\rangle_j\langle e|,\qquad
\sigma_j^+=(\sigma_j^-)^\dagger,\qquad
\sigma_j^z=|e\rangle_j\langle e|-|g\rangle_j\langle g|.$$

We use three angular frequencies: $\omega_c$ is the resonance frequency of the selected empty-cavity mode, and $\omega_a=(E_e-E_g)/\hbar$ is the transition frequency of each emitter. The reference frequency $\omega_r$ specifies the optical phase factor $e^{-i\omega_r t}$ removed from the Heisenberg field operator to define its amplitude. It is a freely chosen reference, usually near the cavity and atomic resonances. Choosing $\omega_r=\omega_c$ gives zero cavity detuning $\Delta_c=\omega_c-\omega_r=0$; this and the atomic detuning $\Delta_a=\omega_a-\omega_r$ both vanish when $\omega_r=\omega_c=\omega_a$. All three are angular frequencies, measured in radians per second.

The lowering operator describes an atomic de-excitation; $\sigma_j^z$ has eigenvalues $+1$ and $-1$. For one cavity mode, $a$ annihilates a photon and $[a,a^\dagger]=1$. After separating the reference phase factor, the positive-frequency electric-field operator is approximated by

$$\widehat{\mathcal E}^{(+)}(\mathbf r,t)
=\mathcal E_{\rm vac}u(\mathbf r)a(t)e^{-i\omega_r t}.$$

Thus **single-mode** means retaining one spatial profile $u(\mathbf r)$ and its time-dependent amplitude. It does not mean that this amplitude is real. Assume identical transition frequencies and a uniform real atom–mode coupling $g$; this neglects inhomogeneous broadening and spatial variations of the coupling.

**Operator representation.** Let $a_H(t)$ and $\sigma^-_{j,H}(t)$ denote Heisenberg operators with their optical-frequency time dependence. We introduce amplitudes with the reference oscillation removed:

$$a(t)=e^{i\omega_r t}a_H(t),\qquad
\widetilde\sigma^-_j(t)=e^{i\omega_r t}\sigma^-_{j,H}(t),\qquad
\widetilde\sigma^z_j(t)=\sigma^z_{j,H}(t).$$

Consequently $a_H(t)=a(t)e^{-i\omega_r t}$, which explains the exponential in the electric-field formula above. For example, a free cavity obeys $\dot a_H=-i\omega_c a_H$, hence $\dot a=-i(\omega_c-\omega_r)a$. The corresponding atomic equation contains $\omega_a-\omega_r$.

Equivalently, choose the interaction representation with reference Hamiltonian

$$H_{\rm ref}=\hbar\omega_r\mathcal N,\qquad
\mathcal N=a^\dagger a+\frac12\sum_j(\sigma_j^z+1).$$

Within the rotating-wave approximation the interaction conserves $\mathcal N$. The transformed Hamiltonian is therefore $H_{\rm lab}-H_{\rm ref}$, up to an irrelevant constant, giving the detunings in the Hamiltonian below. We suppress the tildes on atomic operators from here onward. The conventional phrase “rotating frame” refers to this rotation of complex phases in operator space. The rotating-wave approximation is a separate step: it removes counter-rotating interaction terms.

In the electric-dipole and rotating-wave approximations, the Hamiltonian is

$$\frac{H}{\hbar}=\Delta_c a^\dagger a
+\frac{\Delta_a}{2}\sum_j\sigma_j^z
+g\sum_j(a^\dagger\sigma_j^-+a\sigma_j^+),$$

where $\Delta_c=\omega_c-\omega_r$ and $\Delta_a=\omega_a-\omega_r$. The term $a^\dagger\sigma_j^-$ creates a photon while de-exciting an emitter; its Hermitian conjugate describes absorption. The rotating-wave approximation discards rapidly oscillating terms that create or remove both excitations together.

Unlike the scaled numerical example, this microscopic model initially uses physical time: $g$, the detunings, and the decay rates have units of inverse time. Choosing a time unit makes them dimensionless.


### Pumping, losses, and the quantum master equation

The Hamiltonian alone cannot describe a working laser: pumping supplies energy and losses remove it. A minimal Markovian reservoir model is

$$\dot\rho=-\frac{i}{\hbar}[H,\rho]+2\kappa\mathcal D[a]\rho
+\sum_j\left(w\mathcal D[\sigma_j^+]\rho
+\Gamma\mathcal D[\sigma_j^-]\rho
+\frac{\gamma_\phi}{2}\mathcal D[\sigma_j^z]\rho\right),$$

$$\mathcal D[L]\rho=L\rho L^\dagger-\frac12\{L^\dagger L,\rho\}.$$

Here $\rho$ is the joint atom–field density operator. The dissipator $\mathcal D[L]$ represents the irreversible process associated with $L$: $w$ is an effective incoherent pump rate, $\Gamma$ is atomic decay into channels other than the retained cavity mode, and $\gamma_\phi$ destroys atomic phase coherence without changing populations. The effective pump represents a reservoir or auxiliary levels of a multilevel gain medium, rather than resonant coherent excitation of an isolated two-level atom.

The factor $2\kappa$ fixes a convention: $\kappa$ is the **field-amplitude** decay rate. For an empty cavity,

$$\frac{d}{dt}\langle a\rangle=-\kappa\langle a\rangle,
\qquad\frac{d}{dt}\langle a^\dagger a\rangle=-2\kappa\langle a^\dagger a\rangle.$$

For an uncoupled emitter, its inversion $z=\langle\sigma^z\rangle$ obeys

$$\dot z=w(1-z)-\Gamma(1+z)
=-(w+\Gamma)\left(z-\frac{w-\Gamma}{w+\Gamma}\right).$$

Consequently, for this reservoir model,

$$\gamma_\parallel=w+\Gamma,\qquad
\gamma_\perp=\frac{w+\Gamma}{2}+\gamma_\phi,\qquad
D_0=N\frac{w-\Gamma}{w+\Gamma}.$$

The pump and decay drive the total inversion toward $D_0$ when the cavity field is absent. Inversion requires $w>\Gamma$ in this model. More detailed multilevel models produce other effective relations; in the phenomenological calculations below, $D_0$ is varied while the relaxation rates are held fixed. That scan is not a literal variation of $w$ alone in this particular reservoir model.


### Expectation values: where the closure problem appears

Introduce the collective operators

$$S^-=\sum_j\sigma_j^-,\qquad S^+=(S^-)^\dagger,\qquad Z=\sum_j\sigma_j^z.$$

The expectation of an observable is $\langle O\rangle=\operatorname{Tr}(\rho O)$. Its evolution follows from

$$\frac{d}{dt}\langle O\rangle=\frac{i}{\hbar}\langle[H,O]\rangle
+\text{reservoir contributions}.$$

Using $[\sigma^+,\sigma^-]=\sigma^z$ and $[\sigma^z,\sigma^\pm]=\pm2\sigma^\pm$ gives

$$\begin{aligned}
\frac{d}{dt}\langle a\rangle
&=-(\kappa+i\Delta_c)\langle a\rangle-ig\langle S^-\rangle,\\
\frac{d}{dt}\langle S^-\rangle
&=-(\gamma_\perp+i\Delta_a)\langle S^-\rangle+ig\langle aZ\rangle,\\
\frac{d}{dt}\langle Z\rangle
&=\gamma_\parallel(D_0-\langle Z\rangle)
+2ig(\langle a^\dagger S^-\rangle-\langle aS^+\rangle).
\end{aligned}$$

For example, $[a^\dagger S^-+aS^+,S^-]=aZ$, which generates the field-times-inversion term in the polarization equation. These equations are exact within the stated quantum model. They are not closed, because averages of products such as $\langle aZ\rangle$ are not generally products of averages. Equations for these correlations would introduce further moments.


### Semiclassical factorization and normalization

Set $\alpha=\langle a\rangle$, $s=\langle S^-\rangle$, and $D=\langle Z\rangle$. The semiclassical mean-field approximation neglects connected atom–field correlations:

$$\langle aZ\rangle\simeq\alpha D,\qquad
\langle a^\dagger S^-\rangle\simeq\alpha^*s,\qquad
\langle aS^+\rangle\simeq\alpha s^*.$$

For example, $\langle aZ\rangle=\alpha D+\langle(a-\alpha)(Z-D)\rangle$, and we discard the second term. This gives

$$\begin{aligned}
\dot\alpha&=-(\kappa+i\Delta_c)\alpha-igs,\\
\dot s&=-(\gamma_\perp+i\Delta_a)s+ig\alpha D,\\
\dot D&=\gamma_\parallel(D_0-D)+2ig(\alpha^*s-\alpha s^*).
\end{aligned}$$

Define scaled complex amplitudes

$$\mathsf E=\sqrt2\,\alpha,\qquad\mathsf P=-i\sqrt2\,s.$$

The phase factor absorbs the $i$ in the coupling, and the scale factors fix the coefficients to match the notebook. The resulting complex Maxwell–Bloch equations are

$$\boxed{\begin{aligned}
\dot{\mathsf E}&=-(\kappa+i\Delta_c)\mathsf E+g\mathsf P,\\
\dot{\mathsf P}&=-(\gamma_\perp+i\Delta_a)\mathsf P+g\mathsf E D,\\
\dot D&=\gamma_\parallel(D_0-D)-2g\operatorname{Re}(\mathsf E^*\mathsf P).
\end{aligned}}$$

The polarization is a collective atomic coherence, not an excited-state population; populations enter through $D$. At this stage the model has five real variables: two components each for field and polarization, and one inversion. Factorization omits quantum correlations and fluctuation dynamics and is not a uniform approximation arbitrarily close to threshold or in a few-emitter, few-photon regime.


### Resonance, fixed phase, and the three real equations

Assume exact cavity–atom resonance and rotate at the common frequency, so $\Delta_c=\Delta_a=0$. The complex equations preserve the subspace

$$\mathsf E=e^{i\theta}E,\qquad\mathsf P=e^{i\theta}P,\qquad E,P\in\mathbb R,$$

where $\theta$ is constant. Choosing that common phase as our reference gives exactly

$$\boxed{\begin{aligned}
\dot E&=-\kappa E+gP,\\
\dot P&=-\gamma_\perp P+gED,\\
\dot D&=\gamma_\parallel(D_0-D)-2gEP.
\end{aligned}}$$

This is a fixed-phase sector of the resonant model. A common phase choice cannot in general make two arbitrarily phased complex amplitudes real simultaneously; detuning and general relative-phase dynamics require the complex equations.

An excitation-balance check explains the factor $2$. The coherent photon population is $n_{\rm coh}=|\alpha|^2=E^2/2$, and the mean upper-level population is $N_e=(N+D)/2$. Keeping only interaction terms,

$$\left.\dot n_{\rm coh}\right|_{\rm int}=gEP,\qquad
\left.\dot N_e\right|_{\rm int}=-gEP.$$

One de-excitation adds one photon, but decreases $D=N_e-N_g$ by two. The plots below use $E^2$ as a scaled intensity, proportional to photon population. In the full quantum theory $\langle a^\dagger a\rangle$ need not equal $|\langle a\rangle|^2$.


### From collective instability to the order parameter

Linearizing about $(E,P,D)=(0,0,D_0)$ gives $\delta\dot D=-\gamma_\parallel\delta D$ and

$$\frac{d}{dt}\begin{pmatrix}\delta E\\\delta P\end{pmatrix}
=\begin{pmatrix}-\kappa&g\\gD_0&-\gamma_\perp\end{pmatrix}
\begin{pmatrix}\delta E\\\delta P\end{pmatrix}.$$

The two growth rates solve

$$\lambda^2+(\kappa+\gamma_\perp)\lambda
+\kappa\gamma_\perp-g^2D_0=0.$$

The trace is negative and the determinant changes sign at $D_c=\kappa\gamma_\perp/g^2$. Thus a collective field–polarization perturbation changes from decaying to growing at threshold. At $D_0=D_c$, one eigenvalue is zero and the other is $-(\kappa+\gamma_\perp)$; the critical eigenvector satisfies $\delta P=(\kappa/g)\delta E$. The slow variable therefore represents a coupled field–matter mode, which can be parametrized by $E$.

For a nonzero stationary field the first two equations imply

$$P_s=\frac{\kappa}{g}E_s,\qquad D_s=D_c.$$

The third then gives

$$E_s^2=\frac{\gamma_\parallel}{2\kappa}(D_0-D_c).$$

The field depletes the inversion until gain balances loss. This is the stationary branch plotted below; its stability farther above threshold requires a separate analysis. The field both results from collective polarization and changes the medium that generates it: the circular causality central to Haken's interpretation.

The deterministic equations leave exactly $E=P=0$ unchanged even above threshold. The small seed in the numerical example represents a perturbation; quantum fluctuations provide the physical startup mechanism. In an unseeded finite quantum laser the ensemble mean $\langle a\rangle$ can remain zero because the phase is not selected, despite a nonzero photon population. A nonzero semiclassical amplitude describes a selected phase or a realization over times where phase diffusion can be neglected.

The derivation has retained all three material–field variables. The later slaving approximation needs an additional time-scale argument: near threshold the collective growth rate becomes small compared with noncritical relaxation rates. Simply calling the model semiclassical does not justify setting $\dot P$ or $\dot D$ to zero.

**Further reading.** G. J. de Valcárcel, E. Roldán, and F. Prati, [*Semiclassical Theory of Amplification and Lasing*](https://arxiv.org/abs/quant-ph/0605084), develops the Maxwell–Bloch description and its relation to multilevel gain media. Haken's *Synergetics* supplies the course's interpretation in terms of instability, order parameters, and slaving.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "axes.grid": True,
    "grid.alpha": 0.25,
    "font.size": 11,
})
rng = np.random.default_rng(20260904)

from scipy.integrate import solve_ivp

kap, gamma_perp, gamma_parallel, g = 1.0, 8.0, 0.7, 2.0
D_c = kap*gamma_perp/g**2

def laser(t, state, D0):
    E, P, D = state
    return (-kap*E + g*P,
            -gamma_perp*P + g*E*D,
            gamma_parallel*(D0-D) - 2*g*E*P)

t_eval = np.linspace(0, 60, 4000)
pumps = [0.7*D_c, 1.02*D_c, 1.6*D_c]

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
steady_intensity = []
for D0 in pumps:
    sol = solve_ivp(laser, (0, 60), [1e-4, 0.0, D0], args=(D0,), t_eval=t_eval,
                    rtol=1e-9, atol=1e-11)
    E, P, D = sol.y
    ax[0].plot(sol.t, E**2, label=fr"$D_0/D_c={D0/D_c:.2f}$")
    steady_intensity.append(np.mean(E[sol.t > 50]**2))
ax[0].set(xlabel="time", ylabel=r"intensity $E^2$", title="Laser threshold dynamics")
ax[0].legend()

pump_scan = np.linspace(0.4*D_c, 2.0*D_c, 60)
theory = gamma_parallel*np.maximum(pump_scan-D_c, 0)/(2*kap)
ax[1].plot(pump_scan/D_c, theory, lw=2)
ax[1].set(xlabel=r"pump $D_0/D_c$", ylabel=r"stationary $E^2$",
          title="Order parameter above threshold")
plt.tight_layout()
```

## Slaving and circular causality

If material variables relax faster than the field, set $\dot P\approx0$ and $\dot D\approx0$:

$$P\approx\frac{g}{\gamma_\perp}ED,\qquad
D\approx\frac{D_0}{1+2g^2E^2/(\gamma_\perp\gamma_\parallel)}.$$

Substitution into the field equation and expansion near $E=0$ gives

$$\dot E\approx \mu E-\beta E^3,\qquad
\mu=\kappa\left(\frac{D_0}{D_c}-1\right),\quad \beta>0.$$

The coherent field $E$ is the slow order parameter. The fast atomic variables are approximately functions of it: they are **slaved**. Yet the atoms generate the field through stimulated polarization, while the field organizes the atoms. Haken used this feedback as the paradigmatic case of circular causality: microscopic activity creates the macroscopic order parameter, which then constrains microscopic activity.


```python
# Test the slaving relation on the above-threshold trajectory.
D0 = 1.6*D_c
sol = solve_ivp(laser, (0, 60), [1e-4, 0.0, D0], args=(D0,), t_eval=t_eval,
                rtol=1e-9, atol=1e-11)
E, P, D = sol.y
P_slaved = g*E*D/gamma_perp
D_slaved = D0/(1 + 2*g**2*E**2/(gamma_perp*gamma_parallel))
mask = sol.t > 10

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].plot(sol.t, P, label="full P")
ax[0].plot(sol.t, P_slaved, "--", label="slaved P(E,D)")
ax[0].set(xlabel="time", ylabel="polarization", title="Fast polarization follows the field")
ax[0].legend()
ax[1].plot(E[mask]**2, D[mask], ".", ms=1.5, label="full dynamics")
order = np.argsort(E[mask]**2)
ax[1].plot(E[mask][order]**2, D_slaved[mask][order], "k--", label="slaved inversion")
ax[1].set(xlabel=r"intensity $E^2$", ylabel="inversion D", title="Gain saturation")
ax[1].legend()
plt.tight_layout()
```
