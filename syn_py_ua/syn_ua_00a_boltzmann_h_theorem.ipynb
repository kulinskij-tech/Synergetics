## Від мікроскопічного руху до одночастинкового опису

Для системи $N$ класичних частинок визначимо мікроскопічну фазову густину
<!--  -->
\begin{equation}\widehat n(\mathbf r,\mathbf p,t)=\sum_{i=1}^N
\delta^{(3)}(\mathbf r-\mathbf r_i(t))\delta^{(3)}(\mathbf p-\mathbf p_i(t)),
\qquad \int\widehat n\,d^3r\,d^3p=N.
\label{eq:nmicro}
\end{equation}
<!--  -->
Це сингулярна функція, яка фіксує точні положення частинок, а не гладка густина ймовірності. Усереднення за приготуваннями системи дає розподіл числа частинок:

$$f_1(\mathbf r,\mathbf p,t)=\langle\widehat n\rangle,\qquad
n(\mathbf r,t)=\int f_1\,d^3p,\qquad
\int f_1\,d^3r\,d^3p=N.$$

Отже, $f_1\,d^3r\,d^3p$ — середнє число частинок в елементі фазового простору. За $n>0$ величина $f/n$ є умовною густиною ймовірності імпульсу. Надалі скрізь використовуємо $f$, щоб не змішувати ймовірнісне нормування та нормування на число частинок. Через середній імпульс і теплову енергію визначаємо $\mathbf u$ та $T$:

$$n m\mathbf u=\int\mathbf p f\,d^3p,\qquad
\frac32 n k_BT=\int\frac{|\mathbf p-m\mathbf u|^2}{2m}f_1\,d^3p.$$

Точне рівняння для одночастинкового розподілу містить двочастинковий розподіл; рівняння для останнього — тричастинковий. Це ієрархія ББҐКІ. Розклад на добуток і зв'язану парну кореляцію явно виявляє проблему замикання:

$$f^{(2)}=f f+g^{(2)},$$

за узгодженого нормування редукованих густин (поправки до підрахунку пар за скінченного $N$ несуттєві в розрідженій кінетичній границі). Головне питання полягає в тому, якими кореляціями й коли можна знехтувати, а не лише в тому, як усереднити рівняння Ньютона.


### Мікроскопічна фазова густина задовольняє рівняння неперервності

Щоб перейти від механіки частинок до полів суцільного середовища, спочатку продиференціюємо дельта-функції у $\widehat n$. Позначимо $\delta_a=\delta^{(3)}(\mathbf r-\mathbf r_a(t))\delta^{(3)}(\mathbf p-\mathbf p_a(t))$. З рівнянь Ньютона випливає, у сенсі узагальнених функцій,

$$\partial_t\delta_a=-\mathbf v_a\cdot\nabla_{\mathbf r}\delta_a-\mathbf F_a\cdot\nabla_{\mathbf p}\delta_a.$$

Підсумовуючи за частинками та використовуючи $\mathbf v_a\delta_a=(\mathbf p/m)\delta_a$, отримуємо __точну тотожність__:

$$\partial_t\widehat n+\nabla_{\mathbf r}\cdot(\mathbf v\widehat n)
+\nabla_{\mathbf p}\cdot\widehat{\mathbf J}_p=0,
\qquad \widehat{\mathbf J}_p=\sum_a\mathbf F_a\delta_a,
\qquad \mathbf v=\mathbf p/m.$$

Це диференціяльна форма закону збереження числа частинок \eqref{eq:nmicro} у шестивимірному одночастинковому фазовому просторі. Для заданої зовнішньої сили її внесок у $\widehat{\mathbf J}_p$ дорівнює $\mathbf F_{\rm ext}\widehat n$. Міжчастинкові сили залежать від інших частинок, тому їхнє середнє загалом не можна замінити добутком заданої сили на $f$. Для гладкої парної сили $\mathbf K(\mathbf r-\mathbf r')$ усереднення дає

$$\langle\widehat{\mathbf J}_p\rangle=\mathbf F_{\rm ext}f+
\int\mathbf K(\mathbf r-\mathbf r')f^{(2)}(\mathbf r,\mathbf p;\mathbf r',\mathbf p',t)\,d^3r'\,d^3p'.$$

Тут $f^{(2)}=\langle\sum_{a\ne b}\delta(\mathbf r-\mathbf r_a)\delta(\mathbf p-\mathbf p_a)\delta(\mathbf r'-\mathbf r_b)\delta(\mathbf p'-\mathbf p_b)\rangle$ враховує впорядковані пари різних частинок. Так виникає перше рівняння ланцюжка ББГКІ. Взаємодію твердих куль описують правилами зіткнень, а не гладкою парною силою.

Інтегрування рівняння неперервності у фазовому просторі за імпульсами усуває дивергенцію за імпульсом, якщо відповідний граничний внесок зникає. Отже, закон збереження числа частинок переходить до усередненого опису **без припущення про замикання**:

$$\partial_t n+\nabla\cdot(n\,\mathbf u)=0.$$

Газокінетичне наближення вводиться далі: дію коротких парних зіткнень представляють інтегралом зіткнень Больцмана. Щоб замкнути рівняння суцільного середовища, потрібне ще одне, окреме наближення — повільна зміна полів на масштабі довжини вільного пробігу. У наведеному нижче виведенні методом моментів, після розділу про розподіл Максвелла, ці два етапи розмежовано.


### За вихідним текстом: усереднення, часткові розподіли та перше рівняння ланцюжка

Розпишемо переходи між `fn–fschast` та `uravnf1–force` у джерелі. Нехай $W_N(x_1,\ldots,x_N,t)$ — симетрична густина ймовірності повної системи, нормована на одиницю, а $x_a=(\mathbf r_a,\mathbf p_a)$. Позначимо середню густину числа частинок через $n=N/V_{\rm box}$; $U_{ab}=U(|\mathbf r_a-\mathbf r_b|)$ — парний потенціал. Часткові функції джерела пов’язані з нашими числовими густинами так:

$$f_s^{\rm src}=\left(\frac{N}{n}\right)^s\int W_N\,dx_{s+1}\cdots dx_N,\qquad
f=n f_1^{\rm src},\qquad
f^{(2)}=n^2\left(1-\frac1N\right)f_2^{\rm src}\simeq n^2 f_2^{\rm src}.$$

Тут стала $n$ означає середню густину. Локальну густину явно записуємо як $n(\mathbf r,t)=\int f\,d^3p=n\int f_1^{\rm src}\,d^3p$; у неоднорідному газі вона не обов’язково дорівнює $n$. Це розрізнення запобігає появі зайвого множника густини в рівняннях моментів. Зокрема,

$$\langle\widehat n(x)\rangle=\sum_{a=1}^N\int\delta(x-x_a)W_N\,dx_1\cdots dx_N
=N\int W_N(x,x_2,\ldots)\,dx_2\cdots dx_N.$$

Для $\mathbf K_{ab}=-\nabla_{\mathbf r_a}U_{ab}$ рівняння Ліувілля має вигляд

$$\partial_tW_N+\sum_a\mathbf v_a\cdot\nabla_{\mathbf r_a}W_N
+\sum_a\left(\mathbf F_{0a}+\sum_{b\ne a}\mathbf K_{ab}\right)\cdot\nabla_{\mathbf p_a}W_N=0.$$

Інтегруємо за $x_2,\ldots,x_N$ та множимо на $N$. Усі дивергенції за змінними інтегрування зникають за періодичних граничних умов або достатнього спадання. $N-1$ однакових силових доданків для частинки 1 об’єднуються в парну густину:

$$\left(\partial_t+\mathbf v_1\cdot\nabla_{\mathbf r_1}+\mathbf F_{01}\cdot\nabla_{\mathbf p_1}\right)f(x_1)
=-\int\mathbf K_{12}\cdot\nabla_{\mathbf p_1}f^{(2)}(x_1,x_2)\,dx_2.$$

За нормуванням джерела коефіцієнт праворуч дорівнює $n(1-1/N)$; у використаному там наближенні великої системи він переходить у $n$. Для $g^{(2)}=f^{(2)}-f(x_1)f(x_2)$ внесок добутку дорівнює

$$-\int\mathbf K_{12}\cdot\nabla_{\mathbf p_1}[f(x_1)f(x_2)]\,dx_2
=-\mathbf F_{\rm mf}(\mathbf r_1,t)\cdot\nabla_{\mathbf p_1}f(x_1),\qquad
\mathbf F_{\rm mf}=\int\mathbf K_{12}f(x_2)\,dx_2.$$

Перенесемо його ліворуч. Отримуємо форму `uravnf11`:

$$[\partial_t+\mathbf v\cdot\nabla_{\mathbf r}+(\mathbf F_0+\mathbf F_{\rm mf})\cdot\nabla_{\mathbf p}]f
=-\int\mathbf K_{12}\cdot\nabla_{\mathbf p_1}g^{(2)}\,dx_2.$$

Праву частину визначають кореляції, але це ще не локальний інтеграл зіткнень Больцмана. Точне рівняння для парного розподілу містить наступну невідому функцію:

$$[\partial_t+\mathcal L_{12}]f^{(2)}
=-\int(\mathbf K_{13}\cdot\nabla_{\mathbf p_1}+\mathbf K_{23}\cdot\nabla_{\mathbf p_2})f^{(3)}\,dx_3,$$
$$\mathcal L_{12}=\sum_{a=1}^{2}[\mathbf v_a\cdot\nabla_{\mathbf r_a}+\mathbf F_{0a}\cdot\nabla_{\mathbf p_a}]
+\mathbf K_{12}\cdot(\nabla_{\mathbf p_1}-\nabla_{\mathbf p_2}).$$

Тепер видно загальну структуру ланцюжка. Для $1\le s<N$ означимо

$$f^{(s)}=\frac{N!}{(N-s)!}\int W_N\,dx_{s+1}\cdots dx_N.$$
$$\left[\partial_t+\sum_{a=1}^{s}(\mathbf v_a\cdot\nabla_{\mathbf r_a}+\mathbf F_{0a}\cdot\nabla_{\mathbf p_a})
+\sum_{a<b\le s}\mathbf K_{ab}\cdot(\nabla_{\mathbf p_a}-\nabla_{\mathbf p_b})\right]f^{(s)}
=-\sum_{a=1}^{s}\int\mathbf K_{a,s+1}\cdot\nabla_{\mathbf p_a}f^{(s+1)}\,dx_{s+1}.$$

Ліворуч стоять вільний рух, зовнішні сили та взаємодії всередині вибраної групи з $s$ частинок. Праворуч — взаємодії з частинкою поза цією групою, усереднені за її станом; саме вони вводять $f^{(s+1)}$. Ланцюжок закінчується за $s=N$, коли зовнішньої щодо групи частинки немає. За нормуванням часткових функцій у джерелі праворуч є коефіцієнт $n(1-s/N)$, наближено $n$ за фіксованого $s\ll N$. Отже, потреба в замиканні є фізичним наслідком взаємодій.

У джерелі $f^{(3)}$ розкладено на добуток, три внески парних кореляцій та зв’язану потрійну кореляцію. Нехтування останньою є одним із кроків розрідженого бінарного наближення; нехтування перекриттям актів взаємодії та умова для вхідних кореляцій — додаткові припущення. Далі в кінетиці розрідженого газу з короткодіючими силами залишаємо задану зовнішню силу, а зіткнення описуємо через $C[f]$. Ті самі взаємодії не враховуємо повторно як середнє поле; його збереження потребувало б узгодженого балансу енергії взаємодії.


### Кореляційні функції: структура ланцюжка та бінарне замикання

Це перехід `g2–g3` → `urf2_1` → `i1,i123` → `urf22` з вихідних лекцій. Тут під «ланцюжком» розуміємо **ієрархію ББГКІ**; наведена далі релаксаційна модель БГК є іншим наближенням. Щоб не плутати значення одночастинкової функції з двочастинковою функцією, позначимо $a_i=f(x_i,t)$, $F_{ij}=f^{(2)}(x_i,x_j,t)$ та $F_{123}=f^{(3)}(x_1,x_2,x_3,t)$. Означимо зв’язані кореляції:

$$F_{12}=a_1a_2+G_{12},$$
$$F_{123}=a_1a_2a_3+G_{12}a_3+G_{13}a_2+G_{23}a_1+G_{123}
=F_{12}a_3+G_{13}a_2+G_{23}a_1+G_{123}.$$

Це означення, а не наближення. $a_1a_2a_3$ описує незалежні частинки; $G_{12}a_3$ — корельовану пару та незалежну третю частинку, ще два доданки відповідають іншим парам. $G_{123}$ — частина тричастинкового розподілу, яка не представлена жодним із цих добутків. Це не вся тричастинкова функція. Тому нехтування $G_{123}$ **не** означає занулення $F_{123}$. За фіксованого скінченного $N$ вибране нормування числових густин також включає до зв’язаних функцій поправки підрахунку частинок; у розрідженому наближенні великої системи ними нехтуємо.

Перше рівняння ланцюжка після виділення середнього поля має вигляд

$$\mathcal D_i a_i=I_i,\qquad
\mathcal D_i=\partial_t+\mathbf v_i\cdot\nabla_{\mathbf r_i}
+(\mathbf F_{0i}+\mathbf F_{{\rm mf},i})\cdot\nabla_{\mathbf p_i},\qquad
I_i=-\int\mathbf K_{i3}\cdot\nabla_{\mathbf p_i}G_{i3}\,dx_3.$$

Тепер підставимо повний розклад $F_{123}$ у парне рівняння. Позначимо

$$\mathcal D_{12}=\partial_t+\sum_{i=1}^{2}[\mathbf v_i\cdot\nabla_{\mathbf r_i}
+(\mathbf F_{0i}+\mathbf F_{{\rm mf},i})\cdot\nabla_{\mathbf p_i}],\qquad
\mathcal L_{\rm int}=\mathbf K_{12}\cdot(\nabla_{\mathbf p_1}-\nabla_{\mathbf p_2}).$$

У $\mathcal D_{12}$ є **одна** похідна за часом. Окремі внески отримуємо так:

| Частина $F_{123}$ | Внесок у парне рівняння | Фізичний зміст |
|---|---|---|
| $F_{12}a_3$ | $-\mathbf F_{{\rm mf},1}\cdot\nabla_{\mathbf p_1}F_{12}-\mathbf F_{{\rm mf},2}\cdot\nabla_{\mathbf p_2}F_{12}$ | Середня сила від частинки 3; переносимо до частини перенесення |
| $G_{13}a_2$ під дією $\mathbf K_{13}\cdot\nabla_{\mathbf p_1}$ | $a_2 I_1$ | Зміна розподілу частинки 1 через її кореляції з іншими |
| $G_{23}a_1$ під дією $\mathbf K_{23}\cdot\nabla_{\mathbf p_2}$ | $a_1 I_2$ | Відповідна зміна для частинки 2 |
| Перехресна дія на $G_{13}a_2$ та $G_{23}a_1$ | $X_{12}$ нижче | Третя частинка зв’язує обох членів пари |
| $G_{123}$ | $T_{12}$ нижче | Внесок зв’язаної тричастинкової кореляції |

Наприклад, прямий доданок дорівнює $-\int\mathbf K_{13}\cdot\nabla_{\mathbf p_1}(G_{13}a_2)\,dx_3=a_2I_1$, оскільки $a_2$ не залежить від $\mathbf p_1$. Перехресні доданки:

$$X_{12}=-\nabla_{\mathbf p_1}a_1\cdot\int\mathbf K_{13}G_{23}\,dx_3
-\nabla_{\mathbf p_2}a_2\cdot\int\mathbf K_{23}G_{13}\,dx_3,$$
$$T_{12}=-\int(\mathbf K_{13}\cdot\nabla_{\mathbf p_1}+\mathbf K_{23}\cdot\nabla_{\mathbf p_2})G_{123}\,dx_3.$$

Отже, перегруповане парне рівняння має вигляд

$$\boxed{(\mathcal D_{12}+\mathcal L_{\rm int})F_{12}
=a_2I_1+a_1I_2+X_{12}+T_{12}.}$$

Правило диференціювання добутку та перше рівняння ланцюжка дають $\mathcal D_{12}(a_1a_2)=a_2I_1+a_1I_2$. Віднімемо цю тотожність, зберігаючи пряму взаємодію, й отримаємо рівняння, що показує виникнення кореляцій:

$$\boxed{(\mathcal D_{12}+\mathcal L_{\rm int})G_{12}
=-\mathcal L_{\rm int}(a_1a_2)+X_{12}+T_{12}.}$$

Навіть якщо спочатку пара некорельована, її пряма взаємодія породжує $G_{12}$. Занулення всіх парних кореляцій усунуло б сам механізм зіткнень, який ми прагнемо вивести.

Для короткодіючих сил у розрідженому газі джерело нехтує зв’язаними потрійними кореляціями ($T_{12}$) та перекриттям актів взаємодії, представленим $X_{12}$. Останнє є додатковим припущенням бінарного наближення, а не алгебричним наслідком $G_{123}=0$; воно також потребує контролю вхідних кореляцій і повторних зіткнень. Тоді

$$ (\mathcal D_{12}+\mathcal L_{\rm int})F_{12}
\simeq\mathcal D_{12}(a_1a_2).$$

Разом із $\mathcal D_1a_1=I_1[G_{12}]$ це наближено замкнена парна система джерела. На наступному кроці повільну одночастинкову зміну фіксуємо протягом короткого акту взаємодії. На цьому часовому масштабі права частина має вищий порядок, і залишається рівняння характеристик двочастинкового руху для $F_{12}$. **Інтеграл зіткнень отримуємо, розв’язавши цю парну задачу з вхідною граничною умовою та підставивши результат у перше рівняння ланцюжка.**


## Розріджений газ, молекулярний хаос і парні зіткнення

Розглянемо класичний розріджений газ однакових частинок із короткодіючою взаємодією. Якщо $r_0$ — радіус взаємодії, умова $n r_0^3\ll1$ дозволяє нехтувати одночасними багаточастинковими зіткненнями. Тривалість зіткнення мала порівняно із середнім часом вільного пробігу, а розподіл слабко змінюється в області зіткнення. Враховуємо пружні парні зіткнення та нехтуємо квантовою статистикою.

**Припущення молекулярного хаосу** факторизує розподіл частинок, що зближуються перед зіткненням:

$$f^{(2)}_{\rm in}(\mathbf r,\mathbf p_1,\mathbf p_2,t)
\simeq f(\mathbf r,\mathbf p_1,t)f(\mathbf r,\mathbf p_2,t).$$

Це припущення про вхідні пари на кінетичному масштабі, а не твердження про відсутність кореляцій після зіткнень. Частинки, що розлітаються, загалом корельовані. Саме таке фізичне замикання відповідає використанню імпульсів частинок у віддаленому минулому.

Для однакових мас позначимо $\mathbf q=\mathbf p_1-\mathbf p_2$, а через $\mathbf n$ — нормаль у точці контакту. Зручне правило зіткнення твердих куль:

$$\mathbf p'_1=\mathbf p_1-(\mathbf q\cdot\mathbf n)\mathbf n,\qquad
\mathbf p'_2=\mathbf p_2+(\mathbf q\cdot\mathbf n)\mathbf n.$$

Воно зберігає сумарний імпульс і кінетичну енергію пари:

$$\mathbf p_1+\mathbf p_2=\mathbf p'_1+\mathbf p'_2,\qquad
\frac{p_1^2+p_2^2}{2m}=\frac{p_1'^2+p_2'^2}{2m}.$$

Перетворення оборотне. Отже, втрату симетрії щодо обернення часу в замкненому кінетичному описі не можна пояснювати непружністю цих зіткнень.


### Від характеристик парного руху до різниці приходу та виходу

Розпишемо `urf220–collint_boltz1`. Нехай $r_0$ — радіус дії сил, $v_T$ — характерна відносна швидкість, $\epsilon=nr_0^3\ll1$. Оцінки дають

$$\tau_0\sim r_0/v_T,\qquad \ell\sim1/(nr_0^2),\qquad
\tau_{\rm coll}\sim\ell/v_T,\qquad \tau_0/\tau_{\rm coll}\sim\epsilon.$$

Тривалість одного акту взаємодії значно менша за середній час між зіткненнями. Протягом цього акту повільно змінні одночастинкові поля можна вважати сталими; потрійними зіткненнями нехтуємо. Це розділення масштабів відрізняється від подальшої умови суцільного середовища $\ell/L\ll1$.

На характеристиках парного руху в головному порядку

$$[\partial_t+\mathcal L_{12}]f^{(2)}\simeq0,\qquad
\frac{d}{ds}f^{(2)}(x_1(s),x_2(s),t+s)\simeq0.$$

Задамо факторизацію на вхідній гілці, до взаємодії. Якщо $\mathbf P_1,\mathbf P_2$ — асимптотичні вхідні імпульси траєкторії, що проходить через $(x_1,x_2)$, то локально

$$f^{(2)}(x_1,x_2,t)\simeq f(\mathbf r_1,\mathbf P_1,t)f(\mathbf r_1,\mathbf P_2,t),$$
$$\mathbf P_1+\mathbf P_2=\mathbf p_1+\mathbf p_2,\qquad
\frac{P_1^2+P_2^2}{2m}=\frac{p_1^2+p_2^2}{2m}+U(r_{12}).$$

Асимптотичні імпульси залежать від поточного відносного положення та обох імпульсів. Отже, під час зіткнення це не факторизація за поточними імпульсами: кореляції, породжені взаємодією, зберігаються.

Для визначеності покладемо $\mathbf r=\mathbf r_1-\mathbf r_2$, $\mathbf g=\mathbf v_1-\mathbf v_2$. Після фіксації повільних змінних парне рівняння має вигляд

$$\mathbf g\cdot\nabla_{\mathbf r}f^{(2)}
+\mathbf K_{12}\cdot(\nabla_{\mathbf p_1}-\nabla_{\mathbf p_2})f^{(2)}\simeq0.$$

У першому рівнянні ланцюжка силовий внесок переходить у $\int\mathbf g\cdot\nabla_{\mathbf r}f^{(2)}\,d^3r\,d^3p_2$: додаткова похідна за $\mathbf p_2$ інтегрується в нуль, оскільки $\mathbf K_{12}$ не залежить від імпульсу. За потреби перед просторовим інтегруванням віднімаємо сталий некорельований фон. Виберемо вісь $z$ уздовж $\mathbf g$ та поперечні прицільні координати $(b,\varphi)$. Тоді

$$\mathbf g\cdot\nabla_{\mathbf r}=|\mathbf g|\partial_z,\qquad
d^3r=b\,db\,d\varphi\,dz,$$
$$\int_{-\infty}^{\infty}|\mathbf g|\partial_z f^{(2)}\,dz
=|\mathbf g|[f^{(2)}(+\infty)-f^{(2)}(-\infty)].$$

На вхідному кінці отримуємо $f(\mathbf p_1)f(\mathbf p_2)$. На вихідному кінці розподіл виражається через вхідні імпульси оберненого акту розсіяння, позначені $\mathbf p'_1,\mathbf p'_2$. Оборотність забезпечує однакову міру зіткнень, тому

$$C[f](\mathbf p_1)=\int d^3p_2\int b\,db\,d\varphi\,|\mathbf g|
[f(\mathbf p'_1)f(\mathbf p'_2)-f(\mathbf p_1)f(\mathbf p_2)].$$

Нарешті, $b\,db\,d\varphi=(d\sigma/d\Omega)d\Omega$ (за потреби з підсумовуванням гілок розсіяння) дає $B=|\mathbf g|\,d\sigma/d\Omega$ у наступному розділі. Після обчислення граничних значень **інтеграла за $dz$ більше немає**. Повторний $dz$ в остаточній формулі джерела є друкарським залишком попереднього рядка. Вирішальне статистичне припущення — вхідна гранична умова, а не оборотне відображення розсіяння.


## Рівняння Больцмана як різниця приходу й виходу

За $\mathbf v=\mathbf p/m$ та зовнішньої сили $\mathbf F(\mathbf r,t)$, незалежної від імпульсу,

$$\left(\partial_t+\mathbf v\cdot\nabla_{\mathbf r}
+\mathbf F\cdot\nabla_{\mathbf p}\right)f=C[f].$$

За фіксованих $(\mathbf r,t)$ скоротимо позначення: $f_i=f(\mathbf p_i)$, $f'_i=f(\mathbf p'_i)$. Тоді

$$C[f](\mathbf p_1)=\int d^3p_2\,d\Omega\,
B(\mathbf p_1,\mathbf p_2,\Omega)\big(f'_1f'_2-f_1f_2\big),\qquad B\ge0.$$

Доданок виходу враховує пари, які залишають стан з імпульсом $\mathbf p_1$; доданок приходу — обернені зіткнення, які його заселяють. Завдяки мікроскопічній оборотності те саме зіткнювальне перетворення можна використовувати для обох напрямів. Ядро містить відносну швидкість і диференціальний переріз розсіяння: $B=|\mathbf v_1-\mathbf v_2|\,d\sigma/d\Omega$ за звичайної параметризації кутом розсіяння. За параметризації нормаллю контакту кутовий множник змінюється; додатні геометричні коефіцієнти послідовно включаємо до $B$.

Для доведення потрібні дві симетрії: перестановка однакових частинок та інваріантність міри зіткнень щодо обміну вхідних і вихідних станів. Введемо

$$d\Gamma=d^3p_1\,d^3p_2\,d\Omega\,B,\qquad X=f_1f_2,\qquad Y=f'_1f'_2.$$

Міра невід'ємна. Доведення спирається на її симетрію, а не на конкретний множник для твердих куль. Інтеграл зіткнень нелінійний, бо зіткнення відбуваються між парами: їхня частота пропорційна добутку розподілів.


## Симетризація та інваріанти зіткнень

Для пробної функції $\varphi(\mathbf p)$ розглянемо швидкість зміни її середнього через зіткнення. Перестановка частинок $1\leftrightarrow2$ не змінює інтеграл, тому усереднення дає

<a id="boltzmann-symmetry-particle-exchange"></a>

$$I_\varphi=\int \varphi_1 C[f]_1\,d^3p_1
=\frac12\int d\Gamma\,(\varphi_1+\varphi_2)(Y-X).$$

Тепер поміняємо місцями змінні до й після зіткнення. Міра не змінюється, а $Y-X$ змінює знак:

<a id="boltzmann-symmetry-collision-reversal"></a>

$$I_\varphi=-\frac12\int d\Gamma\,(\varphi'_1+\varphi'_2)(Y-X).$$

Усереднення цих двох форм дає ключову тотожність:

<a id="boltzmann-symmetric-weak-identity"></a>

$$\boxed{I_\varphi=\frac14\int d\Gamma\,
(\varphi_1+\varphi_2-\varphi'_1-\varphi'_2)(Y-X).}$$

Множник $1/4$ виникає внаслідок двох симетризацій. Якщо сума значень $\varphi$ для пари зберігається в кожному дозволеному зіткненні, дужка дорівнює нулю. Звідси

$$\int C[f]\,d^3p=0,\qquad
\int \mathbf p\,C[f]\,d^3p=\mathbf0,\qquad
\int\frac{p^2}{2m}C[f]\,d^3p=0.$$

Це закони збереження числа частинок, імпульсу й кінетичної енергії під час зіткнень. За звичайних умов регулярності та невиродженості пружного розсіяння у трьох вимірах загальний інваріант зіткнень має вигляд
<!--  -->
$$\varphi(\mathbf p)=a+\mathbf b\cdot\mathbf p+c\,p^2.$$

## Покрокове доведення H-теореми

Спочатку розглянемо просторово однорідний газ без зовнішньої сили: $\partial_t f=C[f]$. Визначимо H-функціонал на одиницю об'єму:

$$h[f]=\int f\ln\frac{f}{f_{\rm ref}}\,d^3p,$$

де $f_{\rm ref}>0$ — стала з тією самою розмірністю, що й $f$. Логарифм стає безрозмірним; зміна цієї сталої лише додає величину, пропорційну збережуваній густині числа частинок. Густина кінетичної ентропії $s=-k_Bh$ з точністю до такої сталої.

Припускаємо існування потрібних інтегралів і похідних та спочатку беремо $f>0$. Оскільки $\int C[f]d^3p=0$,

$$\frac{dh}{dt}=\int\left(1+\ln\frac{f}{f_{\rm ref}}\right)C[f]d^3p
=\int\ln\frac{f}{f_{\rm ref}}\,C[f]d^3p.$$

У симетризованій тотожності покладемо $\varphi=\ln(f/f_{\rm ref})$ у даний момент часу. Різниця парних сум дорівнює $\ln(X/Y)$, отже

$$\frac{dh}{dt}=\frac14\int d\Gamma\,(Y-X)\ln\frac XY
=-\frac14\int d\Gamma\,(X-Y)(\ln X-\ln Y).$$

Логарифм — зростаюча функція. За $X>Y$ обидві різниці в останньому добутку додатні; за $X<Y$ обидві від'ємні. Тому

$$\boxed{(X-Y)\ln(X/Y)\ge0,\qquad
\frac{dh}{dt}\le0,\qquad
\sigma=\frac{ds}{dt}=\frac{k_B}{4}\int d\Gamma\,(X-Y)\ln\frac XY\ge0.}$$

Це класична H-теорема Больцмана. Нулі розподілу опрацьовують граничним переходом, використовуючи $f\ln f\to0$ за $f\downarrow0$, за умови скінченності виробництва ентропії. Знак встановлюється для кожного каналу зіткнень, а не підбирається за графіком ентропії.

**Перевірка знака.** Домовленість «H спадає, ентропія зростає» допомагає уникнути типової помилки.


## Умова рівності та розподіл Максвелла

Нульове виробництво ентропії вимагає $X=Y$ майже для кожного дозволеного зіткнення з ненульовим ядром. Отже,

$$f_1f_2=f'_1f'_2
\quad\Longleftrightarrow\quad
\ln f_1+\ln f_2=\ln f'_1+\ln f'_2.$$

Логарифм є інваріантом зіткнень. За наведених вище припущень

$$\ln f=a+\mathbf b\cdot\mathbf p+c p^2,\qquad c<0.$$

Нормованість вимагає від'ємного квадратичного доданка. Виділяючи повний квадрат і зіставляючи збережувані моменти, отримуємо

$$\boxed{f_M(\mathbf r,\mathbf p,t)=
\frac{n(\mathbf r,t)}{[2\pi m k_BT(\mathbf r,t)]^{3/2}}
\exp\left[-\frac{|\mathbf p-m\mathbf u(\mathbf r,t)|^2}{2m k_BT(\mathbf r,t)}\right].}$$

Перевірте три моменти: $\int f_Md^3p=n$, $\int\mathbf p f_Md^3p=nm\mathbf u$, а енергія хаотичного руху дорівнює $3nk_BT/2$. Зокрема, у змінних імпульсу знаменник показника експоненти дорівнює $2mk_BT$, а у змінних швидкості — $2k_BT/m$.

**Локальний розподіл Максвелла** задовольняє $C[f_M]=0$ у кожній точці, але просторово змінні $n,\mathbf u,T$ загалом дають ненульовий доданок перенесення. Тому це не обов'язково розв'язок повного рівняння Больцмана або стан глобальної рівноваги. В однорідному газі без зовнішньої сили моменти визначають стаціонарний розподіл Максвелла; за наявності зовнішнього потенціалу рівновага може бути просторово неоднорідною.

Для однорідного $f$ із такими самими числом частинок, імпульсом та енергією, як у $f_M$, величина $\ln(f_M/f_{\rm ref})$ є комбінацією інваріантів, тому

$$h[f]-h[f_M]=\int f\ln\frac{f}{f_M}\,d^3p
=nD_{\rm KL}(f/n\,\|\,f_M/n)\ge0.$$

Отже, розподіл Максвелла мінімізує $h$ за цих обмежень. H-теорема встановлює монотонність і характеризує нульове виробництво ентропії; сама по собі вона не задає універсальної швидкості збіжності та не є повною теоремою про збіжність для довільних початкових умов.


### Від газової кінетики до механіки суцільного середовища: метод моментів

Розгорнемо тепер матеріал `sec_gaskinetic_ua.tex` про рівняння переносу. Розглядаємо тривимірний розріджений одноатомний газ із пружними зіткненнями, масою частинки $m$ та незалежною від імпульсу зовнішньою силою $\mathbf F$ на одну частинку. Використовуємо вже означену числову густину розподілу $f$, тому додаткового множника $n$ в її моментах немає. Припускаємо достатньо швидке спадання за $|\mathbf p|\to\infty$ та скінченність потрібних моментів. За повторюваними декартовими індексами підсумовуємо; $\partial_j=\partial/\partial r_j$, $D_t=\partial_t+\mathbf u\cdot\nabla$.

Помножимо рівняння Больцмана на довільну гладку функцію $\psi(\mathbf r,\mathbf p,t)$ та проінтегруємо. Диференціювання добутків та інтегрування частинами дають

$$\partial_t\int\psi f\,d^3p+\partial_j\int v_j\psi f\,d^3p
=\int f(\partial_t\psi+v_j\partial_j\psi+F_j\partial_{p_j}\psi)\,d^3p
+\int\psi C[f]\,d^3p.$$

Зокрема, $\int\psi F_j\partial_{p_j}f\,d^3p=-\int f F_j\partial_{p_j}\psi\,d^3p$. Звідси походить знак силового доданка. Для $\psi=m,p_i,p^2/(2m)$ інтеграл зіткнень не дає внеску завдяки доведеним вище інваріантам зіткнень.

| Вага $\psi$ | Густина $\int\psi f\,d^3p$ | Потік у напрямку $j$ | Зовнішнє джерело |
|---|---|---|---|
| $m$ | $\rho=mn$ | $\rho u_j$ | $0$ |
| $p_i$ | $\rho u_i$ | $\rho u_i u_j+\Pi_{ij}$ | $nF_i$ |
| $p^2/(2m)$ | $\mathcal E$ | $\mathcal E u_j+\Pi_{ij}u_i+q_j$ | $n\mathbf F\cdot\mathbf u$ |

Таблиця показує перехід від розподілу у фазовому просторі до тривимірних полів суцільного середовища. Баланси є точними наслідками **моделі Больцмана**; визначальні співвідношення для їх замикання ще потрібно знайти.


### Маса та імпульс: як виникають тиск і в’язкі напруги

Відокремимо середній рух від теплового: $\mathbf v=\mathbf u+\mathbf c$, де $\mathbf c$ — швидкість відносно локального потоку. За означенням $\int\mathbf c f\,d^3p=0$. Тому

$$\int m v_i v_j f\,d^3p=\rho u_i u_j+\Pi_{ij},\qquad
\Pi_{ij}=m\int c_i c_j f\,d^3p.$$

Перехресні доданки зникають, але квадратичний внесок теплового руху залишається. Отже, хаотичний рух молекул переносить імпульс навіть за нульової середньої швидкості. Розкладемо цей потік імпульсу:

$$\Pi_{ij}=P\delta_{ij}+\pi_{ij},\quad
P=\tfrac13\Pi_{ii}=nk_BT,\quad \pi_{ii}=0.$$

Скалярний тиск позначаємо $P$, щоб відрізняти його від імпульсу частинки $\mathbf p$. У цій одноатомній кінетичній моделі другий момент визначає $T$ і поза локальною рівновагою. Балансові рівняння набувають вигляду

$$\boxed{\partial_t\rho+\nabla\cdot(\rho\mathbf u)=0,}$$
$$\partial_t(\rho u_i)+\partial_j(\rho u_i u_j)=-\partial_i P-\partial_j\pi_{ij}+nF_i.$$

Розкриваючи похідні в лівій частині другого рівняння, отримуємо $\rho D_tu_i+u_i[\partial_t\rho+\partial_j(\rho u_j)]$. Вираз у дужках зникає за рівнянням неперервності, тому

$$\boxed{\rho D_tu_i=-\partial_iP-\partial_j\pi_{ij}+nF_i.}$$

**Узгодження знаків.** $\Pi$ — тензор теплового потоку імпульсу. Тензор напруг Коші в механіці суцільного середовища має вигляд $\sigma_{ij}=-P\delta_{ij}+\tau_{ij}$, де $\tau_{ij}=-\pi_{ij}$. Отже, те саме рівняння записується як $\rho D_t\mathbf u=\nabla\cdot\boldsymbol\sigma+n\mathbf F$. Знак мінус пов’язаний з означеннями, а не з від’ємною в’язкістю.


### Енергія: відокремлення руху потоку, внутрішньої енергії та теплового потоку

Означимо внутрішню енергію одиниці маси $e$, повну кінетичну енергію одиниці об’єму $\mathcal E$ та тепловий потік $\mathbf q$:

$$\rho e=\int\tfrac m2c^2f\,d^3p=\tfrac32 nk_BT,\quad
\mathcal E=\tfrac12\rho u^2+\rho e,\quad
q_j=\int\tfrac m2c^2c_jf\,d^3p.$$

Тут $\mathcal E$ не включає енергію зовнішнього потенціалу. Розкриємо $v^2v_j=(u^2+2u_ic_i+c^2)(u_j+c_j)$. Лінійні за $\mathbf c$ доданки зникають після інтегрування; решта дає

$$\int\tfrac m2v^2v_j f\,d^3p
=\mathcal E u_j+\Pi_{ij}u_i+q_j.$$

Ці внески описують конвективне перенесення енергії, механічну роботу та перенесення тепла. Зокрема, $q_j$ уже є потоком енергії, тому додатково множити його на швидкість не потрібно. Енергетичний момент дає

$$\boxed{\partial_t\mathcal E+\nabla\cdot[(\mathcal E+P)\mathbf u
+\boldsymbol\pi\cdot\mathbf u+\mathbf q]=n\mathbf F\cdot\mathbf u.}$$

Щоб отримати рівняння для теплового руху, скалярно помножимо баланс імпульсу на $\mathbf u$ та використаємо неперервність:

$$\partial_t(\tfrac12\rho u^2)+\partial_j(\tfrac12\rho u^2u_j)
=-\partial_j(u_i\Pi_{ij})+\Pi_{ij}\partial_ju_i+nF_iu_i.$$

Віднімемо це рівняння від повного енергетичного балансу. Робота зовнішньої сили скорочується, оскільки вона діє на середній рух, і маємо

$$\rho D_te=-P\nabla\cdot\mathbf u-\pi_{ij}\partial_ju_i-\nabla\cdot\mathbf q.$$

Для одноатомного газу $e=c_vT$, $c_v=3k_B/(2m)$, тому

$$\boxed{\rho c_vD_tT=-P\nabla\cdot\mathbf u
+\tau_{ij}\partial_ju_i-\nabla\cdot\mathbf q.}$$

Стиснення нагріває газ, робота в’язких напруг перетворює кінетичну енергію потоку на внутрішню, а тепловий потік перерозподіляє енергію. Збереження енергії у зіткненнях **не** усуває в’язкість або теплопровідність: зіткнення змінюють розподіл і тим самим визначають потоки.


### Пропущене обчислення другого моменту: `natayg0` → `natayg` → `pressure`

Уведемо $Q_{ijk}=m\int c_ic_jc_kf\,d^3p$ (у джерелі $S_{ijk}$), отже $Q_{iik}=2q_k$, та $\mathcal C_{ij}=m\int c_ic_jC[f]\,d^3p$. Множимо кінетичне рівняння на $mv_iv_j$, **а не на $mv_iv_jf$**. Інтегрування частинами дає

$$\int mv_iv_jF_k\partial_{p_k}f\,d^3p=-n(F_iu_j+F_ju_i),$$
$$\partial_t(\rho u_iu_j+\Pi_{ij})+
\partial_k[\rho u_iu_ju_k+u_k\Pi_{ij}+u_j\Pi_{ik}+u_i\Pi_{jk}+Q_{ijk}]
=n(F_iu_j+F_ju_i)+\mathcal C_{ij}.$$

Потік отримуємо розкриттям усіх восьми доданків $(u_i+c_i)(u_j+c_j)(u_k+c_k)$: три лінійні за $\mathbf c$ доданки зникають, три квадратичні дають $u_i\Pi_{jk}$ та його перестановки, а кубічний — $Q_{ijk}$. Інваріанти зіткнень дозволяють праворуч замінити $v_iv_j$ на $c_ic_j$.

Внесок середнього руху перетворимо без нерозписаної симетризації:

$$\partial_t(\rho u_iu_j)+\partial_k(\rho u_iu_ju_k)
=\rho(u_jD_tu_i+u_iD_tu_j)
=-u_j\partial_k\Pi_{ik}-u_i\partial_k\Pi_{jk}+n(u_jF_i+u_iF_j).$$

Тепер розкриємо $\partial_k(u_j\Pi_{ik})=u_j\partial_k\Pi_{ik}+\Pi_{ik}\partial_ku_j$ та аналогічний вираз із $i\leftrightarrow j$. Два доданки з дивергенцією тензора скорочуються з внесками середнього руху; зовнішні сили також скорочуються. Залишається

$$\boxed{D_t\Pi_{ij}+\Pi_{ij}\partial_ku_k+\Pi_{ik}\partial_ku_j
+\Pi_{jk}\partial_ku_i+\partial_kQ_{ijk}=\mathcal C_{ij}.}$$

Візьмемо слід. Із пружності зіткнень випливає $\mathcal C_{ii}=0$, оскільки $mc^2=mv^2-2m\mathbf u\cdot\mathbf v+mu^2$ у кожній точці є комбінацією інваріантів. За $\Pi_{ii}=3P$

$$3D_tP+3P\nabla\cdot\mathbf u+2(P\delta_{ik}+\pi_{ik})\partial_ku_i+2\partial_kq_k=0,$$
$$\boxed{D_tP+\frac53P\nabla\cdot\mathbf u
+\frac23[\pi_{ik}\partial_ku_i+\nabla\cdot\mathbf q]=0.}$$

Отримали `pressure`; воно узгоджується з рівнянням теплової енергії, бо $\rho e=3P/2$. Рівняння сліду усуває ізотропну частину; решта, безслідна частина рівняння, визначає в’язкі напруги.


### Малі градієнти: виведення `viscnatyag` з рівняння напруг

Означимо безслідний тензор швидкостей деформації

$$S_{ij}=\tfrac12(\partial_iu_j+\partial_ju_i)-\tfrac13\delta_{ij}\nabla\cdot\mathbf u.$$

Позначимо $\delta=\ell/L\ll1$, окремо від параметра розрідженості $nr_0^3$. В одиницях часу зіткнень гідродинамічна похідна має порядок $O(\delta)$, а $\pi/P$ та $Q/(Pv_T)$ — порядок $O(\delta)$. Тому похідні цих нерівноважних моментів мають порядок $O(\delta^2)$. Для визначального співвідношення першого порядку покладаємо $\Pi=P\mathbb I$ та $Q=0$ **лише в частині перенесення** попереднього рівняння напруг, залишаючи праворуч зіткнювальну релаксацію $\pi$:

$$\mathcal C_{ij}^{(1)}=[D_tP+P\nabla\cdot\mathbf u]\delta_{ij}
+P(\partial_iu_j+\partial_ju_i).$$

Рівняння сліду в цьому порядку дає $D_tP=-(5/3)P\nabla\cdot\mathbf u$. Підстановка приводить до

$$\mathcal C_{ij}^{(1)}=P\left(\partial_iu_j+\partial_ju_i-\frac23\delta_{ij}\nabla\cdot\mathbf u\right)=2PS_{ij}.$$

На наступному кроці джерела (`pi1`) вводиться релаксаційний опис безслідного моменту інтеграла зіткнень. Сформулюємо це припущення явно, увівши додатну швидкість релаксації напруг $\nu_\pi$:

$$\mathcal C_{ij}^{(1)}=-\nu_\pi\pi_{ij},\qquad
\pi_{ij}=-\frac{2P}{\nu_\pi}S_{ij},\qquad \eta=\frac{P}{\nu_\pi}>0.$$

Отже, $\pi_{ij}=-\eta(\partial_iu_j+\partial_ju_i-\tfrac23\delta_{ij}\nabla\cdot\mathbf u)$ — саме `viscnatyag` джерела. Воно випливає з рівняння моменту **разом із замиканням зіткнень**, а не лише із симетрії. Скалярна швидкість релаксації є ілюстративним замиканням; для загального ядра Больцмана відгук напруг знаходять розв’язанням лінеаризованого рівняння зіткнень. У цьому порядку ізотропний вираз у дужках не має внеску першого порядку за градієнтами; його дисипативні поправки мають другий порядок, що видно з явного рівняння сліду.


### Момент теплового потоку: заповнення переходів `heat0–heat03`

Спочатку візьмемо використаний у джерелі звичайний момент із вагою $mv^2v_i/2$. Позначимо $J_i=\mathcal E u_i+\Pi_{ij}u_j+q_i$ та $M_{ik}=\int(m/2)v^2v_iv_kf\,d^3p$. Диференціювання ваги за імпульсом дає точне рівняння

$$\partial_tJ_i+\partial_kM_{ik}
=\frac{\mathcal E}{m}F_i+\frac{F_k}{m}(\rho u_iu_k+\Pi_{ik})
+\mathcal C_i^q+u_j\mathcal C_{ij},\qquad
\mathcal C_i^q=\int\tfrac m2 c^2c_iC[f]\,d^3p.$$

Тотожність для зіткнень випливає з розкладу $v^2v_i$ та занулення збережуваних нульового, першого й енергетичного моментів. Це пояснює внесок напруг, пропорційний $\pi_{ij}u_j$, у `heat2`.

Щоб явно виконати віднімання внесків середнього руху, не приховуючи похідних, застосуємо еквівалентну центральну вагу $\psi_i=mc^2c_i/2$ у загальній тотожності для моментів. Уведемо

$$R_{ik}=\int\tfrac m2c^2c_ic_kf\,d^3p,\qquad
A_j=D_tu_j-F_j/m.$$

Тут $\mathscr D=\partial_t+\mathbf v\cdot\nabla_{\mathbf r}+\mathbf F\cdot\nabla_{\mathbf p}$. Вага залежить від $\mathbf r,t$ через $\mathbf u$. Її похідна перенесення дорівнює

$$\mathscr D c_j=-A_j-c_k\partial_ku_j,\qquad
\mathscr D\psi_i=-\frac m2(c^2\delta_{ij}+2c_ic_j)(A_j+c_k\partial_ku_j).$$

Інтегруємо кожен доданок окремо. Коефіцієнт при $A_j$ дорівнює $3P\delta_{ij}/2+\Pi_{ij}$; коефіцієнт при $\partial_ku_j$ — $q_k\delta_{ij}+Q_{ijk}$. Крім того, $\int v_k\psi_i f=u_kq_i+R_{ik}$. Отже, точне рівняння центрального моменту має вигляд

$$D_tq_i+q_i\partial_ku_k+\partial_kR_{ik}
=-\left(\frac32P\delta_{ij}+\Pi_{ij}\right)A_j
-(q_k\delta_{ij}+Q_{ijk})\partial_ku_j+\mathcal C_i^q.$$

Так виконується те саме усунення енергії середнього руху, що й у `denskin02–heat03`. Тепер залишимо лише доданки перенесення першого порядку за градієнтами. Покладемо $\Pi_{ij}=P\delta_{ij}$, використаємо рівняння Ейлера $A_i=-\partial_iP/\rho$ та четвертий момент розподілу Максвелла, де $\theta=k_BT/m=P/\rho$:

$$R_{ik}^{(0)}=\frac m2 n\langle c^2c_ic_k\rangle_M
=\frac52P\theta\delta_{ik}.$$

Оскільки $q,Q$ мають перший порядок, їхні похідні та добутки на градієнти швидкості дають лише другий порядок. Тому рівняння теплового потоку зводиться до

$$\frac52\partial_i(P\theta)=\frac52\frac P\rho\partial_iP+\mathcal C_i^{q(1)},$$
$$\mathcal C_i^{q(1)}=\frac52[\theta\partial_iP+P\partial_i\theta-\theta\partial_iP]
=\frac52\frac{k_B}{m}P\partial_iT.$$

Доданки з градієнтом тиску явно скорочуються. Уведемо додатну швидкість релаксації теплового потоку: $\mathcal C_i^{q(1)}=-\nu_q q_i$. Тоді

$$\boxed{q_i=-\lambda\partial_iT,\qquad \lambda=\frac52\frac{k_B}{m}\frac P{\nu_q}>0.}$$

Це закон Фур’є. Розрізнення $\nu_q$ та $\nu_\pi$ запобігає ототожненню релаксації теплового потоку й напруг; наведена далі модель БГК відповідає окремому вибору $\nu_q=\nu_\pi=1/\tau_R$. У `heat2` джерела внесок теплового потоку має бути від’ємним релаксаційним доданком правильної розмірності. Якщо $\lambda$ — коефіцієнт теплопровідності, цей доданок дорівнює $-\tfrac52(k_B/m)(P/\lambda)q_i$, а не $+(P/\lambda)q_i$. Отже, відтворення методу моментів джерела потребує цього виправлення. Його заключне посилання на закон Фіка для тепла слід читати як закон Фур’є.


### Замикання балансів: рівняння Ейлера та Нав’є–Стокса–Фур’є

П’ять балансових рівнянь визначають еволюцію $n$, трьох компонент $\mathbf u$ та $T$, але містять додаткові моменти $\pi_{ij}$ і $q_i$. Рівняння для цих моментів містять ще вищі моменти. **Обчислення моментів саме по собі не замикає систему.**

Для полів, що повільно змінюються на масштабах довжини вільного пробігу $\ell$ та часу зіткнень $\tau_{\rm coll}$, вводимо $\mathrm{Kn}=\ell/L\ll1$ і також вимагаємо повільної зміни в макроскопічному часі. Поза кінетичними приграничними шарами та областями різких градієнтів шукаємо $f=f_M+f^{(1)}+\cdots$, де $f^{(1)}$ має перший порядок за числом Кнудсена в розкладі моментів. Локальний розподіл Максвелла вибираємо з тими самими збережуваними моментами:

$$\int (1,\mathbf p,p^2/(2m))f^{(1)}\,d^3p=0.$$

Його ізотропність та симетрія непарних моментів дають $\Pi_{ij}[f_M]=P\delta_{ij}$ і $\mathbf q[f_M]=0$. У головному порядку баланси замикаються як **рівняння Ейлера для стисливого газу**:

$$D_t\rho=-\rho\nabla\cdot\mathbf u,\qquad
\rho D_t\mathbf u=-\nabla P+n\mathbf F,\qquad
\rho c_vD_tT=-P\nabla\cdot\mathbf u,\qquad P=\rho k_BT/m.$$

Хоча $C[f_M]=0$, змінний $f_M(\mathbf r,\mathbf p,t)$ загалом має ненульову похідну перенесення. Отже, замикання Ейлера є наближенням головного порядку (локальний розподіл Максвелла не є точним розв’яком кінетичного рівняння).

У першому порядку лінеаризоване рівняння зіткнень для $f^{(1)}$ дає для ізотропного одноатомного розрідженого газу

$$\tau_{ij}=-\pi_{ij}=2\eta S_{ij},\qquad
S_{ij}=\tfrac12(\partial_ju_i+\partial_iu_j)-\tfrac13\delta_{ij}\nabla\cdot\mathbf u,
\qquad \mathbf q=-\lambda\nabla T.$$

Тут $\eta>0$ — динамічна зсувна в’язкість, $\lambda>0$ — коефіцієнт теплопровідності. У цій моделі внесок об’ємної в’язкості відсутній. Значення коефіцієнтів визначаються моделлю зіткнень, а не лише симетрією. Підстановка дає **систему Нав’є–Стокса–Фур’є для стисливого газу**:

$$D_t\rho=-\rho\nabla\cdot\mathbf u,$$
$$\rho D_tu_i=-\partial_iP+\partial_j(2\eta S_{ij})+nF_i,$$
$$\rho c_vD_tT=-P\nabla\cdot\mathbf u+2\eta S_{ij}S_{ij}
+\nabla\cdot(\lambda\nabla T).$$

Якщо $\eta$ та $\lambda$ змінні, їх слід залишати під похідними. За сталого $\eta$ в’язка сила дорівнює $\eta[\Delta\mathbf u+\tfrac13\nabla(\nabla\cdot\mathbf u)]$. Нестисливість потребує окремого граничного переходу й не випливає з обчислення моментів. Так само однокомпонентний газ не має дифузійного потоку маси відносно барицентричної швидкості; для виведення закону Фіка потрібен опис додаткового компонента або домішки.


### Приклад замикання: обчислення коефіцієнтів у моделі БГК

Щоб отримати не лише тензорну форму, замінимо інтеграл зіткнень ілюстративною тривимірною моделлю БГК: $C[f]=-(f-f_M)/\tau_R$, де $f_M$ має ті самі локальні масу, імпульс та енергію, що й $f$. Така заміна зберігає п’ять інваріантів зіткнень. Вона не є інтегралом зіткнень твердих куль.

Позначимо $\mathscr D=\partial_t+\mathbf v\cdot\nabla+\mathbf F\cdot\nabla_{\mathbf p}$. У першому порядку $f^{(1)}=-\tau_R\mathscr D f_M$, причому часові похідні полів обчислюємо за рівняннями Ейлера. Нехай $\theta=k_BT/m$. Диференціювання гауссіана з використанням $D_t\ln n=-\nabla\cdot\mathbf u$, $D_t\ln T=-\tfrac23\nabla\cdot\mathbf u$ та $D_t\mathbf u=\mathbf F/m-\nabla P/\rho$ дає

$$\frac{\mathscr D f_M}{f_M}
=\frac{c_ic_j}{\theta}S_{ij}
+c_i\left(\frac{c^2}{2\theta}-\frac52\right)\partial_i\ln T.$$

Після підстановки рівнянь Ейлера силові доданки та доданки з градієнтом густини скорочуються. Коефіцієнт $5/2$ забезпечує нульовий сумарний імпульс поправки, що переносить тепло. Для $\langle A\rangle_M=n^{-1}\int A f_M\,d^3p$ потрібні такі гауссові моменти:

$$\langle c_ic_jc_kc_l\rangle_M=\theta^2(\delta_{ij}\delta_{kl}+\delta_{ik}\delta_{jl}+\delta_{il}\delta_{jk}),$$
$$\langle c^2c_ic_j\rangle_M=5\theta^2\delta_{ij},\qquad
\langle c^4c_ic_j\rangle_M=35\theta^3\delta_{ij}.$$

Підставимо $f^{(1)}$ в означення напруг і теплового потоку. Непарні гауссові моменти зникають, а $S_{ii}=0$:

$$\pi_{ij}=m\int(c_ic_j-\tfrac13c^2\delta_{ij})f^{(1)}\,d^3p
=-2P\tau_R S_{ij},$$
$$q_i=-\frac{mn\tau_R}{2}\left(\frac{35}{2}\theta^2-\frac{25}{2}\theta^2\right)\partial_i\ln T
=-\frac52\frac{k_B}{m}P\tau_R\partial_iT.$$

Отже, $\eta=P\tau_R$ і $\lambda=\tfrac52(k_B/m)P\tau_R$. Обчислення показує, як кінетичне замикання визначає коефіцієнти переносу. Числове співвідношення між ними специфічне для БГК; повний оператор Больцмана потребує розв’язання лінеаризованої задачі зіткнень. Ця поправка першого порядку є асимптотичним обчисленням моментів і не обов’язково задає додатний розподіл у далеких швидкісних хвостах.


### Баланс ентропії суцільного середовища та його зв’язок з H-теоремою

Нехай $s$ — локально-рівноважна ентропія **одиниці маси**, тому густина ентропії дорівнює $\rho s$. У співвідношенні Ґіббса використовуємо внутрішню енергію, а не енергію середнього руху:

$$T\,ds=de+P\,d(1/\rho).$$

З неперервності маємо $D_t(1/\rho)=(\nabla\cdot\mathbf u)/\rho$. Поєднання з рівнянням внутрішньої енергії усуває оборотний внесок стиснення:

$$\rho T D_ts=\tau_{ij}\partial_ju_i-\nabla\cdot\mathbf q.$$

Використаємо $\nabla\cdot(\mathbf q/T)=(\nabla\cdot\mathbf q)/T-\mathbf q\cdot\nabla T/T^2$ та підставимо визначальні співвідношення. У наближенні Нав’є–Стокса–Фур’є

$$\boxed{\partial_t(\rho s)+\nabla\cdot\left(\rho s\mathbf u+\frac{\mathbf q}{T}\right)
=\frac{2\eta}{T}S_{ij}S_{ij}+\frac{\lambda}{T^2}|\nabla T|^2\ge0.}$$

В’язкий внесок можна також записати як $\pi_{ij}\pi_{ij}/(2\eta T)$; множник $1/T$ необхідний. Це макроскопічний прояв виробництва ентропії внаслідок зіткнень поблизу локальної рівноваги. Рівняння також визначає потік ентропії через межу. Повна ентропія не спадає лише за відсутності граничного обміну ентропією або за його належного врахування в балансі. Наступний розділ містить загальніший кінетичний баланс ентропії.

**Перевірки для обговорення.** Для локального простого зсуву $u_x=ay$, $u_y=u_z=0$ отримайте $S_{xy}=a/2$, $\tau_{xy}=\eta a$ та виробництво ентропії $\eta a^2/T$ за однорідної температури. Щоб зсув за однорідної температури залишався стаціонарним, потрібне відведення тепла. За стаціонарної теплопровідності умова $\nabla\cdot\mathbf q=0$ сумісна з додатним виробництвом ентропії $\lambda|\nabla T|^2/T^2$: ентропія відводиться через межі. Нарешті, поясніть, чому з $\int p_iC[f]\,d^3p=0$ не випливає $\eta=0$.

**Відповідність джерелу.** Доповнення спирається на `microur1`, `neprer`, `tensnapryag0–1`, `contin_denseq`, `mombalance`, `denskin1–3`, `subsec_navie`, `viscnatyag` та підрозділ про баланс ентропії у `sec_gaskinetic_ua.tex`. Послідовно використовуємо нормований на число частинок розподіл $f$, тепловий потік $\mathbf q$ (у джерелі $\mathbf S$), тиск $P$ та величини $e,s$ на одиницю маси. Явне обчислення в моделі БГК ілюструє замикання й не замінює повної задачі зіткнень із вихідного матеріалу.


## Локальне виробництво, граничний потік і відкрита система

Для неоднорідного газу визначимо

$$s(\mathbf r,t)=-k_B\int f\ln(f/f_{\rm ref})\,d^3p,
\qquad \mathbf J_s=-k_B\int\mathbf v f\ln(f/f_{\rm ref})\,d^3p.$$

Домножуючи кінетичне рівняння на $-k_B[1+\ln(f/f_{\rm ref})]$ та інтегруючи за імпульсами, дістаємо

$$\boxed{\partial_t s+\nabla\cdot\mathbf J_s=\sigma\ge0.}$$

Силовий доданок зникає після інтегрування, якщо сила не залежить від імпульсу, а розподіл досить швидко спадає за великих імпульсів. Інтегрування за просторовим об'ємом із зовнішньою нормаллю $\mathbf n$ дає

$$\frac{dS}{dt}=\int_V\sigma\,d^3r-\oint_{\partial V}\mathbf J_s\cdot\mathbf n\,dA.$$

За періодичних граничних умов, належного спадання на нескінченності або нерухомих теплоізолювальних стінок із дзеркальним відбиттям сумарний потік ентропії відсутній, тому $S$ не спадає. Самої замкненості щодо обміну частинками недостатньо, якщо стінки обмінюються теплом. Термостатовані стінки, резервуари або зовнішнє збудження з дисипацією потребують урахування відповідного обміну ентропією.

Тут виникає зв'язок із синергетикою. Для відкритої системи з підведенням енергії

$$\dot S_{\rm system}=\dot S_{\rm production}+\dot S_{\rm exchange},
\qquad\dot S_{\rm production}\ge0.$$

Нерівноважний стаціонарний стан може відводити назовні ентропію, яку виробляє. Тому когерентне поле лазера не порушує другого закону: накачування й втрати пов'язують лазер з оточенням. H-функціонал класичного газу не стає автоматично функцією Ляпунова для лазера чи довільної моделі утворення структур. Так само амплітуду параметра порядку не слід ототожнювати з термодинамічною ентропією.


## Де виникає стріла часу

Точна нормована $N$-частинкова фазова густина $\rho_N$ підпорядковується рівнянню Ліувілля. Гамільтонівський фазовий потік нестисливий, тому за належних граничних умов тонкозерниста ентропія Ґіббса зберігається:

$$S_G=-k_B\int\rho_N\ln(\rho_N/\rho_{\rm ref})\,d\Gamma_N,
\qquad \frac{dS_G}{dt}=0.$$

Це не суперечить зростанню одночастинкової кінетичної ентропії. Редукований опис не зберігає всіх кореляцій. Замикання інтеграла зіткнень відбирає слабко корельовані вхідні пари, тоді як зіткнення породжують вихідні кореляції. Обернення всіх імпульсів мікроскопічної траєкторії перетворює ці спеціальні вихідні кореляції на вхідні; обернений стан загалом уже не задовольняє того самого припущення молекулярного хаосу. Так знімається позірна суперечність у запереченні Лошмідта про оборотність на рівні кінетичного наближення.

Повернення є ще одним застереженням: скінченна гамільтонівська система в обмеженій доступній області фазового простору за умов теореми про повернення може знову опинитися як завгодно близько до попереднього стану. Строго спадний кінетичний H-функціонал для всіх мікроскопічних станів і всіх часів був би з цим несумісний. Рівняння Больцмана має область застосовності та кінетичний граничний опис; воно не є точним переписуванням кожної траєкторії скінченної системи.

Розрізняйте три рівні: оборотну мікроскопічну динаміку, замкнене наближене рівняння для $f$ та макроскопічний баланс ентропії. H-теорема є строгим наслідком кінетичного рівняння за умов існування відповідних інтегралів; фізична застосовність самого рівняння — окреме питання.

## Канал зіткнень, який можна дослідити повністю

Розглянемо чотири безрозмірні швидкості у двох вимірах:

$$\mathbf v_1=(1,0),\quad\mathbf v_2=(-1,0),\quad
\mathbf v_3=(0,1),\quad\mathbf v_4=(0,-1),\qquad 1+2\rightleftharpoons3+4.$$

Цей канал зберігає імпульс та енергію: кожна протилежна пара має нульовий сумарний імпульс і однакову енергію. Нехай $z_i>0$ — населеності, нормовані умовою $\sum_i z_i=1$. За сталої швидкості $k=1$

$$q=z_1z_2-z_3z_4,\qquad
\dot{\mathbf z}=(-q,-q,q,q),\qquad
H_d=\sum_i z_i\ln z_i.$$

Безпосереднє диференціювання дає

<a id="boltzmann-channel-entropy-rate"></a>

$$\dot H_d=-\big(z_1z_2-z_3z_4\big)
\ln\frac{z_1z_2}{z_3z_4}\le0.$$

Множника $1/4$ тут немає, оскільки один оборотний канал враховано один раз, а не підсумовано його чотири еквівалентні позначення. Для $\mathbf z(0)=(0.4,0.4,0.1,0.1)$ із симетрії випливає $z_1=z_2=a$, $z_3=z_4=1/2-a$, тому

$$\dot a=-(a-1/4),\qquad a(t)=1/4+0.15e^{-t}.$$

Наступна комірка використовує цей точний розв'язок і перевіряє інваріанти та ентропійну тотожність. Це обмежена дискретна модель зіткнень, а не моделювання неперервного газу: її рівновага за заданих початкових інваріантів — $(1/4,1/4,1/4,1/4)$, а не неперервний розподіл Максвелла.


```python
import numpy as np
import matplotlib.pyplot as plt

ht_lang = "uk"
def ht_text(en, ua):
    return ua if ht_lang == "uk" else en

ht_t = np.linspace(0, 8, 801)
ht_a = .25 + .15*np.exp(-ht_t)
ht_z = np.column_stack([ht_a, ht_a, .5-ht_a, .5-ht_a])
ht_v = np.array([[1.,0.],[-1.,0.],[0.,1.],[0.,-1.]])
ht_q = ht_z[:,0]*ht_z[:,1] - ht_z[:,2]*ht_z[:,3]
ht_dz = np.column_stack([-ht_q,-ht_q,ht_q,ht_q])
ht_H = np.sum(ht_z*np.log(ht_z),axis=1)
ht_H_eq = -np.log(4.)
ht_prod = ht_q*np.log((ht_z[:,0]*ht_z[:,1])/(ht_z[:,2]*ht_z[:,3]))
ht_Hdot = np.sum((1+np.log(ht_z))*ht_dz,axis=1)
ht_number = ht_z.sum(axis=1)
ht_momentum = ht_z@ht_v
ht_energy = ht_z@(np.sum(ht_v**2,axis=1)/2)
assert np.all(ht_z>0)
assert np.allclose(ht_number,1.,atol=1e-14)
assert np.max(np.abs(ht_momentum))<1e-14
assert np.allclose(ht_energy,.5,atol=1e-14)
assert np.max(np.diff(ht_H))<=1e-14
assert np.allclose(ht_Hdot,-ht_prod,atol=1e-14)

ht_fig, ht_axes = plt.subplots(2,2,figsize=(11,8),layout="constrained")
ht_ax=ht_axes[0,0]
for ht_i,ht_vec in enumerate(ht_v):
    ht_color="#0072B2" if ht_i<2 else "#D55E00"
    ht_ax.annotate("",xy=ht_vec,xytext=(0,0),arrowprops=dict(arrowstyle="->",lw=2,color=ht_color))
    ht_ax.text(*(1.15*ht_vec),str(ht_i+1),ha="center",va="center",fontsize=13,color=ht_color)
ht_ax.set(xlim=(-1.4,1.4),ylim=(-1.4,1.4),xlabel="$v_x$",ylabel="$v_y$",
          title=ht_text("One channel: 1 + 2 ↔ 3 + 4","Один канал: 1 + 2 ↔ 3 + 4"))
ht_ax.set_aspect("equal"); ht_ax.grid(alpha=.15)
ht_axes[0,1].plot(ht_t,ht_a,label="$z_1=z_2$",color="#0072B2")
ht_axes[0,1].plot(ht_t,.5-ht_a,label="$z_3=z_4$",color="#D55E00")
ht_axes[0,1].axhline(.25,color="black",ls="--",lw=1)
ht_axes[0,1].set(xlabel="$t$",ylabel="$z_i$",title=ht_text("Redistribution at fixed invariants","Перерозподіл за сталих інваріантів"))
ht_axes[0,1].legend()
ht_axes[1,0].plot(ht_t,ht_H-ht_H_eq,color="#0072B2")
ht_axes[1,0].set(xlabel="$t$",ylabel="$H_d-H_{d,eq}$",title=ht_text("H decreases toward its minimum","H спадає до мінімуму"))
ht_axes[1,1].plot(ht_t,ht_prod,color="#009E73")
ht_axes[1,1].set(xlabel="$t$",ylabel=r"$-\dot H_d$",title=ht_text("Nonnegative entropy production","Невід'ємне виробництво ентропії"))
for ht_ax in ht_axes.flat: ht_ax.grid(alpha=.15)
plt.show()
print(ht_text("Maximum number / momentum / energy drift:","Максимальний дрейф числа / імпульсу / енергії:"),
      np.max(abs(ht_number-1)),np.max(abs(ht_momentum)),np.max(abs(ht_energy-.5)))
print(ht_text("Maximum H-rate identity error:","Максимальна похибка тотожності для швидкості H:"),np.max(abs(ht_Hdot+ht_prod)))
```

## 10. Наочна релаксація до гауссового розподілу

Щоб показати зміну розподілу без реалізації повного інтеграла зіткнень, використаємо однорідну **релаксаційну модель БГК**: $\partial_t f=\nu(f_M-f)$, де $f_M$ має ті самі збережувані моменти. В однорідній задачі ці моменти сталі, тому

$$f(t)=f_M+e^{-\nu t}(f_0-f_M).$$

Це додатна опукла комбінація. Узгодження моментів та інваріантна форма $\ln f_M$ дають власну ентропійну нерівність моделі:

$$\dot h=-\nu\int(f-f_M)\ln(f/f_M)\,d^3p\le0.$$

Вона має ту саму знакову структуру, що й H-теорема, але не відтворює повної динаміки Больцмана. Зокрема, швидкість релаксації задається, а не обчислюється за ядром розсіяння.

На рисунку використовуємо нормований маргінальний розподіл однієї компоненти швидкості $g(v)$ у безрозмірних одиницях із початковими двома максимумами:

$$g_0(v)=\tfrac12\mathcal N(v;-u_b,s_0^2)+\tfrac12\mathcal N(v;u_b,s_0^2),
\quad u_b=2,\ s_0=0.6,\quad
M(v)=\mathcal N(v;0,u_b^2+s_0^2).$$

Це маргінальна ілюстрація БГК, а не газ однакових частинок лише з одновимірними пружними зіткненнями: останні просто обмінюють швидкості частинок. Для маргінального розподілу узгоджено нормування, середню швидкість і другий момент. Надлишок H дорівнює $\int g\ln(g/M)dv$ — це інформаційна міра для даного маргінального розподілу, а не автоматично повна ентропія тривимірного газу. Скінченність області інтегрування та крок сітки перевіряємо явно.


```python
def ht_gaussian(v,mean,variance):
    return np.exp(-(v-mean)**2/(2*variance))/np.sqrt(2*np.pi*variance)

def ht_integral(y,x,axis=-1):
    # Explicit trapezoidal quadrature, compatible with older NumPy versions.
    y=np.moveaxis(np.asarray(y),axis,-1)
    return np.sum((y[...,1:]+y[...,:-1])*np.diff(x)/2,axis=-1)

def ht_bgk_grid(limit,points,times):
    v=np.linspace(-limit,limit,points)
    g0=.5*ht_gaussian(v,-2,.6**2)+.5*ht_gaussian(v,2,.6**2)
    M=ht_gaussian(v,0,4+.6**2)
    weight=np.exp(-times[:,None])
    g=weight*g0[None,:]+(1-weight)*M[None,:]
    assert np.all(g>0) and np.all(M>0)
    moments=np.column_stack([ht_integral(g,v),ht_integral(g*v,v),ht_integral(g*v*v,v)])
    excess=ht_integral(g*np.log(g/M),v)
    prod=ht_integral((g-M)*np.log(g/M),v)
    H=ht_integral(g*np.log(g),v)
    HM=ht_integral(M*np.log(M),v)
    return v,g0,M,g,moments,excess,prod,H-HM

ht_times=np.linspace(0,8,161)
ht_result=ht_bgk_grid(16,3201,ht_times)
ht_vgrid,ht_g0,ht_M,ht_g,ht_mom,ht_excess,ht_bgk_prod,ht_deltaH=ht_result
ht_target=np.array([1.,0.,4.36])
assert np.max(abs(ht_mom-ht_target))<1e-10
assert np.min(ht_excess)>-1e-12
assert np.max(np.diff(ht_excess))<1e-12
assert np.min(ht_bgk_prod)>-1e-12
assert np.max(abs(ht_excess-ht_deltaH))<1e-10
ht_fine=ht_bgk_grid(18,7201,ht_times)
assert np.max(abs(ht_excess-ht_fine[5]))<1e-9

ht_fig,ht_axes=plt.subplots(1,2,figsize=(12,4.5),layout="constrained")
for ht_time,ht_color in zip([0,.5,2,8],["#D55E00","#CC79A7","#0072B2","#009E73"]):
    ht_index=np.argmin(abs(ht_times-ht_time))
    ht_axes[0].plot(ht_vgrid,ht_g[ht_index],label=fr"$t={ht_time:g}$",color=ht_color,lw=1.7)
ht_axes[0].plot(ht_vgrid,ht_M,"k--",lw=1.2,label="$M(v)$")
ht_axes[0].set(xlim=(-7,7),xlabel="$v$",ylabel="$g(v,t)$",
    title=ht_text("BGK marginal: two peaks relax","Маргінальний розподіл БГК: релаксація"))
ht_axes[0].legend(fontsize=9)
ht_axes[1].semilogy(ht_times,ht_excess,color="#0072B2",label=r"$\int g\ln(g/M)\,dv$")
ht_axes[1].semilogy(ht_times,ht_bgk_prod,color="#009E73",ls="--",label=r"$-\dot H$")
ht_axes[1].set(xlabel="$t$",title=ht_text("Excess H and entropy production","Надлишок H та виробництво ентропії"))
ht_axes[1].legend(fontsize=9)
for ht_ax in ht_axes: ht_ax.grid(alpha=.2)
plt.show()
print(ht_text("Maximum moment errors (mass, mean, second moment):","Максимальні похибки моментів (маса, середнє, другий момент):"),np.max(abs(ht_mom-ht_target),axis=0))
print(ht_text("Grid/domain refinement: maximum excess-H difference:","Уточнення сітки й області: максимальна різниця надлишку H:"),np.max(abs(ht_excess-ht_fine[5])))
```

У виведенні «Симетризація та інваріанти зіткнень» відтворіть два усереднення: [перестановку частинок $1\leftrightarrow2$](#boltzmann-symmetry-particle-exchange), а потім [обмін змінних до та після зіткнення $(\mathbf p_1,\mathbf p_2)\leftrightarrow(\mathbf p'_1,\mathbf p'_2)$](#boltzmann-symmetry-collision-reversal). Покажіть, як два множники $1/2$ дають $1/4$ у [виділеній рамкою тотожності для $I_\varphi$](#boltzmann-symmetric-weak-identity). Порівняйте цей підрахунок із [формулою для $\dot H_d$ в оборотному каналі $1+2\rightleftharpoons3+4$](#boltzmann-channel-entropy-rate), де канал враховано один раз, і поясніть отриманий числовий множник.
2. Перевірте збереження імпульсу й енергії для зіткнення твердих куль. Який крок H-теореми порушився б для довільного правила зіткнення?
3. Виведіть нормування та три моменти $f_M$. Покажіть, чому довільний локальний розподіл Максвелла не зануляє доданок перенесення.
4. Доведіть тотожність із відносною ентропією за фіксованих числа частинок, імпульсу та енергії. Укажіть, де саме використано ці обмеження.
5. Змініть початкові дискретні населеності, залишивши їх додатними. Для несиметричних даних використовуйте $\dot{\mathbf z}=(-q,-q,q,q)$ і знайдіть рівновагу за нових інваріантів; не застосовуйте спеціальний симетричний аналітичний розв'язок.
6. Змініть відстань між пучками та їхню ширину в прикладі БГК, узгоджено змінивши цільову дисперсію. Порівняйте спадання H, перевіряючи збереження моментів і точність квадратур. Чому це не вимірювання справжньої частоти зіткнень у рівнянні Больцмана?
7. Поясніть, що відбувається з вхідними кореляціями за мікроскопічного обернення часу. Розрізніть кінетичну ентропію та тонкозернисту ентропію Ґіббса.
8. Для стаціонарної відкритої системи з додатним виробництвом ентропії визначте знак сумарного обміну ентропією. Поясніть, чому це допускає, але саме по собі не передбачає, когерентне поле лазера або просторову структуру.

## Література та відповідність вихідним матеріалам

Основне джерело — авторський український курс «Нерівноважна термодинаміка і статистична фізика»: `stphysneq_ua.tex`, що містить `sec_gaskinetic_ua.tex`. Ноутбук наслідує розгляд мікроскопічної густини, використання імпульсів вхідних пар та підрозділ про властивості інтеграла зіткнень і H-теорему. Додані розділи про суцільне середовище розгортають баланси переносу та ілюструють їх замикання в моделі БГК.

| Позначки у джерелі | Матеріал ноутбука |
|---|---|
| `micro1`, `normmicro1`, `g2`, `f2_factor` | Мікроскопічна густина, нормування, кореляції, кінетичне замикання |

**Редакційні уточнення в цій адаптації.** До симетризації всі множники інтеграла зіткнень включено до $B$. Добуток $(Y-X)\ln(X/Y)$ недодатний, а не невід'ємний. У знаменнику показника експоненти розподілу Максвелла в імпульсних змінних стоїть $2mk_BT$. Інваріанти зіткнень містять незалежний лінійний доданок за імпульсом. Локальна зміна ентропії включає перенесення; внесок зіткнень не є повною локальною похідною за часом у неоднорідному газі. Оригінальні TeX-файли не змінено.
