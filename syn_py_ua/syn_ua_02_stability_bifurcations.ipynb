## Необхідний вступ: стаціонарні точки, якобіани та власні моди
Розглянемо систему ЗДР:
<!--  -->
$$\dot{\mathbf{x}} = \mathbf{F}(\mathbf{x};\mu)$$
<!--  -->
де $\mathbf{x}$ - вектор стану, $\mu$ набір параметрів. Стаціонарна точка $\mathbf{x}_*$ задовольняє $\mathbf{F}(\mathbf{x}_*;\mu)=0$. Малу відхилу запишемо як $\boldsymbol\xi=\mathbf{x}-\mathbf{x}_*$. Розклад Тейлора дає

$$\dot{\boldsymbol\xi}=J_*\boldsymbol\xi+O(\|\boldsymbol\xi\|^2),\qquad
(J_*)_{ij}=\left.\frac{\partial F_i}{\partial x_j}\right|_{\mathbf{x}_*}.$$

Поблизу стаціонарної точки власна мода поводиться як $e^{\lambda t}$. Від'ємна дійсна частина $\lambda$ означає згасання, додатна — нестійкість, а ненульова уявна частина — коливання. Лінеаризація є локальним наближенням: за нульової дійсної частини вирішальними стають нелінійні члени, а самі власні значення нічого не кажуть про віддалені атрактори чи межі басейнів притягання.


## Класифікація стаціонарних точок на фазовій площині

Перш ніж вивчати зміну рівноваг під час біфуркацій, сформуємо наочне уявлення про їхню локальну поведінку. Для гладкої автономної системи

$$\dot x=f(x,y),\qquad \dot y=g(x,y),$$

стаціонарна точка $(x_*,y_*)$ є перетином **нуль-клін** $f=0$ та $g=0$. На $f=0$ вектор швидкості вертикальний, а на $g=0$ — горизонтальний; виняток становлять точки рівноваги, де обидві компоненти нульові. Нуль-кліна загалом не є траєкторією. Перенесемо рівновагу в початок координат: $\boldsymbol\xi=(x-x_*,y-y_*)^T$. Для двічі неперервно диференційовного векторного поля

$$\dot{\boldsymbol\xi}=A\boldsymbol\xi+O(\|\boldsymbol\xi\|^2),\qquad
A=J_* = \begin{pmatrix}f_x&f_y\\g_x&g_y\end{pmatrix}_{(x_*,y_*)}.$$

У лінійній системі дійсна власна пара задовольняє $A\mathbf v=\lambda\mathbf v$, а розв'язок $\boldsymbol\xi(t)=c e^{\lambda t}\mathbf v$ лежить на інваріантній прямій, заданій власним вектором. Для різних дійсних власних значень загальний розв'язок має вигляд

$$\boldsymbol\xi(t)=c_1e^{\lambda_1t}\mathbf v_1+c_2e^{\lambda_2t}\mathbf v_2.$$

Власні значення визначають зростання або згасання, а власні вектори — відповідні напрями. Для сідла стійкий і нестійкий власні підпростори одновимірні. Для стійкого вузла стійким підпростором є вся площина, хоча на рисунку виділяємо лише дві власні прямі. Для нелінійного сідла відповідні інваріантні криві загалом викривлені й лише дотикаються до цих прямих у точці рівноваги.

Позначимо

$$T=\operatorname{tr}A=a+d,\qquad D=\det A=ad-bc,\qquad Q=T^2-4D.$$

Тут $D$ — визначник, а не інверсія населеностей із модуля 01. Розкриваючи $\det(\lambda I-A)=0$, дістаємо

$$\lambda^2-T\lambda+D=0,\qquad
\lambda_\pm=\frac{T\pm\sqrt Q}{2},\qquad
T=\lambda_++\lambda_-,\quad D=\lambda_+\lambda_-.$$

Отже, класифікацію за власними значеннями можна подати геометрично на площині $(T,D)$. Парабола $D=T^2/4$ відділяє дійсні власні значення від комплексних. У лінійному потоці площа малої області змінюється в $e^{Tt}$ разів. Від'ємний слід означає стискання площі, але сам по собі не гарантує притягання: сідло може стискати площу й водночас розтягувати збурення вздовж одного напряму.

| Умови | Тип лінійної рівноваги | Поведінка за $t\to+\infty$ |
|---|---|---|
| $D<0$ | Сідло | Один напрям зростання й один згасання; нестійкість |
| $D>0,\ Q>0,\ T<0$ | Стійкий вузол | Обидві моди згасають без коливних власних мод |
| $D>0,\ Q>0,\ T>0$ | Нестійкий вузол | Обидві моди зростають |
| $Q<0,\ T<0$ | Стійкий фокус | Обертання зі згасанням до рівноваги |
| $Q<0,\ T>0$ | Нестійкий фокус | Обертання зі зростанням відстані від рівноваги |
| $D>0,\ T=0$ | Лінійний центр | Замкнені орбіти; стійкість без асимптотичної стійкості |
| $Q=0,\ T\ne0$ | Вузол із кратним власним значенням | Зірковий або дефектний вузол; стійкість визначає знак $T$ |
| $D=0$ | Принаймні одне нульове власне значення | Вироджений лінійний випадок; потрібен аналіз матриці й нелінійних членів |

Стійкість за Ляпуновим означає, що достатньо малі збурення залишаються малими для всіх наступних моментів часу. Асимптотична стійкість додатково вимагає збіжності до рівноваги. Центр має першу властивість, але не другу. Для стійкої рівноваги евклідова відстань не обов'язково спадає монотонно кожної миті: неортогональні власні вектори можуть спричиняти тимчасове зростання збурення.


```python
# planar-fixed-points-trace-diagram
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

pf_lang = "uk"
def pf_text(en, ua):
    return ua if pf_lang == "uk" else en

pf_t = np.linspace(-4.5, 4.5, 701)
pf_d = np.linspace(-2.0, 6.0, 601)
pf_T, pf_D = np.meshgrid(pf_t, pf_d)
pf_regions = np.where(pf_D < 0, 0,
    np.where(pf_D < pf_T**2/4, np.where(pf_T < 0, 1, 2),
             np.where(pf_T < 0, 3, 4)))
pf_fig, pf_ax = plt.subplots(figsize=(10, 6.5), layout="constrained")
pf_ax.pcolormesh(pf_T, pf_D, pf_regions,
    cmap=ListedColormap(["#eadcf2", "#b9dfeb", "#f8d6af", "#d3eadc", "#f3c9ce"]),
    shading="auto", rasterized=True)
pf_ax.plot(pf_t, pf_t**2/4, color="#333333", lw=1.8)
pf_ax.axhline(0, color="#333333", lw=1.4)
pf_ax.plot([0, 0], [0, 6], color="#555555", lw=2, ls="--")
for tx, dy, en, ua in [
    (-3.1, .65, "Stable nodes", "Стійкі вузли"),
    (3.1, .65, "Unstable nodes", "Нестійкі вузли"),
    (-2.25, 4.9, "Stable foci", "Стійкі фокуси"),
    (2.25, 4.9, "Unstable foci", "Нестійкі фокуси"),
    (0, -1.15, "Saddles", "Сідла")]:
    pf_ax.text(tx, dy, pf_text(en, ua), ha="center", fontsize=12)
pf_ax.text(.13, 3.2, pf_text("Linear centers", "Лінійні центри"),
           rotation=90, va="center", fontsize=10)
pf_ax.annotate(pf_text("Repeated roots: Q = 0", "Кратні корені: Q = 0"),
    xy=(4.0, 4.0), xytext=(3.1, 5.75), ha="center",
    arrowprops=dict(arrowstyle="->", color="#333333"), fontsize=10)
pf_ax.text(-4.3, .12, pf_text("D = 0: zero eigenvalue", "D = 0: нульове власне значення"), fontsize=9)
pf_ax.set(xlim=(-4.5,4.5), ylim=(-2,6), xlabel=r"$T=\mathrm{tr}\,A$",
    ylabel=r"$D=\det A$", title=pf_text("Trace–determinant map", "Діаграма слід–визначник"))
pf_ax.grid(False)
plt.show()
```

### Як читати атлас фазових портретів

Кожна панель показує систему $\dot{\boldsymbol\xi}=A\boldsymbol\xi$ в однаковому квадраті. Стрілки задають напрям часу, чорна точка позначає ізольовану рівновагу, а в заголовку наведено власні значення. Сині штрихові та помаранчеві пунктирні прямі виділяють стійкі й нестійкі дійсні власні напрями; це не нуль-кліни. На останній панелі є ціла пряма рівноваг, показана чорним. Порівнюйте геометрію та напрям руху, а не густоту ліній: вона не відображає ймовірність або однакові часові інтервали.

Для $\lambda=\alpha\pm i\omega$ у відповідних дійсних координатах маємо

$$\dot u=\alpha u-\omega v,\quad \dot v=\omega u+\alpha v,
\qquad \dot r=\alpha r,\quad\dot\theta=\omega.$$

Отже, $\alpha$ визначає зростання радіуса, а $\omega$ — обертання. За іншого лінійного вибору координат кола перетворюються на еліпси. Слід і визначник не задають напрям обертання чи орієнтацію власних векторів: щоб установити напрям стрілок, обчисліть векторне поле в конкретній точці.

Для кратного ненульового власного значення $\lambda=T/2$ слід розрізняти $A=\lambda I$ (**зірковий вузол**, усі прямі інваріантні) та дефектну матрицю лише з одним незалежним власним вектором. У жорданових координатах для останньої

$$A=\begin{pmatrix}\lambda&1\\0&\lambda\end{pmatrix},\qquad
\boldsymbol\xi(t)=e^{\lambda t}\begin{pmatrix}u_0+t v_0\\v_0\end{pmatrix}.$$

Множник $t$ викривлює траєкторії, але не переважає експоненційного згасання за $\lambda<0$. Назви «неправильний», «вироджений» і «дефектний» вузол у підручниках можуть відрізнятися; число незалежних власних векторів однозначне. Кратне ненульове власне значення не порушує гіперболічності.

За $D=0$ рівновага лінійної системи не ізольована: множина рівноваг дорівнює $\ker A$. За $T=D=0$ матриця $A=0$ робить усі точки стаціонарними, тоді як ненульова нільпотентна матриця, наприклад $\begin{pmatrix}0&1\\0&0\end{pmatrix}$, породжує зсув: $x=x_0+t y_0$, $y=y_0$. Лише за слідом і визначником ці випадки не розрізнити.


```python
# planar-fixed-points-atlas
pf_cases = [
    ("Stable node", "Стійкий вузол", [[-1,0],[0,-2]], r"-1,\,-2"),
    ("Saddle", "Сідло", [[1,0],[0,-1]], r"1,\,-1"),
    ("Unstable node", "Нестійкий вузол", [[1,0],[0,2]], r"1,\,2"),
    ("Stable focus", "Стійкий фокус", [[-.35,-1],[1,-.35]], r"-0.35\pm i"),
    ("Center", "Центр", [[0,-1],[1,0]], r"\pm i"),
    ("Unstable focus", "Нестійкий фокус", [[.35,-1],[1,.35]], r"0.35\pm i"),
    ("Star node", "Зірковий вузол", [[-1,0],[0,-1]], r"-1,\,-1"),
    ("Defective node", "Дефектний вузол", [[-1,1],[0,-1]], r"-1,\,-1"),
    ("Line of equilibria", "Пряма рівноваг", [[-1,0],[0,0]], r"-1,\,0"),
]
pf_grid = np.linspace(-2, 2, 65)
pf_X, pf_Y = np.meshgrid(pf_grid, pf_grid)
pf_fig, pf_axes = plt.subplots(3,3,figsize=(12,12),layout="constrained")
for pf_ax, (pf_en, pf_ua, pf_matrix, pf_eiglabel) in zip(pf_axes.flat, pf_cases):
    pf_A = np.asarray(pf_matrix, dtype=float)
    pf_U = pf_A[0,0]*pf_X + pf_A[0,1]*pf_Y
    pf_V = pf_A[1,0]*pf_X + pf_A[1,1]*pf_Y
    pf_ax.streamplot(pf_grid, pf_grid, pf_U, pf_V, density=.85,
                     color="#475569", linewidth=.85, arrowsize=1.1)
    pf_values, pf_vectors = np.linalg.eig(pf_A)
    if np.all(np.abs(pf_values.imag) < 1e-10) and not np.allclose(pf_A, np.eye(2)*pf_A[0,0]):
        pf_seen = []
        for pf_value, pf_vector in zip(pf_values.real, pf_vectors.T.real):
            if abs(pf_value) < 1e-10:
                continue
            pf_vector = pf_vector / np.linalg.norm(pf_vector)
            if any(abs(np.dot(pf_vector, v)) > .9999 for v in pf_seen):
                continue
            pf_seen.append(pf_vector)
            pf_line = np.outer(np.linspace(-4,4,2), pf_vector)
            pf_ax.plot(pf_line[:,0], pf_line[:,1],
                color="#0072B2" if pf_value < 0 else "#D55E00",
                ls="--" if pf_value < 0 else ":", lw=2)
    if np.isclose(np.linalg.det(pf_A), 0):
        pf_ax.axvline(0, color="black", lw=2.5)
    else:
        pf_ax.plot(0,0,"ko",ms=4)
    pf_ax.set(xlim=(-2,2),ylim=(-2,2),xlabel=r"$\xi_1$",ylabel=r"$\xi_2$",
              title=pf_text(pf_en,pf_ua)+"\n"+r"$\lambda="+pf_eiglabel+"$")
    pf_ax.set_aspect("equal")
    pf_ax.grid(alpha=.15)
pf_fig.suptitle(pf_text("Planar linear systems: a phase-portrait atlas",
                       "Лінійні системи на площині: атлас фазових портретів"), fontsize=15)
plt.show()
```

### Що лінеаризація визначає, а чого не визначає

Стаціонарна точка є **гіперболічною**, якщо дійсні частини всіх власних значень якобіана ненульові. Для гладкого векторного поля теорема Гартмана—Гробмана встановлює локальну топологічну еквівалентність із лінеаризованим потоком. Отже, гіперболічні притягувальна чи відштовхувальна рівновага та сідло зберігають локальний характер стійкості; це не означає, що нелінійні траєкторії є точними еліпсами або прямими. Кратні власні значення з ненульовою дійсною частиною допустимі. Для негіперболічної точки потрібно дослідити нелінійні члени.

Три наведені нижче системи мають у початку координат той самий якобіан $\begin{pmatrix}0&-1\\1&0\end{pmatrix}$:

$$\dot x=-y+\sigma x(x^2+y^2),\qquad
\dot y=x+\sigma y(x^2+y^2),\qquad \sigma\in\{-1,0,1\}.$$

Оскільки $r\dot r=x\dot x+y\dot y$ та $r^2\dot\theta=x\dot y-y\dot x$, маємо

$$\dot r=\sigma r^3,\qquad \dot\theta=1,\qquad
r(t)=\frac{r_0}{\sqrt{1-2\sigma r_0^2t}}.$$

За $\sigma=-1$ початок координат притягує з алгебраїчним згасанням; за $\sigma=0$ це центр; за $\sigma=1$ він відштовхує. Останній вираз застосовуємо лише до занулення знаменника. Отже, однакові власні значення лінеаризації не визначають нелінійну стійкість. Рисунок побудовано за точними розв'язками, щоб числовий дрейф не перетворив центр на хибну спіраль.

Межі на діаграмі також мають різний зміст. Перетин $Q=0$ за $T\ne0$ перетворює вузол на фокус без втрати стійкості. На $D=0$ виникає нульове власне значення, що може уможливити біфуркацію стаціонарних станів. Перетин $T=0$ за $D>0$ є кандидатом на біфуркацію Гопфа, але додатково потрібні ненульова швидкість перетину та відповідні умови нелінійної невиродженості. Сама діаграма слід–визначник не доводить народження граничного циклу.


```python
# planar-fixed-points-nonlinear-comparison
pf_fig, pf_axes = plt.subplots(1,3,figsize=(12,4.3),layout="constrained")
pf_r0 = .3
pf_times = np.linspace(0,4.4,900)
pf_radii = []
for pf_ax, pf_sigma, pf_en, pf_ua in zip(pf_axes,[-1,0,1],
        ["Algebraic attraction", "Center", "Nonlinear repulsion"],
        ["Алгебраїчне притягання", "Центр", "Нелінійне відштовхування"]):
    pf_r = pf_r0/np.sqrt(1-2*pf_sigma*pf_r0**2*pf_times)
    pf_radii.append(pf_r)
    for pf_phase in np.linspace(0,2*np.pi,4,endpoint=False):
        pf_theta = pf_phase+pf_times
        pf_x, pf_y = pf_r*np.cos(pf_theta), pf_r*np.sin(pf_theta)
        pf_ax.plot(pf_x,pf_y,color="#0072B2",lw=1.6)
        pf_ax.plot(pf_x[0],pf_y[0],"o",color="#D55E00",ms=4)
        pf_k = 620
        pf_ax.annotate("",xy=(pf_x[pf_k+12],pf_y[pf_k+12]),
            xytext=(pf_x[pf_k],pf_y[pf_k]),
            arrowprops=dict(arrowstyle="->",color="#0072B2",lw=1.7))
    pf_ax.plot(0,0,"ko",ms=4)
    pf_ax.set(xlim=(-.72,.72),ylim=(-.72,.72),xlabel="$x$",ylabel="$y$",
              title=pf_text(pf_en,pf_ua)+"\n"+fr"$\sigma={pf_sigma},\quad\lambda=\pm i$")
    pf_ax.set_aspect("equal")
    pf_ax.grid(alpha=.2)
pf_fig.suptitle(pf_text("Same linearization, three nonlinear outcomes",
                       "Однакова лінеаризація — три нелінійні режими"),fontsize=14)
plt.show()
```

### Фізичний приклад і графічне дослідження

Для осцилятора з тертям $\ddot x+2\zeta\omega_0\dot x+\omega_0^2x=0$, де $\omega_0>0$ та $v=\dot x$,

$$A=\begin{pmatrix}0&1\\-\omega_0^2&-2\zeta\omega_0\end{pmatrix},\quad
T=-2\zeta\omega_0,\quad D=\omega_0^2,\quad Q=4\omega_0^2(\zeta^2-1).$$

За $0<\zeta<1$ маємо стійкий фокус (коливне згасання), за $\zeta>1$ — стійкий вузол (аперіодичне згасання), $\zeta=1$ відповідає дефектному стійкому вузлу (критичне згасання), а $\zeta=0$ — лінійному центру. Від'ємне тертя спричиняє нестійкість. Так класифікація набуває фізичного змісту й перестає бути лише переліком назв.

## Лінійна стійкість і нормальні форми

Підстановка $\mathbf{x}=\mathbf{x}_*+\boldsymbol\xi$ дає $\dot{\boldsymbol\xi}=J_*\boldsymbol\xi+O(\|\boldsymbol\xi\|^2)$. Локальна біфуркація відбувається, коли власне значення перетинає уявну вісь. Нормальна форма зберігає члени найнижчого порядку, яких не можна позбутися гладкою заміною змінних.

| Перехід | Нормальна форма | Характерний скейлінг |
| --- | --- | --- |
| Сідло-вузлова біфуркація | $\dot x=\mu-x^2$ | $x_*\sim\mu^{1/2}$ |
| Транскритична біфуркація | $\dot x=\mu x-x^2$ | гілки обмінюються стійкістю |
| Надкритична вилкова біфуркація | $\dot x=\mu x-x^3$ | $|x_*|\sim\mu^{1/2}$ |
| Надкритична біфуркація Гопфа | $\dot z=(\mu+i\omega)z-|z|^2z$ | $|z|\sim\mu^{1/2}$ |

Для вилкової біфуркації швидкість релаксації зі стійкого боку прямує до нуля за $\mu\to0$. Це і є критичне сповільнення.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "axes.grid": True,
    "grid.alpha": 0.25,
    "font.size": 11,
})
rng = np.random.default_rng(20260904)

mu = np.linspace(-1.0, 1.0, 500)

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].plot(mu, np.zeros_like(mu), "k--", label="x=0")
positive = mu >= 0
ax[0].plot(mu[positive], np.sqrt(mu[positive]), "C0", lw=2)
ax[0].plot(mu[positive], -np.sqrt(mu[positive]), "C0", lw=2)
ax[0].set(xlabel=r"control $\mu$", ylabel=r"equilibrium $x_*$", title="Pitchfork branches")

tau_below = 1 / np.maximum(np.abs(mu[mu < 0]), 1e-12)
ax[1].loglog(-mu[mu < 0], tau_below)
ax[1].set(xlabel=r"distance $|\mu|$", ylabel=r"relaxation time $\tau$", title="Critical slowing down")
plt.tight_layout()
```

```python
def rk4_scalar(f, x0, t, parameter):
    x = np.empty_like(t)
    x[0] = x0
    for n, dt in enumerate(np.diff(t)):
        k1 = f(x[n], parameter)
        k2 = f(x[n] + dt*k1/2, parameter)
        k3 = f(x[n] + dt*k2/2, parameter)
        k4 = f(x[n] + dt*k3, parameter)
        x[n+1] = x[n] + dt*(k1 + 2*k2 + 2*k3 + k4)/6
    return x

t = np.linspace(0, 80, 4001)
for m in [-0.5, -0.1, 0.05, 0.5]:
    trajectory = rk4_scalar(lambda x, mu_: mu_*x - x**3, 0.08, t, m)
    plt.plot(t, trajectory, label=fr"$\mu={m}$")
plt.xlabel("time")
plt.ylabel("order parameter x")
plt.title("Pitchfork normal form")
plt.legend()
plt.show()
```

## Автоколивання: енергія від сталого джерела

**Автоколивання** — це стійкий періодичний режим, амплітуду й період якого визначає внутрішня нелінійна динаміка системи, а не періодична зовнішня сила. Вільні коливання згасають через тертя; вимушені повторюють зовнішній ритм; автоколивна система живиться від сталого джерела енергії й сама дозує надходження енергії протягом циклу. Отже, це найпростіший часовий приклад структури у відкритій дисипативній системі.

Класична модель Бальтазара ван дер Поля виникла з теорії лампового генератора:

$$\ddot x-\mu(1-x^2)\dot x+\omega_0^2x=0,\qquad \mu>0.$$

У змінних $v=\dot x$ маємо

$$\dot x=v,\qquad \dot v=\mu(1-x^2)v-\omega_0^2x.$$

Поблизу початку координат коефіцієнт при $v$ додатний: мала флуктуація підсилюється, тобто стаціонарний стан нестійкий. За великих $|x|$ знак ефективного тертя змінюється і рух втрачає енергію. Для

$$E=\frac12\left(v^2+\omega_0^2x^2\right)$$

безпосередньо випливає

$$\dot E=\mu(1-x^2)v^2.$$

Тому енергія в середньому надходить на малих амплітудах і відводиться на великих. Стійкий **граничний цикл** встановлюється тоді, коли за повний період $\langle\dot E\rangle=0$. Його існування не означає відсутності дисипації: втрати постійно компенсує джерело через нелінійний зворотний зв'язок.

> **Синергетичний зміст.** Сталий потік енергії підтримує нестійкість спокою, нелінійність обмежує зростання, а система сама відбирає макроскопічний ритм. Фаза циклу стає нейтральною змінною, що згодом дає природний перехід до синхронізації.


```python
from scipy.integrate import solve_ivp

def van_der_pol(t, state, mu, omega0=1.0):
    x, v = state
    return v, mu * (1.0 - x**2) * v - omega0**2 * x

mus = [0.2, 1.0, 5.0]
initial_states = [(0.10, 0.0), (1.25, 0.75), (3.0, -0.5)]
t_eval = np.linspace(0.0, 80.0, 8001)

fig, axes = plt.subplots(2, 3, figsize=(13, 7))
for column, mu_value in enumerate(mus):
    reference = None
    for state0 in initial_states:
        sol = solve_ivp(
            van_der_pol, (t_eval[0], t_eval[-1]), state0,
            args=(mu_value,), t_eval=t_eval, method="DOP853",
            rtol=1e-8, atol=1e-10, max_step=0.04,
        )
        axes[1, column].plot(sol.y[0], sol.y[1], alpha=0.45, lw=1.0)
        tail = sol.t > 0.75 * sol.t[-1]
        axes[1, column].plot(sol.y[0, tail], sol.y[1, tail], lw=2.2)
        if state0 == initial_states[0]:
            reference = sol

    axes[0, column].plot(reference.t, reference.y[0], color=f"C{column}")
    axes[0, column].set(
        title=fr"$\mu={mu_value}$: $x(t)$", xlabel="час", ylabel="$x$",
    )
    axes[1, column].set(
        title="збіжність до граничного циклу", xlabel="$x$", ylabel="$v$",
    )
    axes[1, column].set_aspect("equal", adjustable="datalim")

plt.tight_layout()
plt.show()
```

```python
from scipy.signal import find_peaks

def cycle_diagnostics(mu_value, t_end=220.0):
    sample_times = np.linspace(0.0, t_end, 22001)
    sol = solve_ivp(
        van_der_pol, (0.0, t_end), (0.15, 0.0),
        args=(mu_value,), t_eval=sample_times, method="DOP853",
        rtol=1e-9, atol=1e-11, max_step=0.04,
    )
    mask = sol.t > 0.60 * t_end
    time_tail = sol.t[mask]
    x_tail, v_tail = sol.y[:, mask]
    peak_indices, _ = find_peaks(x_tail, prominence=0.2)
    cycle_peaks = peak_indices[-6:]
    period = np.mean(np.diff(time_tail[cycle_peaks]))
    cycle_slice = slice(cycle_peaks[0], cycle_peaks[-1] + 1)
    cycle_time = time_tail[cycle_slice]
    cycle_x, cycle_v = x_tail[cycle_slice], v_tail[cycle_slice]
    amplitude = 0.5 * (np.max(cycle_x) - np.min(cycle_x))
    power = mu_value * (1.0 - cycle_x**2) * cycle_v**2
    mean_power = np.trapezoid(power, cycle_time) / (cycle_time[-1] - cycle_time[0])
    return amplitude, period, mean_power

mu_scan = np.array([0.1, 0.2, 0.5, 1.0, 2.0, 5.0])
diagnostics = np.array([cycle_diagnostics(value) for value in mu_scan])

print("   mu    амплітуда      період     <dE/dt>")
for value, (amplitude, period, mean_power) in zip(mu_scan, diagnostics):
    print(f"{value:5.2f}    {amplitude:8.4f}    {period:8.4f}    {mean_power: .2e}")

fig, ax = plt.subplots(1, 3, figsize=(13, 3.6))
ax[0].plot(mu_scan, diagnostics[:, 0], "o-")
ax[0].set(xlabel=r"$\mu$", ylabel="амплітуда", title="Розмір циклу")
ax[1].plot(mu_scan, diagnostics[:, 1], "o-", color="C1")
ax[1].set(xlabel=r"$\mu$", ylabel="період", title="Уповільнення релаксаційного циклу")
ax[2].axhline(0.0, color="k", lw=1)
ax[2].plot(mu_scan, diagnostics[:, 2], "o-", color="C2")
ax[2].set(xlabel=r"$\mu$", ylabel=r"$\langle\dot E\rangle$", title="Баланс енергії")
plt.tight_layout()
plt.show()
```

## Родина моделей і межі аналогії

| Модель | Рівняння або структура | Що вона додає |
|---|---|---|
| Релея | $\ddot x+\omega_0^2x=\varepsilon(1-\dot x^2/3)\dot x$ | Нелінійне підведення й відведення енергії залежить від швидкості, а не від координати. |
| Льєнара | $\ddot x+f(x)\dot x+g(x)=0$ | Загальна площинна рамка; за додаткових умов теорема Льєнара гарантує єдиний стійкий граничний цикл. |
| Стюарта—Ландау | $\dot Z=(\lambda+i\omega)Z-(1+ic)|Z|^2Z$ | Нормальна форма м'якої гопфівської біфуркації: при $\lambda>0$ амплітуда прямує до $|Z|=\sqrt{\lambda}$. |
| ФіцГ'ю—Нагумо | швидкий активатор + повільне відновлення | Перехід від автоколивань до збудливості; спрощена мова для нервових імпульсів і активних середовищ. |

За $0<\mu\ll1$ осцилятор ван дер Поля майже гармонічний, а нелінійність повільно регулює амплітуду. За $\mu\gg1$ з'являються **релаксаційні коливання**: повільне накопичення на ділянках повільного многовиду чергується зі швидкими стрибками. Числова задача стає жорсткою, тому результат слід перевіряти з меншим кроком або неявним методом.

Для слабкого зв'язку стійкий цикл швидко відновлює амплітуду після збурення, тоді як фаза залишається повільною. Саме це обґрунтовує фазову редукцію й моделі Курамото в модулі 08. Не слід, однак, переносити параметр $\mu$ буквально з лампового генератора на серце, нейрон чи хімічну реакцію: спільною є геометрія динаміки, а не мікроскопічний механізм.
