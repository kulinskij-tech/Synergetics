## Необхідний вступ: моделі, простір станів і масштаби

Модель задає вектор стану $\mathbf{x}(t)$ та закон еволюції, наприклад $\dot{\mathbf{x}}=\mathbf{F}(\mathbf{x};\mu)$. Стан містить достатній набір змінних, щоб після задання керувань і зовнішніх впливів передбачити майбутнє самої моделі. Траєкторія — це крива у просторі станів; спостережувана $O(\mathbf{x})$ — величина, яку з неї обчислюють. Керувальний параметр $\mu$ задають ззовні, тоді як параметр порядку є колективною спостережуваною, обраною тому, що вона розрізняє макроскопічні режими.

Перед моделюванням оберіть характерні масштаби $X$ і $T$, запишіть $x=X\tilde x$, $t=T\tilde t$ та виділіть безрозмірні групи параметрів. Обезрозмірювання виявляє конкуренцію часових масштабів, зменшує ризик помилок з одиницями й дозволяє зіставляти на перший погляд різні системи. Слово «безрозмірний» у курсі означає, що цей крок уже виконано.


## До Гакена: звідки береться порядок без зовнішнього архітектора?
Довгий час живі системи, органічного та біологічного походження протиставлялись неживим системам неорганічної природи. Проте фізика в своєму загальному розумінні вивчає Природу, тому 
проникнення фізичних методів та підходів з моделювання складних систем у біологію ([про органічну хімію можна навіть не згадувати](https://www.nobelprize.org/prizes/chemistry/2024/press-release/)) ще раз свідчить що фізікалістичний підхід поки залишається єдиним конструктивним підходом дослідження та моделювання складних систем, до яких безумовно відносяться живі системи. Очевидною особливістю біологічних систем є здатність утворення та розвитку складних впорядкованих структур - самоорганізація на багатьох рівнів від клітинного до соціального. Як показав ще Е.Шредінгер, це не суперечить законам термодинаміки, оскільки всі живі біологічні системи не є замкнутими і обмінюються енергією і речовиною з навколишнім середовищем. Ентропія відкритої системи з часом може зменшуватись за рахунок її відтоку у зовнішнє середовище у вигляді високоетропійних продуктів життєдіяльності. Таким чином, необхідною умовою самоорганізації є наявність зовнішнього потоку енергії та його дисипація в системі. Завдяки цьому потоку система стає активною і набуває здатності до утворення структур. Найбільш простим прикладом такої ситуації є автоколивання. Отже ефекти самоорганізації не є характеристичною властивістю біологічних об'єктів, і повинні спостерігатися (зрозуміло ц більш простій формі) у "неживих" системах неорганічного походження. У всіх системах процес самоорганізації йде за участю великої кількості об'єктів які можна вважати ''елементарними'' (атоми, молекули, клітини, нейрони тінш) і визначається __кооперативною поведінкою__. Щоб наголосити на цій обставині, [H. Haken](https://en.wikipedia.org/wiki/Hermann_Haken) запровадив спеціальний термін __Синергетика__. Ця назва походить від грецького __synergeia__, що означає спільну, чи кооперативну, дію.
Вперше ідею сінегрії було введено англійським фізіологом [Charles Scott Sherrington](https://en.wikipedia.org/wiki/Charles_Scott_Sherrington) більге ста років тому в ході досліджень м'язових систем і управління ними з боку спинного мозку. У своїх ідеях про розум і пізнання Шеррінгтон представив ідею про те, що нейрони працюють як групи в ``мільйонній демократії`` для досягнення результатів, а не під централізованим контролем. Синергетика (від synergy) не виникла на порожньому місці. Термодинаміка XIX століття зробила поняття ентропії та необоротності неминучими, а досліди з конвекцією показали, що віддалений від рівноваги шар рідини може замість однорідного стану утворити макроскопічні вали. У XX столітті два кроки особливо загострили питання спонтанної організації. Виявляється також, що основні форми кооперативної поведінки, властиві живим організмам, мають свої аналоги серед неорганічних систем. Зараз Синергетика відноситься до [теорії складних систем](https://en.wikipedia.org/wiki/Complex_system).

- У книжці *What Is Life?* (1944) **Ервін Шредінгер** запитав, як організм зберігає впорядкованість, не порушуючи термодинаміки. Його образ «від'ємної ентропії», який згодом часто називали *негентропією*, найкраще розуміти як твердження про **відкриту систему**: організм споживає вільну енергію і відводить ентропію в довкілля. Локальне зростання порядку не скасовує другого начала.
- У статті [“The Chemical Basis of Morphogenesis” (1952)](https://royalsocietypublishing.org/rstb/article/237/641/37/112910/The-chemical-basis-of-morphogenesis) **Алан Тюрінг** запропонував уже не лише термодинамічне питання, а конкретний динамічний механізм. Локальні хімічні реакції разом із дифузією можуть дестабілізувати однорідний стан і вибрати просторову довжину хвилі. Дифузія, яку зазвичай пов'язують зі згладжуванням, здатна сприяти порушенню симетрії, якщо компоненти реагують і поширюються на різних масштабах.

## Синтез Гакена: від лазера до загальної дослідницької програми

Лазер зробив проблему виникнення порядку обчислюваною. Накачування підтримує інверсію населеностей, втрати в резонаторі розсіюють енергію, а безліч мікроскопічних випромінювачів взаємодіють через спільне електромагнітне поле. На порозі когерентна амплітуда поля втрачає стійкість і починає зростати — вона стає **параметром порядку**. Поблизу порога поляризація та інверсія часто релаксують швидше за поле і можуть бути **підпорядковані** йому. Поле породжується випромінювачами знизу вгору, але макроскопічне поле водночас визначає їх фазово когерентний відгук згори вниз. Гакен назвав цей взаємний зв'язок циклічною причинністю.

У такому сенсі проблема Шредінгерової «від'ємної ентропії» дістає в роботах Гакена операційне продовження: потрібно назвати джерело й дисипацію, знайти нестійкість, виділити повільну колективну змінну і вивести, як вона організовує решту ступенів вільності. У модулі 01 ми пройдемо цей шлях для лазерної моделі Максвелла—Блоха; наступні модулі перевірять, наскільки далеко переноситься та сама логіка.


## Сучасне продовження: спільні питання в дуже різних системах

Сучасні автори не завжди користуються словом *синергетика*, проте знов і знов ставлять цілком гакенівські питання про колективні змінні, нестійкість, синхронізацію, мультистабільність і редукцію.

Тому коректніше говорити не про буквальне перейменування, а про зміну наукової мови та розширення поля. Значна частина програми Гакена нині розвивається під ширшими назвами **наука про складність (теорія складних систем)**, нелінійна динаміка, самоорганізація, формування структур і мережева наука. Історична назва *синергетика* залишається особливо змістовною там, де пояснення спирається на параметри порядку, принцип підпорядкування та циклічну причинність.

> **Чому важко знайти нові огляди саме з назвою «синергетика»?** Ця програма нині живе радше як концептуальна й математична традиція, ніж як головне ключове слово в базах даних. [Довідковий том Springer 2020 року](https://doi.org/10.1007/978-1-0716-0421-2) прямо розміщує синергетику в *Encyclopedia of Complexity and Systems Science*. Серія статей 2025–2026 років повертається до програми Гакена в контексті складної динаміки, координації, міст і рекурентного аналізу; див., зокрема, [Atmanspacher (2025)](https://doi.org/10.1140/epjs/s11734-025-01706-9) та [Marwan, Braun & Kurths (2026)](https://doi.org/10.1140/epjs/s11734-026-02553-y). Для сучасного пошуку краще поєднувати назву моделі або механізму зі словами *параметр порядку*, *принцип підпорядкування*, *самоорганізація*, *амплітудне рівняння* чи *складні динамічні системи*, а не покладатися лише на слово *синергетика*.

Теорія випадкових нейронних мереж виявляє керований підсиленням перехід від стаціонарної до хаотичної колективної динаміки. Дослідження складних мереж аналізують синхронізацію в біологічних, технологічних і соціальних системах, а статистична фізика транспорту пояснює самоорганізовані затори та утворення пішохідних смуг. Ці системи не тотожні й не зобов'язані мати одне універсальне рівняння. Спільною є вимога зробити запропонований зв'язок між мікро- і макрорівнем перевірюваним.

## Основна ідея

Синергетика намагається зясувати, як поблизу нестійкості багатовимірна відкрита система може бути описана невеликою кількістю колективних змінних. В аналізі можна виділити п'ять рівнів:

1. Окреслити межі системи, джерела, дисипацію та симетрію.
2. Визначити **керувальний параметр** $\mu$, здатний провести систему через перехід.
3. Дослідити стійкість опорного стану і знайти м'яку моду.
4. Побудувати **параметр порядку** $A$ та, якщо це обґрунтовано, виключити швидкі стійкі моди.
5. Перевірити редуковану динаміку за даними або моделюванням і вказати межі, де вона перестає працювати.

Канонічною локальною формою є рівняння амплітуди

$$\dot A = \mu A-g|A|^2A+\eta(t),$$

але фізичний зміст $A$, допустимі члени та модель шуму мають випливати із симетрії системи й розділення масштабів.
