## Необхідний вступ: мережі та дифузійний зв'язок

Мережа складається з вузлів із локальними станами та ребер, які задають взаємодії. Матриця суміжності $A_{ij}$ фіксує зв'язки, а лапласіан графа $L=K-A$ порівнює кожен вузол із сусідами. Стандартний неперервний дифузійний зв'язок має вигляд

$$\dot{\mathbf x}_i=\mathbf F(\mathbf x_i)-\epsilon\sum_jL_{ij}\mathbf H(\mathbf x_j).$$

На синхронізованому многовиді $\mathbf x_1=\cdots=\mathbf x_N$. Згасання поперечних до нього збурень визначається одночасно локальною динамікою, силою зв'язку та власними значеннями лапласіана. У дискретних зв'язаних відображеннях важливий порядок оновлення: усі нові стани слід обчислювати з одного старого часового шару, якщо асинхронність не є навмисною частиною моделі.


## Ґратка зв'язаних відображень як лабораторія

Кільце логістичних відображень із дифузійним зв'язком можна записати як

$$x_i^{n+1}=(1-\epsilon)f(x_i^n)+\frac{\epsilon}{2}\left[f(x_{i-1}^n)+f(x_{i+1}^n)\right],\quad f(x)=rx(1-x).$$

Локальний хаос конкурує з просторовим зв'язком. Корисними спостережуваними є просторова дисперсія, кореляція сусідів, довжини ламінарних доменів і часовий ляпуновський ріст. Називати стан із малою дисперсією «організованим» доречно лише відносно чітко визначеної спостережуваної та перевірки робастності.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "axes.grid": True,
    "grid.alpha": 0.25,
    "font.size": 11,
})
rng = np.random.default_rng(20260904)

def coupled_logistic(epsilon, r=3.9, sites=120, steps=700, burn=200):
    x = rng.uniform(0.1, 0.9, sites)
    history = []
    for n in range(steps):
        fx = r*x*(1-x)
        x = ((1-epsilon)*fx
             + 0.5*epsilon*np.roll(fx, 1)
             + 0.5*epsilon*np.roll(fx, -1))
        if n >= burn:
            history.append(x.copy())
    return np.asarray(history)

eps_values = [0.0, 0.15, 0.35, 0.65]
fig, ax = plt.subplots(1, 4, figsize=(13, 4), sharey=True)
for axis, eps in zip(ax, eps_values):
    history = coupled_logistic(eps)
    axis.imshow(history, aspect="auto", origin="lower", cmap="viridis")
    axis.set(title=fr"$\epsilon={eps}$", xlabel="site")
ax[0].set_ylabel("time after transient")
plt.tight_layout()
```

## Малий зворотний зв'язок і керування хаосом

Ідея Отта—Гребоджі—Йорка використовує повторні наближення хаотичної траєкторії до нестійкої періодичної орбіти. Маленька, залежна від стану зміна доступного керувального параметра може стабілізувати цю орбіту. Збурення залишається малим, бо хаос сам підводить траєкторію близько до цілі, а зворотний зв'язок лише забезпечує локальне стискання. Практичний аналіз має враховувати межі зміни параметра, похибку спостереження, геометрію басейну та шум.
