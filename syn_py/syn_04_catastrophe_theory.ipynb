## Prerequisite bridge: gradient flow, metastability, and hysteresis

In a one-dimensional gradient system $\dot x=-\partial_xV(x;\mu)$, the potential decreases along trajectories because $\dot V=-(\partial_xV)^2\le0$. Local minima of $V$ are stable equilibria and local maxima are unstable. Multiple minima represent competing metastable macrostates separated by barriers.

Hysteresis is path dependence under parameter variation: a state may persist until its local minimum disappears at a fold, then jump to another branch. A potential is powerful when the dynamics is genuinely gradient-like; driven systems with circulation in state space generally require more than a scalar landscape.


## Cusp normal form

The canonical potential

$$V(x;a,b)=\frac{x^4}{4}+\frac{a x^2}{2}+bx$$

generates gradient dynamics $\dot x=-\partial_xV=-(x^3+ax+b)$. Equilibria satisfy $x^3+ax+b=0$ and are stable when $V''=3x^2+a>0$. Folds obey both $V'=0$ and $V''=0$, giving

$$a=-3x^2,\qquad b=2x^3,$$

and the cusp discriminant $4a^3+27b^2=0$.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "axes.grid": True,
    "grid.alpha": 0.25,
    "font.size": 11,
})
rng = np.random.default_rng(20260904)

def real_equilibria(a, b, tol=1e-9):
    roots = np.roots([1.0, 0.0, a, b])
    roots = roots[np.abs(roots.imag) < tol].real
    return np.sort(roots)

a_values = [-1.4, -0.5, 0.4]
x = np.linspace(-2.0, 2.0, 800)
fig, ax = plt.subplots(1, 3, figsize=(13, 3.8), sharey=True)
for axis, a in zip(ax, a_values):
    for b in [-0.4, 0.0, 0.4]:
        V = x**4/4 + a*x**2/2 + b*x
        axis.plot(x, V, label=fr"$b={b}$")
    axis.set(title=fr"$a={a}$", xlabel="state x")
ax[0].set_ylabel("potential V")
ax[-1].legend()
plt.tight_layout()
```

```python
# Equilibrium surface projected onto the (b, x) plane for fixed a < 0.
a = -1.0
x_branch = np.linspace(-1.4, 1.4, 1000)
b_branch = -(x_branch**3 + a*x_branch)
stable = 3*x_branch**2 + a > 0

plt.plot(b_branch[stable], x_branch[stable], "C0.", ms=2, label="stable")
plt.plot(b_branch[~stable], x_branch[~stable], "C3.", ms=2, label="unstable")
plt.xlabel("control b")
plt.ylabel("equilibrium x")
plt.title("Cusp slice and hysteresis branches")
plt.legend()
plt.show()
```
