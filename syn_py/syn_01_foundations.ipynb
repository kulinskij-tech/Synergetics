## Prerequisite bridge: why a laser needs threshold and feedback

An excited atom can emit spontaneously with an uncontrolled phase, or it can undergo stimulated emission into an already occupied electromagnetic mode. **Population inversion** means that the pumped gain medium has more atoms available for stimulated downward transitions than for absorption. A resonator repeatedly returns selected field modes through the medium, but mirrors and the medium also produce losses.

Below threshold, losses dominate and phase-incoherent spontaneous emission remains small. At threshold, gain balances loss for one mode. Above threshold, the field stimulates further emission into its own phase and spatial pattern, while depletion of the inversion limits growth. Coherence is therefore produced by nonlinear field–matter feedback. The Maxwell–Bloch variables used below are collective summaries: field amplitude $E$, material polarization $P$ (phase-coherent dipole response), and inversion $D$ (available gain).


### Populations, inversion, and material polarization

Consider $N$ effective two-level emitters, labeled by $j$, with upper and lower states $|e\rangle$ and $|g\rangle$. A **level population** is the mean number of emitters occupying that energy level. For emitter $j$, let $p_{e,j}$ and $p_{g,j}$ be the probabilities of occupying the upper and lower levels. Within the effective two-level model,

$$p_{e,j}+p_{g,j}=1,\qquad N_e=\sum_j p_{e,j},\qquad N_g=\sum_j p_{g,j},\qquad N_e+N_g=N.$$

The **population inversion** is their difference:

$$D=N_e-N_g,\qquad -N\le D\le N,\qquad N_e=\frac{N+D}{2},\qquad N_g=\frac{N-D}{2}.$$

Thus $D>0$ means that the upper laser level has the larger population. The variable $D$ in this notebook is the total inversion of the retained emitters. Inversion per emitter is $D/N$; inversion per unit volume is $D/V_g$, where $V_g$ is the active-medium volume. $D_0$ is the inversion maintained by pump and relaxation in the absence of the cavity field, while $D_c$ denotes its laser-threshold value.

For a quantum state described by the density operator $\rho$, an average is $\langle O\rangle=\operatorname{Tr}(\rho O)$, with state and operator expressed in the same representation. The reduced density matrix $\rho_j$ describes emitter $j$ after tracing out the field and other emitters. In the ordered basis $(|e\rangle,|g\rangle)$,

$$\rho_j=\begin{pmatrix}p_{e,j}&c_j\\c_j^*&p_{g,j}\end{pmatrix},\qquad
c_j=\langle\sigma_j^-\rangle,\qquad
\langle\sigma_j^z\rangle=p_{e,j}-p_{g,j}.$$

Here $\sigma_j^-=|g\rangle_j\langle e|$ lowers the emitter from the upper to the lower level, $\sigma_j^+=(\sigma_j^-)^\dagger$ raises it, and $\sigma_j^z=|e\rangle_j\langle e|-|g\rangle_j\langle g|$ measures its inversion. The off-diagonal entry $c_j$ is the **atomic coherence**: its magnitude and phase determine the emitter's oscillating mean transition dipole. A diagonal mixture has $c_j=0$, even if $D>0$. For example, $(|e\rangle+e^{i\phi}|g\rangle)/\sqrt2$ has equal populations and $c_j=e^{-i\phi}/2$.

**Material polarization** is electric dipole moment per unit volume. For a uniform active medium with identical transition dipoles, the laboratory polarization is given by the following expressions; here the averages use the laboratory density operator and the displayed dipole and transition operators are in the Schrödinger representation:

$$\mathbf d_{ge}=\langle g|\widehat{\mathbf d}|e\rangle,\qquad
\widehat{\mathbf d}_j=\mathbf d_{ge}\sigma_j^-+\mathbf d_{ge}^*\sigma_j^+,\qquad
\mathbf P_{\rm mat}=\frac1{V_g}\sum_j\langle\widehat{\mathbf d}_j\rangle.$$

Here we retain the optical transition contribution to the dipole. Coherences with aligned phases add to a macroscopic polarization; different phases can cancel in the sum. The collective coherence is $s=\sum_jc_j$. The later scalar variable $P$ represents a scaled amplitude of the component coupled to the selected cavity mode. Its normalization is specified in the semiclassical reduction; physical $\mathbf P_{\rm mat}$ has SI units C/m$^2$.

The operator $a$ annihilates a photon in the retained cavity mode, and $a^\dagger$ creates one. Similarly, $\alpha=\langle a\rangle$ is the coherent field amplitude in the chosen phase convention. The mean photon number is $n_{\rm ph}=\langle a^\dagger a\rangle$, and $n_{\rm coh}=|\alpha|^2$ measures its coherent contribution. Their difference is $\langle(a-\alpha)^\dagger(a-\alpha)\rangle\ge0$. The scalar $E$ below is a scaled field amplitude; $E^2$ is the plotted scaled intensity.


## The laser as Haken's prototype

Haken's synergetics grew from laser theory. A pumped gain medium contains many microscopic emitters whose spontaneous phases are initially incoherent. In a resonator, stimulated emission and field–matter feedback create a threshold: one cavity mode can acquire a stable macroscopic amplitude and phase relation. The result is not imposed phase-by-phase from outside; it is a collective instability of an open, driven, dissipative system.

In a real-variable, single-mode semiclassical reduction, let $E$ denote the cavity-field amplitude, $P$ the collective material polarization, and $D$ the population inversion:

$$\dot E=-\kappa E+gP,$$
$$\dot P=-\gamma_\perp P+gED,$$
$$\dot D=\gamma_\parallel(D_0-D)-2gEP.$$

Here $D_0$ is set by the pump, while $\kappa$, $\gamma_\perp$, and $\gamma_\parallel$ are field, polarization, and inversion relaxation rates. Linearizing the $(E,P)$ subsystem about the nonlasing state $(E,P,D)=(0,0,D_0)$ gives the threshold

$$D_c=\frac{\kappa\gamma_\perp}{g^2}.$$

Below $D_c$, field fluctuations decay. Above it, the nonlasing state is unstable and nonlinear depletion of $D$ saturates the coherent field.


### From quantum emitters to collective laser variables

The three equations above summarize several physical approximations. We now derive them before using them numerically. The purpose is to see how microscopic emitters and a common cavity field give rise to the collective variables of synergetics. No fast-variable elimination is needed yet; slaving is a further reduction discussed later.

Consider $N$ identical effective two-level emitters, with upper and lower laser levels $|e\rangle$ and $|g\rangle$. Define

$$\sigma_j^-=|g\rangle_j\langle e|,\qquad
\sigma_j^+=(\sigma_j^-)^\dagger,\qquad
\sigma_j^z=|e\rangle_j\langle e|-|g\rangle_j\langle g|.$$

We use three angular frequencies: $\omega_c$ is the resonance frequency of the selected empty-cavity mode, and $\omega_a=(E_e-E_g)/\hbar$ is the transition frequency of each emitter. The reference frequency $\omega_r$ specifies the optical phase factor $e^{-i\omega_r t}$ removed from the Heisenberg field operator to define its amplitude. It is a freely chosen reference, usually near the cavity and atomic resonances. Choosing $\omega_r=\omega_c$ gives zero cavity detuning $\Delta_c=\omega_c-\omega_r=0$; this and the atomic detuning $\Delta_a=\omega_a-\omega_r$ both vanish when $\omega_r=\omega_c=\omega_a$. All three are angular frequencies, measured in radians per second.

The lowering operator describes an atomic de-excitation; $\sigma_j^z$ has eigenvalues $+1$ and $-1$. For one cavity mode, $a_H(t)$ is the laboratory Heisenberg photon-annihilation operator, with $[a_H(t),a_H^\dagger(t)]=1$. The positive-frequency electric-field operator in the same representation is approximated by

$$\widehat{\mathcal E}^{(+)}_H(\mathbf r,t)
=\mathcal E_{\rm vac}u(\mathbf r)a_H(t).$$

Thus **single-mode** means retaining one spatial profile $u(\mathbf r)$ and its time-dependent amplitude. It does not mean that this amplitude is real. Assume identical transition frequencies and a uniform real atom–mode coupling $g$; this neglects inhomogeneous broadening and spatial variations of the coupling.

Here $u(\mathbf r)$ is the dimensionless spatial mode profile, with its normalization absorbed into $\mathcal E_{\rm vac}$, the vacuum electric-field scale fixed by the chosen mode normalization (units V/m). The superscript $(+)$ denotes the positive-frequency component; the Hermitian electric field includes its adjoint. The real coupling $g$ measures the dipole–mode interaction strength in angular-frequency units; its sign can be absorbed in the phase convention.

**Operator representation.** The subscript $H$ labels laboratory Heisenberg operators. A tilde labels every operator after removal of the reference optical phase:

$$\widetilde a(t)=e^{i\omega_r t}a_H(t),\qquad
\widetilde\sigma^-_j(t)=e^{i\omega_r t}\sigma^-_{j,H}(t),\qquad
\widetilde\sigma^z_j(t)=\sigma^z_{j,H}(t).$$

Adjoint operators acquire the opposite phase: $\widetilde a^\dagger=e^{-i\omega_r t}a_H^\dagger$ and $\widetilde\sigma_j^+=e^{-i\omega_r t}\sigma_{j,H}^+$. The tilded quantities remain operators. The laboratory field is recovered as

$$\widehat{\mathcal E}^{(+)}_H(\mathbf r,t)
=\mathcal E_{\rm vac}u(\mathbf r)a_H(t)
=\mathcal E_{\rm vac}u(\mathbf r)\widetilde a(t)e^{-i\omega_r t}.$$

For a free cavity, $\dot a_H=-i\omega_c a_H$, so differentiating the definition gives

$$\frac{d\widetilde a}{dt}=i\omega_r\widetilde a+e^{i\omega_r t}\dot a_H
=-i(\omega_c-\omega_r)\widetilde a.$$

To specify the representation of the dynamics, use the time-independent Schrödinger operators $a_S,\sigma_{j,S}$ in

$$H_{\rm ref}=\hbar\omega_r\mathcal N,\qquad
\mathcal N=a_S^\dagger a_S+\frac12\sum_j(\sigma_{j,S}^z+1).$$

Here $\mathcal N$ counts photons plus excited emitters. For the closed Hamiltonian part, the state transformation $|\psi_{\rm r}\rangle=e^{iH_{\rm ref}t/\hbar}|\psi_S\rangle$ is an interaction representation relative to $H_{\rm ref}$. Its Hamiltonian is

$$H_{\rm r}=e^{iH_{\rm ref}t/\hbar}H_{\rm lab}e^{-iH_{\rm ref}t/\hbar}-H_{\rm ref}
=H_{\rm lab}-H_{\rm ref},$$

where the last equality uses $[H_{\rm lab},\mathcal N]=0$ in the rotating-wave approximation. The tilded operators satisfy Heisenberg equations generated by $H_{\rm r}$. The conventional expression “rotating frame” refers to this change of optical phase. The rotating-wave approximation removes counter-rotating interaction terms.

An additive scalar constant in the Hamiltonian has no effect on these equations and is omitted below.

**Notation from here onward:** $a\equiv\widetilde a$, $\sigma_j^{\pm,z}\equiv\widetilde\sigma_j^{\pm,z}$, and $H\equiv H_{\rm r}$. Both field and atomic operators thus use the same phase convention. In the density-operator equations below, $\rho$ denotes the state in this transformed representation, and unadorned operators inside traces are its Schrödinger-picture observables. Their averages equal the averages of the corresponding tilded Heisenberg operators evaluated with the initial state.

In the electric-dipole and rotating-wave approximations, the Hamiltonian is

$$\frac{H}{\hbar}=\Delta_c a^\dagger a
+\frac{\Delta_a}{2}\sum_j\sigma_j^z
+g\sum_j(a^\dagger\sigma_j^-+a\sigma_j^+),$$

where $\Delta_c=\omega_c-\omega_r$ and $\Delta_a=\omega_a-\omega_r$. The term $a^\dagger\sigma_j^-$ creates a photon while de-exciting an emitter; its Hermitian conjugate describes absorption. The rotating-wave approximation discards rapidly oscillating terms that create or remove both excitations together.

Unlike the scaled numerical example, this microscopic model initially uses physical time: $g$, the detunings, and the decay rates have units of inverse time. Choosing a time unit makes them dimensionless.


### Pumping, losses, and the quantum master equation

The Hamiltonian alone cannot describe a working laser: pumping supplies energy and losses remove it. A minimal Markovian reservoir model is

$$\dot\rho=-\frac{i}{\hbar}[H,\rho]+2\kappa\mathcal D[a]\rho
+\sum_j\left(w\mathcal D[\sigma_j^+]\rho
+\Gamma\mathcal D[\sigma_j^-]\rho
+\frac{\gamma_\phi}{2}\mathcal D[\sigma_j^z]\rho\right),$$

$$\mathcal D[L]\rho=L\rho L^\dagger-\frac12\{L^\dagger L,\rho\}.$$

Here $\rho$ is the joint atom–field density operator. The dissipator $\mathcal D[L]$ represents the irreversible process associated with $L$: $w$ is an effective incoherent pump rate, $\Gamma$ is atomic decay into channels other than the retained cavity mode, and $\gamma_\phi$ destroys atomic phase coherence without changing populations. The effective pump represents a reservoir or auxiliary levels of a multilevel gain medium, rather than resonant coherent excitation of an isolated two-level atom.

The factor $2\kappa$ fixes a convention: $\kappa$ is the **field-amplitude** decay rate. For an empty cavity,

$$\frac{d}{dt}\langle a\rangle=-\kappa\langle a\rangle,
\qquad\frac{d}{dt}\langle a^\dagger a\rangle=-2\kappa\langle a^\dagger a\rangle.$$

For an uncoupled emitter, its inversion $z=\langle\sigma^z\rangle$ obeys

$$\dot z=w(1-z)-\Gamma(1+z)
=-(w+\Gamma)\left(z-\frac{w-\Gamma}{w+\Gamma}\right).$$

Consequently, for this reservoir model,

$$\gamma_\parallel=w+\Gamma,\qquad
\gamma_\perp=\frac{w+\Gamma}{2}+\gamma_\phi,\qquad
D_0=N\frac{w-\Gamma}{w+\Gamma}.$$

The pump and decay drive the total inversion toward $D_0$ when the cavity field is absent. Inversion requires $w>\Gamma$ in this model. More detailed multilevel models produce other effective relations; in the phenomenological calculations below, $D_0$ is varied while the relaxation rates are held fixed. That scan is not a literal variation of $w$ alone in this particular reservoir model.


### Expectation values: where the closure problem appears

Introduce the collective operators

$$S^-=\sum_j\sigma_j^-,\qquad S^+=(S^-)^\dagger,\qquad Z=\sum_j\sigma_j^z.$$

The expectation of an observable is $\langle O\rangle=\operatorname{Tr}(\rho O)$. Its evolution follows from

$$\frac{d}{dt}\langle O\rangle=\frac{i}{\hbar}\langle[H,O]\rangle
+\text{reservoir contributions}.$$

Using $[\sigma^+,\sigma^-]=\sigma^z$ and $[\sigma^z,\sigma^\pm]=\pm2\sigma^\pm$ gives

$$\begin{aligned}
\frac{d}{dt}\langle a\rangle
&=-(\kappa+i\Delta_c)\langle a\rangle-ig\langle S^-\rangle,\\
\frac{d}{dt}\langle S^-\rangle
&=-(\gamma_\perp+i\Delta_a)\langle S^-\rangle+ig\langle aZ\rangle,\\
\frac{d}{dt}\langle Z\rangle
&=\gamma_\parallel(D_0-\langle Z\rangle)
+2ig(\langle a^\dagger S^-\rangle-\langle aS^+\rangle).
\end{aligned}$$

For example, $[a^\dagger S^-+aS^+,S^-]=aZ$, which generates the field-times-inversion term in the polarization equation. These equations are exact within the stated quantum model. They are not closed, because averages of products such as $\langle aZ\rangle$ are not generally products of averages. Equations for these correlations would introduce further moments.


### Semiclassical factorization and normalization

Set $\alpha=\langle a\rangle$, $s=\langle S^-\rangle$, and $D=\langle Z\rangle$. The semiclassical mean-field approximation neglects connected atom–field correlations:

$$\langle aZ\rangle\simeq\alpha D,\qquad
\langle a^\dagger S^-\rangle\simeq\alpha^*s,\qquad
\langle aS^+\rangle\simeq\alpha s^*.$$

For example, $\langle aZ\rangle=\alpha D+\langle(a-\alpha)(Z-D)\rangle$, and we discard the second term. This gives

$$\begin{aligned}
\dot\alpha&=-(\kappa+i\Delta_c)\alpha-igs,\\
\dot s&=-(\gamma_\perp+i\Delta_a)s+ig\alpha D,\\
\dot D&=\gamma_\parallel(D_0-D)+2ig(\alpha^*s-\alpha s^*).
\end{aligned}$$

Define scaled complex amplitudes

$$\mathsf E=\sqrt2\,\alpha,\qquad\mathsf P=-i\sqrt2\,s.$$

The phase factor absorbs the $i$ in the coupling, and the scale factors fix the coefficients to match the notebook. The resulting complex Maxwell–Bloch equations are

$$\boxed{\begin{aligned}
\dot{\mathsf E}&=-(\kappa+i\Delta_c)\mathsf E+g\mathsf P,\\
\dot{\mathsf P}&=-(\gamma_\perp+i\Delta_a)\mathsf P+g\mathsf E D,\\
\dot D&=\gamma_\parallel(D_0-D)-2g\operatorname{Re}(\mathsf E^*\mathsf P).
\end{aligned}}$$

The polarization is a collective atomic coherence, not an excited-state population; populations enter through $D$. At this stage the model has five real variables: two components each for field and polarization, and one inversion. Factorization omits quantum correlations and fluctuation dynamics and is not a uniform approximation arbitrarily close to threshold or in a few-emitter, few-photon regime.


### Resonance, fixed phase, and the three real equations

Assume exact cavity–atom resonance and rotate at the common frequency, so $\Delta_c=\Delta_a=0$. The complex equations preserve the subspace

$$\mathsf E=e^{i\theta}E,\qquad\mathsf P=e^{i\theta}P,\qquad E,P\in\mathbb R,$$

where $\theta$ is constant. Choosing that common phase as our reference gives exactly

$$\boxed{\begin{aligned}
\dot E&=-\kappa E+gP,\\
\dot P&=-\gamma_\perp P+gED,\\
\dot D&=\gamma_\parallel(D_0-D)-2gEP.
\end{aligned}}$$

This is a fixed-phase sector of the resonant model. A common phase choice cannot in general make two arbitrarily phased complex amplitudes real simultaneously; detuning and general relative-phase dynamics require the complex equations.

An excitation-balance check explains the factor $2$. The coherent photon population is $n_{\rm coh}=|\alpha|^2=E^2/2$, and the mean upper-level population is $N_e=(N+D)/2$. Keeping only interaction terms,

$$\left.\dot n_{\rm coh}\right|_{\rm int}=gEP,\qquad
\left.\dot N_e\right|_{\rm int}=-gEP.$$

One de-excitation adds one photon, but decreases $D=N_e-N_g$ by two. The plots below use $E^2$ as a scaled intensity, proportional to photon population. In the full quantum theory $\langle a^\dagger a\rangle$ need not equal $|\langle a\rangle|^2$.


### From collective instability to the order parameter

Linearizing about $(E,P,D)=(0,0,D_0)$ gives $\delta\dot D=-\gamma_\parallel\delta D$ and

$$\frac{d}{dt}\begin{pmatrix}\delta E\\\delta P\end{pmatrix}
=\begin{pmatrix}-\kappa&g\\gD_0&-\gamma_\perp\end{pmatrix}
\begin{pmatrix}\delta E\\\delta P\end{pmatrix}.$$

The two growth rates solve

$$\lambda^2+(\kappa+\gamma_\perp)\lambda
+\kappa\gamma_\perp-g^2D_0=0.$$

The trace is negative and the determinant changes sign at $D_c=\kappa\gamma_\perp/g^2$. Thus a collective field–polarization perturbation changes from decaying to growing at threshold. At $D_0=D_c$, one eigenvalue is zero and the other is $-(\kappa+\gamma_\perp)$; the critical eigenvector satisfies $\delta P=(\kappa/g)\delta E$. The slow variable therefore represents a coupled field–matter mode, which can be parametrized by $E$.

For a nonzero stationary field the first two equations imply

$$P_s=\frac{\kappa}{g}E_s,\qquad D_s=D_c.$$

The third then gives

$$E_s^2=\frac{\gamma_\parallel}{2\kappa}(D_0-D_c).$$

The field depletes the inversion until gain balances loss. This is the stationary branch plotted below; its stability farther above threshold requires a separate analysis. The field both results from collective polarization and changes the medium that generates it: the circular causality central to Haken's interpretation.

The deterministic equations leave exactly $E=P=0$ unchanged even above threshold. The small seed in the numerical example represents a perturbation; quantum fluctuations provide the physical startup mechanism. In an unseeded finite quantum laser the ensemble mean $\langle a\rangle$ can remain zero because the phase is not selected, despite a nonzero photon population. A nonzero semiclassical amplitude describes a selected phase or a realization over times where phase diffusion can be neglected.

The derivation has retained all three material–field variables. The later slaving approximation needs an additional time-scale argument: near threshold the collective growth rate becomes small compared with noncritical relaxation rates. Simply calling the model semiclassical does not justify setting $\dot P$ or $\dot D$ to zero.

**Further reading.** G. J. de Valcárcel, E. Roldán, and F. Prati, [*Semiclassical Theory of Amplification and Lasing*](https://arxiv.org/abs/quant-ph/0605084), develops the Maxwell–Bloch description and its relation to multilevel gain media. Haken's *Synergetics* supplies the course's interpretation in terms of instability, order parameters, and slaving.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "axes.grid": True,
    "grid.alpha": 0.25,
    "font.size": 11,
})
rng = np.random.default_rng(20260904)

from scipy.integrate import solve_ivp

kap, gamma_perp, gamma_parallel, g = 1.0, 8.0, 0.7, 2.0
D_c = kap*gamma_perp/g**2

def laser(t, state, D0):
    E, P, D = state
    return (-kap*E + g*P,
            -gamma_perp*P + g*E*D,
            gamma_parallel*(D0-D) - 2*g*E*P)

t_eval = np.linspace(0, 60, 4000)
pumps = [0.7*D_c, 1.02*D_c, 1.6*D_c]

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
steady_intensity = []
for D0 in pumps:
    sol = solve_ivp(laser, (0, 60), [1e-4, 0.0, D0], args=(D0,), t_eval=t_eval,
                    rtol=1e-9, atol=1e-11)
    E, P, D = sol.y
    ax[0].plot(sol.t, E**2, label=fr"$D_0/D_c={D0/D_c:.2f}$")
    steady_intensity.append(np.mean(E[sol.t > 50]**2))
ax[0].set(xlabel="time", ylabel=r"intensity $E^2$", title="Laser threshold dynamics")
ax[0].legend()

pump_scan = np.linspace(0.4*D_c, 2.0*D_c, 60)
theory = gamma_parallel*np.maximum(pump_scan-D_c, 0)/(2*kap)
ax[1].plot(pump_scan/D_c, theory, lw=2)
ax[1].set(xlabel=r"pump $D_0/D_c$", ylabel=r"stationary $E^2$",
          title="Order parameter above threshold")
plt.tight_layout()
```

## Slaving and circular causality

If material variables relax faster than the field, set $\dot P\approx0$ and $\dot D\approx0$:

$$P\approx\frac{g}{\gamma_\perp}ED,\qquad
D\approx\frac{D_0}{1+2g^2E^2/(\gamma_\perp\gamma_\parallel)}.$$

Substitution into the field equation and expansion near $E=0$ gives

$$\dot E\approx \mu E-\beta E^3,\qquad
\mu=\kappa\left(\frac{D_0}{D_c}-1\right),\quad \beta>0.$$

The coherent field $E$ is the slow order parameter. The fast atomic variables are approximately functions of it: they are **slaved**. Yet the atoms generate the field through stimulated polarization, while the field organizes the atoms. Haken used this feedback as the paradigmatic case of circular causality: microscopic activity creates the macroscopic order parameter, which then constrains microscopic activity.


```python
# Test the slaving relation on the above-threshold trajectory.
D0 = 1.6*D_c
sol = solve_ivp(laser, (0, 60), [1e-4, 0.0, D0], args=(D0,), t_eval=t_eval,
                rtol=1e-9, atol=1e-11)
E, P, D = sol.y
P_slaved = g*E*D/gamma_perp
D_slaved = D0/(1 + 2*g**2*E**2/(gamma_perp*gamma_parallel))
mask = sol.t > 10

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].plot(sol.t, P, label="full P")
ax[0].plot(sol.t, P_slaved, "--", label="slaved P(E,D)")
ax[0].set(xlabel="time", ylabel="polarization", title="Fast polarization follows the field")
ax[0].legend()
ax[1].plot(E[mask]**2, D[mask], ".", ms=1.5, label="full dynamics")
order = np.argsort(E[mask]**2)
ax[1].plot(E[mask][order]**2, D_slaved[mask][order], "k--", label="slaved inversion")
ax[1].set(xlabel=r"intensity $E^2$", ylabel="inversion D", title="Gain saturation")
ax[1].legend()
plt.tight_layout()
```

## Gradient Dynamics of the Order Parameter: From Synergetics to Catastrophe Theory

The slaving calculation above has reduced a field–matter system to one **real, signed** amplitude $E$. Its two signs describe phases differing by $\pi$ in the resonant fixed-phase sector; intensity is $E^2$. We can now interpret the same amplitude equation as motion down a potential landscape. This connects Haken's order parameter with [Module 04: potential landscapes and catastrophe theory](https://colab.research.google.com/github/kulinskij-tech/Synergetics/blob/main/syn_py/syn_04_catastrophe_theory.ipynb).

### Keep the coefficients of the actual laser reduction

Define $c=2g^2/(\gamma_\perp\gamma_\parallel)$ and $A=g^2D_0/\gamma_\perp=\kappa D_0/D_c$. Substituting the displayed slaving relations gives, **within the quasistatic approximation**,

$$\dot E=\left[-\kappa+\frac{A}{1+cE^2}\right]E
=(A-\kappa)E-AcE^3+O(E^5).$$

Thus the notebook's coefficients are

$$\mu=\kappa\left(\frac{D_0}{D_c}-1\right),\qquad
\beta(D_0)=Ac=\frac{2g^4D_0}{\gamma_\perp^2\gamma_\parallel}>0.$$

At leading near-threshold order one may use the constant

$$\beta_c=\beta(D_c)=\frac{2\kappa g^2}{\gamma_\perp\gamma_\parallel}
=\frac{2\kappa^2}{\gamma_\parallel D_c},$$

because $\beta(D_0)-\beta_c=O(\mu)$ and $E=O(\sqrt{\mu})$ on the emerging branch. With $\beta_c$, $E_s^2=\mu/\beta_c=\gamma_\parallel(D_0-D_c)/(2\kappa)$ matches the leading onset (and the full model's exact stationary formula). With $\beta(D_0)$, the cubic truncation gives $E_s^2=\gamma_\parallel D_c(D_0-D_c)/(2\kappa D_0)$, differing beyond leading order. Keep $cE^2\ll1$ and stay near threshold; neither polynomial is a global laser model.

**Time-scale qualification.** Small collective growth rate alone does not make $\dot P=\dot D=0$ an exact reduction in physical time. For finite $\gamma_\perp/\kappa$, the full nonlasing Jacobian gives $\lambda_{\rm slow}\simeq M\mu$, where $M=\gamma_\perp/(\gamma_\perp+\kappa)$. The leading center-manifold equation in the field coordinate is $\dot E=M(\mu E-\beta_cE^3)$ (higher-order terms omitted). The notebook's quasistatic coefficients apply when polarization follows sufficiently rapidly, or in the rescaled time $s=Mt$ at this order. This positive factor changes relaxation times, not equilibria or their stability. The plots below explicitly use the quasistatic normal form.

### Integrate the force to obtain a quartic potential

For fixed controls and a chosen positive $\beta$, integrate $V_E=-\mu E+\beta E^3$:

$$V(E;\mu)= -\frac{\mu}{2}E^2+\frac{\beta}{4}E^4+C,
\qquad \dot E=-\frac{\partial V}{\partial E}.$$

The factors $1/2$ and $1/4$ ensure the correct drift. The arbitrary constant $C$ does not affect motion. As an independent check, the unexpanded quasistatic drift has potential $V_{\rm qs}=\kappa E^2/2-[A/(2c)]\ln(1+cE^2)$; expanding it recovers precisely the quadratic and quartic coefficients above. This is an **effective dynamical potential**, with units $E^2/\text{time}$ for unit mobility, not an identified thermodynamic free energy of the pumped laser.

Along a trajectory with constant $\mu,\beta$,

$$\frac{dV}{dt}=V_E\dot E=-(V_E)^2=-(\dot E)^2\le0.$$

Equality holds exactly at equilibria. For $\dot E=-M V_E$ with constant $M>0$, $dV/dt=-M(V_E)^2$. Since $\beta>0$, the quartic potential confines this scalar model; it has no nonconstant periodic orbit. Its Lyapunov property does not assert monotone decrease of physical energy in a driven open system.


### Pitchfork equilibria and stability

Writing $f(E)=\mu E-\beta E^3$, equilibria solve $E(\mu-\beta E^2)=0$, and perturbations obey $\delta\dot E=f'(E_s)\delta E=-V''(E_s)\delta E$.

| Control | Equilibria | Local behavior |
|---|---|---|
| $\mu<0$ | $E_s=0$ | $f'(0)=\mu<0$: asymptotically stable minimum |
| $\mu=0$ | $E_s=0$ | Nonhyperbolic but asymptotically stable: $\dot E=-\beta E^3$ |
| $\mu>0$ | $E_s=0$ | $f'(0)=\mu>0$: unstable maximum |
| $\mu>0$ | $E_s=\pm\sqrt{\mu/\beta}$ | $f'(E_s)=-2\mu<0$: two stable minima |

This is a **supercritical pitchfork** in the signed real amplitude. At threshold the linear test is inconclusive, but the exact solution $E(t)=E(0)/\sqrt{1+2\beta E(0)^2t}$ proves algebraic attraction. Away from threshold the local relaxation times are $1/|\mu|$ below threshold and $1/(2\mu)$ on the stable nonzero branches (unit mobility): critical slowing down as $\mu\to0$. The two field signs have the same intensity. For an unrestricted complex laser amplitude they belong to a continuous phase family, so this real pitchfork must not be interpreted as exactly two possible optical phases.

The first plot uses $\beta_c$ from the laser parameters above and a small reduced pump interval. Solid branches are stable; dashed branches are unstable. The second example integrates the scalar normal form and checks its Lyapunov identity numerically. These are model illustrations, not measured laser data.


```python
# gradient_landscape: near-threshold quasistatic normal form, unit mobility.
beta_c = 2*kap*g**2/(gamma_perp*gamma_parallel)
def gradient_potential(field, mu, beta=beta_c, h=0.0):
    return -mu*field**2/2 + beta*field**4/4 - h*field

field_grid = np.linspace(-0.55, 0.55, 600)
fig, axes = plt.subplots(1, 2, figsize=(11, 4), constrained_layout=True)
for mu_value in [-0.08*kap, 0.0, 0.08*kap]:
    axes[0].plot(field_grid, gradient_potential(field_grid, mu_value),
                 label=fr'$\mu/\kappa={mu_value/kap:.2f}$')
axes[0].set(xlabel=r'$E$', ylabel=r'$V(E)$', title='Potential near threshold')
axes[0].legend()
mu_scan = np.linspace(-0.1*kap, 0.1*kap, 500)
positive_mu = mu_scan[mu_scan > 0]
axes[1].plot(mu_scan[mu_scan <= 0]/kap, np.zeros(np.sum(mu_scan <= 0)),
             color='C0', label='stable')
axes[1].plot(positive_mu/kap, np.zeros_like(positive_mu), 'k--', label='unstable')
for sign in [-1, 1]:
    axes[1].plot(positive_mu/kap, sign*np.sqrt(positive_mu/beta_c), color='C0')
axes[1].plot(0, 0, 'ko', ms=4)
axes[1].set(xlabel=r'$\mu/\kappa=D_0/D_c-1$', ylabel=r'$E_s$',
            title='Supercritical pitchfork')
axes[1].legend()
plt.show()
```

```python
# gradient_lyapunov: fixed controls; change initial fields and rerun.
from scipy.integrate import cumulative_trapezoid
mu_demo = 0.08*kap
times_gradient = np.linspace(0, 100/kap, 24001)
fig, axes = plt.subplots(1, 2, figsize=(11, 4), constrained_layout=True)
lyapunov_errors = []
for initial_field in [-0.45, -0.03, 0.03, 0.45]:
    trajectory = solve_ivp(
        lambda t, y: mu_demo*y-beta_c*y**3,
        (times_gradient[0], times_gradient[-1]), [initial_field],
        t_eval=times_gradient, rtol=1e-10, atol=1e-12)
    assert trajectory.success, trajectory.message
    field_path = trajectory.y[0]
    potential_path = gradient_potential(field_path, mu_demo)
    drift_path = mu_demo*field_path-beta_c*field_path**3
    balance = potential_path-potential_path[0] + cumulative_trapezoid(
        drift_path**2, times_gradient, initial=0)
    lyapunov_errors.append(np.max(np.abs(balance)))
    assert np.max(np.diff(potential_path)) < 1e-10
    axes[0].plot(times_gradient*kap, field_path, label=fr'$E(0)={initial_field}$')
    axes[1].plot(times_gradient*kap, potential_path-potential_path[0])
axes[0].set(xlabel=r'$\kappa t$', ylabel=r'$E(t)$', title='Attraction to either minimum')
axes[0].legend()
axes[1].set(xlabel=r'$\kappa t$', ylabel=r'$V(t)-V(0)$', title='Lyapunov decrease')
print('max |V(t)-V(0)+integral(dE/dt)^2 dt| =', max(lyapunov_errors))
assert max(lyapunov_errors) < 1e-7
plt.show()
```

### Break the symmetry: imperfect pitchfork and cusp unfolding

A small constant bias $h$ breaks $E\mapsto-E$ symmetry:

$$\dot E=\mu E-\beta E^3+h=-\partial_E V_h,\qquad
V_h=\frac{\beta E^4}{4}-\frac{\mu E^2}{2}-hE.$$

Here $h$ has units of amplitude/time. In a fixed-phase model it may represent a weak coherent seed projected onto the retained quadrature; it is not the incoherent pump $D_0$. It is a phenomenological unfolding parameter, not a derivation of a general injected-laser model.

Choose an amplitude scale $E_*>0$, set $x=E/E_*$ and $\tau=\beta E_*^2t$, and divide the potential by $\beta E_*^4$. Then

$$U(x;a,b)=\frac{x^4}{4}+\frac{a x^2}{2}+bx,\qquad
\frac{dx}{d\tau}=-(x^3+ax+b),\quad
a=-\frac{\mu}{\beta E_*^2},\quad b=-\frac{h}{\beta E_*^3}.$$

This is the same sign convention as Module 04. The symmetric pitchfork is the $b=0$ slice through the **two-control cusp unfolding**. For the equilibrium cubic $x^3+ax+b=0$, define the algebraic discriminant

$$\Delta=-4a^3-27b^2=\frac{4\mu^3-27\beta h^2}{\beta^3 E_*^6}.$$

For $\Delta>0$ there are three distinct real roots: two stable minima and the intervening unstable maximum. For $\Delta<0$ there is one real root, a stable minimum. At $\Delta=0$ roots merge: generically a stable and an unstable equilibrium meet at a fold; the triple root at $(a,b)=(0,0)$ is the cusp. Nonhyperbolic roots need nonlinear analysis. A generic double fold root is attracting from one side and repelling from the other; the triple cusp root is algebraically attracting at fixed controls.

Solving $U_x=U_{xx}=0$ gives

$$a=-3x_f^2,\qquad b=2x_f^3,\qquad 4a^3+27b^2=0.$$

Equivalently, $\mu=3\beta E_f^2$, $h=-2\beta E_f^3$, so the three-equilibrium region satisfies $\mu>0$ and $|h|<2\mu^{3/2}/(3\sqrt{3\beta})$. The expression $4a^3+27b^2$ is often called a cusp discriminant too; its sign is opposite to $\Delta$. At fixed $\mu>0$, varying $h$ tilts the wells. A deterministic state can remain in a local minimum until that minimum disappears at a fold. Equal depths ($h=0$) do not by themselves force switching. Slow forward/backward sweeps can therefore show hysteresis; finite sweep rates and noise require further dynamics.

**Explore:** change `a_explore` and `b_explore` below, rerun, and compare the potential extrema, equilibrium sheet section, and marked control point. Try $(-1,0)$, $(-1,0.25)$, $(-1,0.5)$, and $(0.4,0.2)$. Filled circles/solid branches indicate stable roots; crosses/dashed branches indicate unstable roots. The fold check uses the analytic parametrization, avoiding unreliable counting of nearly repeated numerical roots.


```python
# gradient_cusp: dimensionless canonical controls; edit these and rerun.
a_explore, b_explore = -1.0, 0.25
def cusp_potential(x, a, b):
    return x**4/4 + a*x**2/2 + b*x

roots = np.roots([1.0, 0.0, a_explore, b_explore])
real_roots = np.sort(roots.real[np.abs(roots.imag) < 1e-7])
delta = -4*a_explore**3-27*b_explore**2
print('Delta =', delta, '; real roots (numerical) =', real_roots)
if abs(delta) > 1e-6:
    assert len(real_roots) == (3 if delta > 0 else 1)
assert np.allclose(real_roots**3+a_explore*real_roots+b_explore, 0, atol=1e-7)

fig, axes = plt.subplots(1, 3, figsize=(15, 4.2), constrained_layout=True)
x_view = np.linspace(-2.0, 2.0, 700)
axes[0].plot(x_view, cusp_potential(x_view, a_explore, b_explore))
for root in real_roots:
    curvature = 3*root**2+a_explore
    marker = 'o' if curvature > 1e-6 else ('x' if curvature < -1e-6 else 's')
    axes[0].plot(root, cusp_potential(root, a_explore, b_explore),
                 marker=marker, color='C0' if curvature > 1e-6 else 'black', ms=7)
axes[0].set(xlabel=r'$x$', ylabel=r'$U$', title='Tilted potential')

# Parametrize the equilibrium section by x: b = -x^3-a*x.
curvature = 3*x_view**2+a_explore
b_section = -x_view**3-a_explore*x_view
axes[1].plot(np.where(curvature > 0, b_section, np.nan), x_view,
             color='C0', label='stable')
axes[1].plot(np.where(curvature < 0, b_section, np.nan), x_view,
             'k--', label='unstable')
axes[1].axvline(b_explore, color='C1', lw=1)
axes[1].set(xlabel=r'$b$', ylabel=r'$x_s$', xlim=(-0.8, 0.8),
            title=fr'Equilibria at $a={a_explore:g}$')
axes[1].legend()

fold_x = np.linspace(-0.85, 0.85, 801)
fold_a, fold_b = -3*fold_x**2, 2*fold_x**3
assert np.allclose(fold_x**3+fold_a*fold_x+fold_b, 0)
assert np.allclose(3*fold_x**2+fold_a, 0)
assert np.allclose(4*fold_a**3+27*fold_b**2, 0)
a_region = np.linspace(-2.2, 0, 400)
b_limit = 2*(-a_region/3)**1.5
axes[2].fill_between(a_region, -b_limit, b_limit, color='C0', alpha=0.15,
                     label='3 equilibria')
axes[2].plot(fold_a, fold_b, 'k-', label='fold set')
axes[2].plot(a_explore, b_explore, 'o', color='C1', label='chosen controls')
axes[2].set(xlabel=r'$a$', ylabel=r'$b$', title='Cusp control plane')
axes[2].legend(fontsize=9)
plt.show()
```

### What the landscape does and does not imply

For any smooth **autonomous scalar** equation $\dot E=f(E)$ on an interval, $V(E)=-\int^E f(u)\,du$ gives a unit-mobility gradient representation. This elementary construction does not require thermal equilibrium. If controls vary in time, the extra term $\partial_tV$ means that the instantaneous potential need not decrease: for the biased normal form,

$$\frac{dV_h}{dt}=-(\partial_E V_h)^2-\frac{\dot\mu}{2}E^2
+\frac{\dot\beta}{4}E^4-\dot h E.$$

In several dimensions, a Euclidean representation $\dot{\mathbf q}=\mathbf F=-\nabla V$ requires a symmetric Jacobian, $\partial_jF_i=\partial_iF_j$, and suitable domain topology (a simply connected domain makes the curl-free condition sufficient). The full real laser model generally fails this test: $\partial_P\dot E=g$ while $\partial_E\dot P=gD$. A scalar potential for its reduced amplitude therefore does not establish a Euclidean gradient potential for $(E,P,D)$, or exclude oscillations and chaos outside the scalar regime.

A system $\dot{\mathbf q}=-\mathbf M(\mathbf q)\nabla V$ with symmetric positive-definite mobility has $\dot V=-\nabla V^T\mathbf M\nabla V\le0$, but existence of such a representation needs separate justification. A Lyapunov function alone does not imply pure gradient dynamics; circulation can coexist with decay. For a complex amplitude $\mathsf E=X+iY$, the special real-coefficient equation $\dot{\mathsf E}=(\mu-\beta|\mathsf E|^2)\mathsf E$ is gradient in $(X,Y)$ with radial potential $-\mu(X^2+Y^2)/2+\beta(X^2+Y^2)^2/4$. Above threshold it has a circle of minima with neutral phase. Detuning or imaginary coefficients can introduce phase rotation, so this special result cannot be assumed for the general laser equations.
