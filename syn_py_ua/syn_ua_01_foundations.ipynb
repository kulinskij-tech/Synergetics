## Необхідний вступ: навіщо лазеру поріг і зворотний зв'язок

Збуджений атом може випромінити фотон спонтанно, з неконтрольованою фазою, або здійснити вимушений перехід у вже заселену електромагнітну моду. **Інверсія населеностей** означає, що завдяки накачуванню в активному середовищі більше атомів готові до вимушеного випромінювання, ніж до поглинання. Резонатор багаторазово повертає вибрані моди поля в активне середовище, але дзеркала й саме середовище водночас спричиняють втрати.

Нижче порога втрати переважають, тому фазово неузгоджене спонтанне випромінювання залишається слабким. На порозі підсилення однієї моди зрівнюється з втратами. Вище порога поле стимулює подальше випромінювання у власній фазі та просторовій моді, а виснаження інверсії обмежує зростання. Отже, когерентність виникає завдяки нелінійному зворотному зв'язку між полем і речовиною. Змінні Максвелла—Блоха нижче є колективними описами: $E$ — амплітуда поля, $P$ — поляризація середовища (фазово узгоджений дипольний відгук), $D$ — інверсія населеностей, тобто запас підсилення.


### Населеності, інверсія та поляризація речовини

Розглянемо $N$ ефективних дворівневих випромінювачів із номерами $j$, верхнім станом $|e\rangle$ і нижнім станом $|g\rangle$. **Населеність рівня** — середнє число випромінювачів, які перебувають на цьому енергетичному рівні. Для випромінювача $j$ позначимо через $p_{e,j}$ та $p_{g,j}$ імовірності перебування на верхньому та нижньому рівнях. У межах ефективної дворівневої моделі

$$p_{e,j}+p_{g,j}=1,\qquad N_e=\sum_j p_{e,j},\qquad N_g=\sum_j p_{g,j},\qquad N_e+N_g=N.$$

**Інверсія населеностей** — різниця населеностей:

$$D=N_e-N_g,\qquad -N\le D\le N,\qquad N_e=\frac{N+D}{2},\qquad N_g=\frac{N-D}{2}.$$

Отже, $D>0$ означає більшу населеність верхнього лазерного рівня. Змінна $D$ у цьому ноутбуці є сумарною інверсією врахованих випромінювачів. Інверсія на один випромінювач дорівнює $D/N$, а на одиницю об’єму — $D/V_g$, де $V_g$ — об’єм активного середовища. $D_0$ — інверсія, яку підтримують накачування та релаксація за відсутності поля резонатора; $D_c$ — її порогове значення для генерації.

Для квантового стану з оператором густини $\rho$ середнє визначається як $\langle O\rangle=\operatorname{Tr}(\rho O)$; стан і оператор мають бути записані в одному представленні. Редукована матриця густини $\rho_j$ описує випромінювач $j$ після взяття часткового сліду за полем та іншими випромінювачами. В упорядкованому базисі $(|e\rangle,|g\rangle)$

$$\rho_j=\begin{pmatrix}p_{e,j}&c_j\\c_j^*&p_{g,j}\end{pmatrix},\qquad
c_j=\langle\sigma_j^-\rangle,\qquad
\langle\sigma_j^z\rangle=p_{e,j}-p_{g,j}.$$

Тут $\sigma_j^-=|g\rangle_j\langle e|$ переводить випромінювач із верхнього рівня на нижній, $\sigma_j^+=(\sigma_j^-)^\dagger$ виконує зворотний перехід, а $\sigma_j^z=|e\rangle_j\langle e|-|g\rangle_j\langle g|$ вимірює його інверсію. Позадіагональний елемент $c_j$ — **атомна когерентність**: його модуль і фаза визначають осцилюючий середній дипольний момент переходу. Для діагональної суміші $c_j=0$, навіть якщо $D>0$. Наприклад, стан $(|e\rangle+e^{i\phi}|g\rangle)/\sqrt2$ має однакові населеності та $c_j=e^{-i\phi}/2$.

**Поляризація речовини** — електричний дипольний момент одиниці об’єму. Для однорідного активного середовища з однаковими дипольними моментами переходу лабораторну поляризацію задають наведені нижче вирази; тут середні обчислюємо з лабораторним оператором густини, а оператори дипольного моменту й переходів записано у представленні Шредінґера:

$$\mathbf d_{ge}=\langle g|\widehat{\mathbf d}|e\rangle,\qquad
\widehat{\mathbf d}_j=\mathbf d_{ge}\sigma_j^-+\mathbf d_{ge}^*\sigma_j^+,\qquad
\mathbf P_{\rm mat}=\frac1{V_g}\sum_j\langle\widehat{\mathbf d}_j\rangle.$$

Тут враховуємо внесок оптичного переходу в дипольний момент. Когерентності з узгодженими фазами додаються й утворюють макроскопічну поляризацію; за різних фаз внески можуть взаємно компенсуватися. Колективна когерентність дорівнює $s=\sum_jc_j$. Уведена далі скалярна змінна $P$ представляє масштабовану амплітуду компоненти, зв’язаної з вибраною модою резонатора. Її нормування задаємо під час напівкласичної редукції; фізична $\mathbf P_{\rm mat}$ має одиниці СІ Кл/м$^2$.

Оператор $a$ знищує фотон у вибраній моді резонатора, а $a^\dagger$ створює його. Аналогічно, $\alpha=\langle a\rangle$ — когерентна амплітуда поля за вибраної опорної фази. Середнє число фотонів дорівнює $n_{\rm ph}=\langle a^\dagger a\rangle$, а $n_{\rm coh}=|\alpha|^2$ визначає його когерентний внесок. Їхня різниця дорівнює $\langle(a-\alpha)^\dagger(a-\alpha)\rangle\ge0$. Наведена нижче скалярна величина $E$ є масштабованою амплітудою поля; на графіках $E^2$ представляє масштабовану інтенсивність.


## Лазер як прототип синергетичної системи

Синергетика Гакена виросла з теорії лазера. Активне середовище містить багато мікроскопічних випромінювачів, чиї спонтанні фази спочатку не узгоджені. У резонаторі вимушене випромінювання та взаємний вплив поля й речовини створюють поріг: одна мода може набути стійкої макроскопічної амплітуди й фазової узгодженості. Цей порядок не задають ззовні для кожного атома — він постає як колективна нестійкість відкритої системи з підведенням і дисипацією енергії.

У дійсному одномодовому напівкласичному наближенні позначимо через $E$ амплітуду поля резонатора, через $P$ — колективну поляризацію, а через $D$ — інверсію населеностей:

$$\dot E=-\kappa E+gP,$$
$$\dot P=-\gamma_\perp P+gED,$$
$$\dot D=\gamma_\parallel(D_0-D)-2gEP.$$

Параметр $D_0$ задається накачуванням, тоді як $\kappa$, $\gamma_\perp$ і $\gamma_\parallel$ є швидкостями релаксації поля, поляризації та інверсії. Лінеаризація підсистеми $(E,P)$ поблизу безлазерного стану $(E,P,D)=(0,0,D_0)$ дає поріг

$$D_c=\frac{\kappa\gamma_\perp}{g^2}.$$

За $D_0<D_c$ флуктуації поля згасають. За $D_0>D_c$ безлазерний стан стає нестійким, а нелінійне виснаження інверсії насичує когерентне поле.


### Від квантових випромінювачів до колективних змінних лазера

Три наведені вище рівняння спираються на кілька фізичних наближень. Перш ніж переходити до числового розрахунку, простежимо їхнє походження. Наше завдання — зрозуміти, як мікроскопічні випромінювачі та спільне поле резонатора приводять до колективних змінних. Поки що ми не виключаємо швидкі змінні: принцип підпорядкування становить наступний крок редукції.

Розглянемо $N$ однакових ефективних дворівневих випромінювачів із верхнім і нижнім лазерними рівнями $|e\rangle$ та $|g\rangle$. Введемо оператори

$$\sigma_j^-=|g\rangle_j\langle e|,\qquad
\sigma_j^+=(\sigma_j^-)^\dagger,\qquad
\sigma_j^z=|e\rangle_j\langle e|-|g\rangle_j\langle g|.$$

Використовуємо три кутові частоти: $\omega_c$ — резонансна частота вибраної моди порожнього резонатора, а $\omega_a=(E_e-E_g)/\hbar$ — частота переходу кожного випромінювача. Опорна частота $\omega_r$ визначає оптичний фазовий множник $e^{-i\omega_r t}$, який виділяємо з оператора поля Гайзенберґа для означення його амплітуди. Цю частоту вибираємо довільно, зазвичай поблизу резонансних частот резонатора й атомів. Вибір $\omega_r=\omega_c$ дає нульову розстройку резонатора $\Delta_c=\omega_c-\omega_r=0$; ця розстройка та атомна розстройка $\Delta_a=\omega_a-\omega_r$ дорівнюють нулю за $\omega_r=\omega_c=\omega_a$. Усі три величини є кутовими частотами й вимірюються в радіанах за секунду.

Оператор пониження описує перехід атома з верхнього рівня на нижній; власні значення $\sigma_j^z$ дорівнюють $+1$ і $-1$. Для однієї моди резонатора $a_H(t)$ — лабораторний оператор знищення фотона у представленні Гайзенберґа, причому $[a_H(t),a_H^\dagger(t)]=1$. Додатночастотну частину оператора електричного поля в тому самому представленні подамо як

$$\widehat{\mathcal E}^{(+)}_H(\mathbf r,t)
=\mathcal E_{\rm vac}u(\mathbf r)a_H(t).$$

Отже, **одномодовість** означає, що ми залишаємо один просторовий профіль $u(\mathbf r)$ та його залежну від часу амплітуду. Ця амплітуда не обов'язково дійсна. Припустимо також, що всі випромінювачі мають однакову частоту переходу й однакову дійсну сталу зв'язку $g$ із модою. Тим самим нехтуємо неоднорідним розширенням спектральної лінії та просторовою залежністю зв'язку.

Тут $u(\mathbf r)$ — безрозмірний просторовий профіль моди, нормування якого враховано у $\mathcal E_{\rm vac}$ — масштабі вакуумного електричного поля, визначеному вибраним нормуванням моди (одиниці В/м). Індекс $(+)$ позначає додатночастотну компоненту; ермітів оператор електричного поля включає також її спряження. Дійсна стала $g$ визначає силу дипольного зв’язку з модою в одиницях кутової частоти; її знак можна врахувати вибором фази.

**Представлення операторів.** Індекс $H$ позначає лабораторні оператори Гайзенберґа. Тильдою позначаємо кожен оператор після виділення опорної оптичної фази:

$$\widetilde a(t)=e^{i\omega_r t}a_H(t),\qquad
\widetilde\sigma^-_j(t)=e^{i\omega_r t}\sigma^-_{j,H}(t),\qquad
\widetilde\sigma^z_j(t)=\sigma^z_{j,H}(t).$$

Ермітово спряжені оператори набувають протилежної фази: $\widetilde a^\dagger=e^{-i\omega_r t}a_H^\dagger$ та $\widetilde\sigma_j^+=e^{-i\omega_r t}\sigma_{j,H}^+$. Величини з тильдами залишаються операторами. Лабораторне поле відновлюється як

$$\widehat{\mathcal E}^{(+)}_H(\mathbf r,t)
=\mathcal E_{\rm vac}u(\mathbf r)a_H(t)
=\mathcal E_{\rm vac}u(\mathbf r)\widetilde a(t)e^{-i\omega_r t}.$$

Для вільної моди резонатора $\dot a_H=-i\omega_c a_H$, тому диференціювання означення дає

$$\frac{d\widetilde a}{dt}=i\omega_r\widetilde a+e^{i\omega_r t}\dot a_H
=-i(\omega_c-\omega_r)\widetilde a.$$

Щоб визначити представлення динаміки, використаємо незалежні від часу оператори Шредінґера $a_S,\sigma_{j,S}$:

$$H_{\rm ref}=\hbar\omega_r\mathcal N,\qquad
\mathcal N=a_S^\dagger a_S+\frac12\sum_j(\sigma_{j,S}^z+1).$$

Тут $\mathcal N$ враховує сумарне число фотонів і збуджених випромінювачів. Для замкненої гамільтонової частини перетворення стану $|\psi_{\rm r}\rangle=e^{iH_{\rm ref}t/\hbar}|\psi_S\rangle$ задає представлення взаємодії відносно $H_{\rm ref}$. Його гамільтоніан дорівнює

$$H_{\rm r}=e^{iH_{\rm ref}t/\hbar}H_{\rm lab}e^{-iH_{\rm ref}t/\hbar}-H_{\rm ref}
=H_{\rm lab}-H_{\rm ref},$$

де остання рівність використовує $[H_{\rm lab},\mathcal N]=0$ у наближенні обертової хвилі. Оператори з тильдами задовольняють рівняння Гайзенберґа з гамільтоніаном $H_{\rm r}$. Усталений вислів «обертова система відліку» означає цю зміну оптичної фази. Наближення обертової хвилі усуває контробертові доданки взаємодії.

Адитивна скалярна стала в гамільтоніані не впливає на ці рівняння; далі її опускаємо.

**Подальші позначення:** $a\equiv\widetilde a$, $\sigma_j^{\pm,z}\equiv\widetilde\sigma_j^{\pm,z}$ та $H\equiv H_{\rm r}$. Отже, оператори поля й атомів мають однаково вибрану опорну фазу. У наведених далі рівняннях для оператора густини $\rho$ позначає стан у цьому перетвореному представленні, а оператори без додаткових позначок під знаком сліду — його спостережувані у представленні Шредінґера. Їхні середні дорівнюють середнім відповідних операторів Гайзенберґа з тильдами, обчисленим із початковим станом.

В електродипольному наближенні та наближенні обертової хвилі гамільтоніан має вигляд

$$\frac{H}{\hbar}=\Delta_c a^\dagger a
+\frac{\Delta_a}{2}\sum_j\sigma_j^z
+g\sum_j(a^\dagger\sigma_j^-+a\sigma_j^+),$$

Тут $\Delta_c=\omega_c-\omega_r$, а $\Delta_a=\omega_a-\omega_r$. Доданок $a^\dagger\sigma_j^-$ описує народження фотона під час переходу випромінювача на нижній рівень; ермітово спряжений доданок описує поглинання. У наближенні обертової хвилі відкидаємо швидко осцилюючі доданки, які одночасно створюють або знищують і фотонне, і атомне збудження.

На відміну від безрозмірного числового прикладу, у цій мікроскопічній моделі спочатку використовуємо фізичний час: $g$, розстройки частот і швидкості релаксації мають розмірність оберненого часу. Вибір одиниці часу дає змогу перейти до безрозмірних параметрів.


### Накачування, втрати та квантове керувальне рівняння

Самого гамільтоніана недостатньо для опису працюючого лазера: накачування підводить енергію, а втрати її відводять. Найпростіша марковська модель взаємодії з резервуарами задається рівнянням

$$\dot\rho=-\frac{i}{\hbar}[H,\rho]+2\kappa\mathcal D[a]\rho
+\sum_j\left(w\mathcal D[\sigma_j^+]\rho
+\Gamma\mathcal D[\sigma_j^-]\rho
+\frac{\gamma_\phi}{2}\mathcal D[\sigma_j^z]\rho\right),$$

$$\mathcal D[L]\rho=L\rho L^\dagger-\frac12\{L^\dagger L,\rho\}.$$

Тут $\rho$ — оператор густини спільної системи атомів і поля. Дисипатор $\mathcal D[L]$ описує незворотний процес, пов'язаний з оператором $L$: $w$ — ефективна швидкість некогерентного накачування, $\Gamma$ — швидкість атомного розпаду через канали, відмінні від явно врахованої моди резонатора, а $\gamma_\phi$ характеризує втрату атомної фазової когерентності без зміни населеностей. Ефективне накачування моделює дію резервуара або допоміжних рівнів багаторівневого активного середовища, а не резонансне когерентне збудження ізольованого дворівневого атома.

Множник $2\kappa$ фіксує домовленість: $\kappa$ є швидкістю згасання **амплітуди поля**. Для порожнього резонатора

$$\frac{d}{dt}\langle a\rangle=-\kappa\langle a\rangle,
\qquad\frac{d}{dt}\langle a^\dagger a\rangle=-2\kappa\langle a^\dagger a\rangle.$$

Для випромінювача, не зв'язаного з полем резонатора, інверсія $z=\langle\sigma^z\rangle$ задовольняє рівняння

$$\dot z=w(1-z)-\Gamma(1+z)
=-(w+\Gamma)\left(z-\frac{w-\Gamma}{w+\Gamma}\right).$$

Отже, у цій моделі резервуарів

$$\gamma_\parallel=w+\Gamma,\qquad
\gamma_\perp=\frac{w+\Gamma}{2}+\gamma_\phi,\qquad
D_0=N\frac{w-\Gamma}{w+\Gamma}.$$

За відсутності поля резонатора накачування й атомний розпад спрямовують сумарну інверсію до $D_0$. Інверсія виникає за $w>\Gamma$. У докладніших багаторівневих моделях ефективні співвідношення можуть бути іншими. У феноменологічних розрахунках нижче ми змінюємо $D_0$, залишаючи швидкості релаксації сталими. Такий розрахунок не відповідає буквальній зміні лише $w$ у щойно записаній моделі резервуарів.


### Рівняння для середніх і проблема замикання

Введемо колективні оператори

$$S^-=\sum_j\sigma_j^-,\qquad S^+=(S^-)^\dagger,\qquad Z=\sum_j\sigma_j^z.$$

Середнє значення спостережуваної визначається як $\langle O\rangle=\operatorname{Tr}(\rho O)$. Його еволюцію знаходимо з рівняння

$$\frac{d}{dt}\langle O\rangle=\frac{i}{\hbar}\langle[H,O]\rangle
+\text{внески резервуарів}.$$

Використовуючи $[\sigma^+,\sigma^-]=\sigma^z$ та $[\sigma^z,\sigma^\pm]=\pm2\sigma^\pm$, дістаємо

$$\begin{aligned}
\frac{d}{dt}\langle a\rangle
&=-(\kappa+i\Delta_c)\langle a\rangle-ig\langle S^-\rangle,\\
\frac{d}{dt}\langle S^-\rangle
&=-(\gamma_\perp+i\Delta_a)\langle S^-\rangle+ig\langle aZ\rangle,\\
\frac{d}{dt}\langle Z\rangle
&=\gamma_\parallel(D_0-\langle Z\rangle)
+2ig(\langle a^\dagger S^-\rangle-\langle aS^+\rangle).
\end{aligned}$$

Наприклад, комутатор $[a^\dagger S^-+aS^+,S^-]=aZ$ породжує добуток поля та інверсії в рівнянні для поляризації. У межах прийнятої квантової моделі ці рівняння точні. Проте вони не замкнені: середні від добутків, як-от $\langle aZ\rangle$, загалом не дорівнюють добуткам середніх. Рівняння для цих кореляцій містили б наступні моменти.


### Напівкласична факторизація та нормування

Позначимо $\alpha=\langle a\rangle$, $s=\langle S^-\rangle$ і $D=\langle Z\rangle$. У напівкласичному наближенні середнього поля нехтуємо кореляційними внесками між атомними змінними та полем:

$$\langle aZ\rangle\simeq\alpha D,\qquad
\langle a^\dagger S^-\rangle\simeq\alpha^*s,\qquad
\langle aS^+\rangle\simeq\alpha s^*.$$

Наприклад, $\langle aZ\rangle=\alpha D+\langle(a-\alpha)(Z-D)\rangle$, і ми відкидаємо другий доданок. Тоді

$$\begin{aligned}
\dot\alpha&=-(\kappa+i\Delta_c)\alpha-igs,\\
\dot s&=-(\gamma_\perp+i\Delta_a)s+ig\alpha D,\\
\dot D&=\gamma_\parallel(D_0-D)+2ig(\alpha^*s-\alpha s^*).
\end{aligned}$$

Введемо нормовані комплексні амплітуди

$$\mathsf E=\sqrt2\,\alpha,\qquad\mathsf P=-i\sqrt2\,s.$$

Фазовий множник поглинає уявну одиницю в членах взаємодії, а числові множники забезпечують саме ті коефіцієнти, які використано в цьому модулі. Одержуємо комплексні рівняння Максвелла—Блоха:

$$\boxed{\begin{aligned}
\dot{\mathsf E}&=-(\kappa+i\Delta_c)\mathsf E+g\mathsf P,\\
\dot{\mathsf P}&=-(\gamma_\perp+i\Delta_a)\mathsf P+g\mathsf E D,\\
\dot D&=\gamma_\parallel(D_0-D)-2g\operatorname{Re}(\mathsf E^*\mathsf P).
\end{aligned}}$$

Поляризація описує колективну атомну когерентність, а не населеність збудженого рівня; населеності входять через $D$. На цьому етапі модель містить п'ять дійсних змінних: по дві компоненти поля й поляризації та одну інверсію. Факторизація не враховує квантові кореляції та динаміку флуктуацій; її точність не гарантована як завгодно близько до порога або за малої кількості випромінювачів і фотонів.


### Резонанс, фіксована фаза і три дійсні рівняння

Припустимо точний резонанс між модою резонатора й атомним переходом та виберемо систему відліку, що обертається з їхньою спільною частотою: $\Delta_c=\Delta_a=0$. Комплексні рівняння зберігають підпростір

$$\mathsf E=e^{i\theta}E,\qquad\mathsf P=e^{i\theta}P,\qquad E,P\in\mathbb R,$$

де $\theta$ стала. Вибравши цю спільну фазу за початок відліку, дістаємо саме

$$\boxed{\begin{aligned}
\dot E&=-\kappa E+gP,\\
\dot P&=-\gamma_\perp P+gED,\\
\dot D&=\gamma_\parallel(D_0-D)-2gEP.
\end{aligned}}$$

Це підпростір резонансної моделі з фіксованою фазою. Вибір одного спільного початку відліку фаз загалом не дозволяє одночасно зробити дійсними дві комплексні амплітуди з довільними фазами; за розстройки частот і для загальної динаміки відносної фази потрібні комплексні рівняння.

Баланс збуджень пояснює множник $2$. Когерентна складова числа фотонів дорівнює $n_{\rm coh}=|\alpha|^2=E^2/2$, а середня населеність верхнього рівня — $N_e=(N+D)/2$. Залишаючи лише внески взаємодії, маємо

$$\left.\dot n_{\rm coh}\right|_{\rm int}=gEP,\qquad
\left.\dot N_e\right|_{\rm int}=-gEP.$$

Один перехід на нижній рівень додає один фотон, але зменшує $D=N_e-N_g$ на два. На графіках нижче величина $E^2$ є нормованою інтенсивністю, пропорційною числу фотонів у когерентній складовій. У повному квантовому описі $\langle a^\dagger a\rangle$ не обов'язково дорівнює $|\langle a\rangle|^2$.


### Від колективної нестійкості до параметра порядку

Лінеаризація поблизу $(E,P,D)=(0,0,D_0)$ дає $\delta\dot D=-\gamma_\parallel\delta D$ та

$$\frac{d}{dt}\begin{pmatrix}\delta E\\\delta P\end{pmatrix}
=\begin{pmatrix}-\kappa&g\\gD_0&-\gamma_\perp\end{pmatrix}
\begin{pmatrix}\delta E\\\delta P\end{pmatrix}.$$

Два показники зростання визначаються рівнянням

$$\lambda^2+(\kappa+\gamma_\perp)\lambda
+\kappa\gamma_\perp-g^2D_0=0.$$

Слід матриці від'ємний, а її визначник змінює знак за $D_c=\kappa\gamma_\perp/g^2$. Отже, на порозі колективне збурення поля й поляризації переходить від згасання до зростання. За $D_0=D_c$ одне власне значення дорівнює нулю, а друге — $-(\kappa+\gamma_\perp)$; критичний власний вектор задовольняє $\delta P=(\kappa/g)\delta E$. Тому повільна змінна описує зв'язану моду поля й речовини, яку можна параметризувати амплітудою $E$.

Для ненульового стаціонарного поля перші два рівняння дають

$$P_s=\frac{\kappa}{g}E_s,\qquad D_s=D_c.$$

Із третього рівняння тоді випливає

$$E_s^2=\frac{\gamma_\parallel}{2\kappa}(D_0-D_c).$$

Поле виснажує інверсію доти, доки підсилення не зрівняється з втратами. Саме цю стаціонарну гілку зображено нижче; її стійкість далеко вище порога потребує окремого аналізу. Поле виникає завдяки колективній поляризації й водночас змінює середовище, що його породжує: у цьому полягає циклічна причинність у трактуванні Гакена.

Детерміністичні рівняння залишають стан із точно нульовими $E=P=0$ незмінним навіть вище порога. Мала початкова амплітуда в числовому прикладі задає збурення; фізичний механізм запуску забезпечують квантові флуктуації. У скінченному квантовому лазері без зовнішнього сигналу, що задає фазу, ансамблеве середнє $\langle a\rangle$ може залишатися нульовим навіть за ненульового числа фотонів. Ненульова напівкласична амплітуда описує вибрану фазу або окрему реалізацію на проміжках часу, коли дифузією фази можна знехтувати.

У цьому виведенні ми зберегли всі три змінні поля й середовища. Подальше наближення підпорядкування потребує додаткового обґрунтування через розділення часових масштабів: поблизу порога швидкість еволюції критичної колективної моди стає малою порівняно зі швидкостями релаксації некритичних мод. Саме по собі напівкласичне наближення ще не дозволяє покласти $\dot P$ або $\dot D$ рівними нулю.

**Додаткове читання.** G. J. de Valcárcel, E. Roldán, F. Prati, [*Semiclassical Theory of Amplification and Lasing*](https://arxiv.org/abs/quant-ph/0605084): виведення рівнянь Максвелла—Блоха та їхній зв'язок із багаторівневими активними середовищами. Інтерпретацію через нестійкість, параметри порядку та принцип підпорядкування розглянуто в основному джерелі курсу — *Synergetics* Гакена.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "axes.grid": True,
    "grid.alpha": 0.25,
    "font.size": 11,
})
rng = np.random.default_rng(20260904)

from scipy.integrate import solve_ivp

kap, gamma_perp, gamma_parallel, g = 1.0, 8.0, 0.7, 2.0
D_c = kap*gamma_perp/g**2

def laser(t, state, D0):
    E, P, D = state
    return (-kap*E + g*P,
            -gamma_perp*P + g*E*D,
            gamma_parallel*(D0-D) - 2*g*E*P)

t_eval = np.linspace(0, 60, 4000)
pumps = [0.7*D_c, 1.02*D_c, 1.6*D_c]

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
steady_intensity = []
for D0 in pumps:
    sol = solve_ivp(laser, (0, 60), [1e-4, 0.0, D0], args=(D0,), t_eval=t_eval,
                    rtol=1e-9, atol=1e-11)
    E, P, D = sol.y
    ax[0].plot(sol.t, E**2, label=fr"$D_0/D_c={D0/D_c:.2f}$")
    steady_intensity.append(np.mean(E[sol.t > 50]**2))
ax[0].set(xlabel="time", ylabel=r"intensity $E^2$", title="Laser threshold dynamics")
ax[0].legend()

pump_scan = np.linspace(0.4*D_c, 2.0*D_c, 60)
theory = gamma_parallel*np.maximum(pump_scan-D_c, 0)/(2*kap)
ax[1].plot(pump_scan/D_c, theory, lw=2)
ax[1].set(xlabel=r"pump $D_0/D_c$", ylabel=r"stationary $E^2$",
          title="Order parameter above threshold")
plt.tight_layout()
```

## Принцип підпорядкування і циклічна причинність

Якщо матеріальні змінні релаксують швидше за поле, покладемо $\dot P\approx0$ і $\dot D\approx0$:

$$P\approx\frac{g}{\gamma_\perp}ED,\qquad
D\approx\frac{D_0}{1+2g^2E^2/(\gamma_\perp\gamma_\parallel)}.$$

Підстановка в рівняння поля та розклад поблизу $E=0$ дають

$$\dot E\approx \mu E-\beta E^3,\qquad
\mu=\kappa\left(\frac{D_0}{D_c}-1\right),\quad \beta>0.$$

Когерентне поле $E$ є повільним параметром порядку. Швидкі атомні змінні наближено стають його функціями — вони **підпорядковуються** полю. Водночас саме атоми породжують поле через вимушену поляризацію, а поле узгоджує атоми. Для Гакена цей взаємний зв'язок був взірцем циклічної причинності: мікроскопічна активність створює макроскопічний параметр порядку, який у відповідь обмежує мікроскопічну динаміку.


```python
# Test the slaving relation on the above-threshold trajectory.
D0 = 1.6*D_c
sol = solve_ivp(laser, (0, 60), [1e-4, 0.0, D0], args=(D0,), t_eval=t_eval,
                rtol=1e-9, atol=1e-11)
E, P, D = sol.y
P_slaved = g*E*D/gamma_perp
D_slaved = D0/(1 + 2*g**2*E**2/(gamma_perp*gamma_parallel))
mask = sol.t > 10

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].plot(sol.t, P, label="full P")
ax[0].plot(sol.t, P_slaved, "--", label="slaved P(E,D)")
ax[0].set(xlabel="time", ylabel="polarization", title="Fast polarization follows the field")
ax[0].legend()
ax[1].plot(E[mask]**2, D[mask], ".", ms=1.5, label="full dynamics")
order = np.argsort(E[mask]**2)
ax[1].plot(E[mask][order]**2, D_slaved[mask][order], "k--", label="slaved inversion")
ax[1].set(xlabel=r"intensity $E^2$", ylabel="inversion D", title="Gain saturation")
ax[1].legend()
plt.tight_layout()
```

## Градієнтна динаміка параметра порядку: від синергетики до теорії катастроф

Наведене вище підпорядкування звело систему «поле–речовина» до однієї **дійсної амплітуди зі знаком** $E$. Два знаки відповідають фазам, що відрізняються на $\pi$, у резонансному секторі з фіксованою фазою; інтенсивність дорівнює $E^2$. Тепер те саме рівняння амплітуди можна розглядати як рух униз потенціальним ландшафтом. Так параметр порядку Гакена пов'язується з [модулем 04: потенціальні ландшафти й теорія катастроф](https://colab.research.google.com/github/kulinskij-tech/Synergetics/blob/main/syn_py_ua/syn_ua_04_catastrophe_theory.ipynb).

### Збережімо коефіцієнти конкретного лазерного наближення

Позначимо $c=2g^2/(\gamma_\perp\gamma_\parallel)$ та $A=g^2D_0/\gamma_\perp=\kappa D_0/D_c$. Підстановка наведених співвідношень підпорядкування дає **в межах квазістатичного наближення**

$$\dot E=\left[-\kappa+\frac{A}{1+cE^2}\right]E
=(A-\kappa)E-AcE^3+O(E^5).$$

Отже, коефіцієнти рівняння цього ноутбука мають вигляд

$$\mu=\kappa\left(\frac{D_0}{D_c}-1\right),\qquad
\beta(D_0)=Ac=\frac{2g^4D_0}{\gamma_\perp^2\gamma_\parallel}>0.$$

У провідному наближенні поблизу порога можна взяти сталу

$$\beta_c=\beta(D_c)=\frac{2\kappa g^2}{\gamma_\perp\gamma_\parallel}
=\frac{2\kappa^2}{\gamma_\parallel D_c},$$

оскільки $\beta(D_0)-\beta_c=O(\mu)$, а на новій гілці $E=O(\sqrt{\mu})$. З $\beta_c$ отримуємо $E_s^2=\mu/\beta_c=\gamma_\parallel(D_0-D_c)/(2\kappa)$: це провідний закон виникнення генерації (і точна стаціонарна формула повної моделі). З $\beta(D_0)$ кубічне наближення дає $E_s^2=\gamma_\parallel D_c(D_0-D_c)/(2\kappa D_0)$, що відрізняється в наступних порядках. Потрібно дотримуватися $cE^2\ll1$ і залишатися поблизу порога; жоден із поліномів не є глобальною моделлю лазера.

**Уточнення часових масштабів.** Мала швидкість колективного зростання сама собою не робить $\dot P=\dot D=0$ точним зведенням у фізичному часі. За скінченного $\gamma_\perp/\kappa$ повний якобіан безлазерного стану дає $\lambda_{\rm slow}\simeq M\mu$, де $M=\gamma_\perp/(\gamma_\perp+\kappa)$. Провідне рівняння на центральному многовиді в координаті поля має вигляд $\dot E=M(\mu E-\beta_cE^3)$ (члени вищих порядків опущено). Квазістатичні коефіцієнти ноутбука застосовні, коли поляризація достатньо швидко слідує за полем, або в перемасштабованому часі $s=Mt$ у цьому порядку. Додатний множник змінює час релаксації, але не рівноваги та їхню стійкість. На графіках нижче явно використано квазістатичну нормальну форму.

### Інтегруємо силу й отримуємо потенціал четвертого степеня

За фіксованих керувальних параметрів і вибраного додатного $\beta$ інтегруємо $V_E=-\mu E+\beta E^3$:

$$V(E;\mu)= -\frac{\mu}{2}E^2+\frac{\beta}{4}E^4+C,
\qquad \dot E=-\frac{\partial V}{\partial E}.$$

Множники $1/2$ і $1/4$ відтворюють правильну швидкість зміни $E$. Довільна стала $C$ не впливає на рух. Незалежна перевірка: нерозкладеному квазістатичному рівнянню відповідає $V_{\rm qs}=\kappa E^2/2-[A/(2c)]\ln(1+cE^2)$; його розклад відтворює саме наведені коефіцієнти при другому й четвертому степенях. Це **ефективний динамічний потенціал**, з одиницями $E^2/\text{час}$ за одиничної рухливості, а не встановлена термодинамічна вільна енергія лазера з накачуванням.

Уздовж траєкторії за сталих $\mu,\beta$

$$\frac{dV}{dt}=V_E\dot E=-(V_E)^2=-(\dot E)^2\le0.$$

Рівність виконується саме в станах рівноваги. Для $\dot E=-M V_E$ зі сталою $M>0$ маємо $dV/dt=-M(V_E)^2$. Оскільки $\beta>0$, потенціал четвертого степеня обмежує рух цієї скалярної моделі; несталих періодичних орбіт у ній немає. Властивість Ляпунова не означає монотонного спадання фізичної енергії відкритої системи з підведенням енергії.


### Рівноваги вилкової біфуркації та їхня стійкість

Позначивши $f(E)=\mu E-\beta E^3$, знаходимо рівноваги з $E(\mu-\beta E^2)=0$; збурення задовольняють $\delta\dot E=f'(E_s)\delta E=-V''(E_s)\delta E$.

| Керувальний параметр | Рівноваги | Локальна поведінка |
|---|---|---|
| $\mu<0$ | $E_s=0$ | $f'(0)=\mu<0$: асимптотично стійкий мінімум |
| $\mu=0$ | $E_s=0$ | Негіперболічна, але асимптотично стійка рівновага: $\dot E=-\beta E^3$ |
| $\mu>0$ | $E_s=0$ | $f'(0)=\mu>0$: нестійкий максимум |
| $\mu>0$ | $E_s=\pm\sqrt{\mu/\beta}$ | $f'(E_s)=-2\mu<0$: два стійкі мінімуми |

Це **надкритична вилкова біфуркація** амплітуди зі знаком. На порозі лінійний аналіз не дає відповіді, проте точний розв'язок $E(t)=E(0)/\sqrt{1+2\beta E(0)^2t}$ доводить алгебраїчне притягання. Поза порогом локальний час релаксації становить $1/|\mu|$ нижче порога та $1/(2\mu)$ на стійких ненульових гілках (одинична рухливість): за $\mu\to0$ виникає критичне сповільнення. Два знаки поля дають однакову інтенсивність. Для необмеженої комплексної амплітуди лазера вони належать неперервній родині фаз, тому дійсну вилкову біфуркацію не слід тлумачити як наявність лише двох можливих оптичних фаз.

Перший графік використовує $\beta_c$, обчислене з наведених лазерних параметрів, і малий інтервал відносного накачування. Суцільні гілки стійкі, штрихові — нестійкі. Другий приклад інтегрує скалярну нормальну форму й чисельно перевіряє тотожність Ляпунова. Це ілюстрації моделі, а не виміряні лазерні дані.


```python
# gradient_landscape: квазістатична нормальна форма поблизу порога; одинична рухливість.
beta_c = 2*kap*g**2/(gamma_perp*gamma_parallel)
def gradient_potential(field, mu, beta=beta_c, h=0.0):
    return -mu*field**2/2 + beta*field**4/4 - h*field

field_grid = np.linspace(-0.55, 0.55, 600)
fig, axes = plt.subplots(1, 2, figsize=(11, 4), constrained_layout=True)
for mu_value in [-0.08*kap, 0.0, 0.08*kap]:
    axes[0].plot(field_grid, gradient_potential(field_grid, mu_value),
                 label=fr'$\mu/\kappa={mu_value/kap:.2f}$')
axes[0].set(xlabel=r'$E$', ylabel=r'$V(E)$', title='Потенціал поблизу порога')
axes[0].legend()
mu_scan = np.linspace(-0.1*kap, 0.1*kap, 500)
positive_mu = mu_scan[mu_scan > 0]
axes[1].plot(mu_scan[mu_scan <= 0]/kap, np.zeros(np.sum(mu_scan <= 0)),
             color='C0', label='стійкі')
axes[1].plot(positive_mu/kap, np.zeros_like(positive_mu), 'k--', label='нестійкі')
for sign in [-1, 1]:
    axes[1].plot(positive_mu/kap, sign*np.sqrt(positive_mu/beta_c), color='C0')
axes[1].plot(0, 0, 'ko', ms=4)
axes[1].set(xlabel=r'$\mu/\kappa=D_0/D_c-1$', ylabel=r'$E_s$',
            title='Надкритична вилкова біфуркація')
axes[1].legend()
plt.show()
```

```python
# gradient_lyapunov: сталі параметри; змініть початкові поля й виконайте знову.
from scipy.integrate import cumulative_trapezoid
mu_demo = 0.08*kap
times_gradient = np.linspace(0, 100/kap, 24001)
fig, axes = plt.subplots(1, 2, figsize=(11, 4), constrained_layout=True)
lyapunov_errors = []
for initial_field in [-0.45, -0.03, 0.03, 0.45]:
    trajectory = solve_ivp(
        lambda t, y: mu_demo*y-beta_c*y**3,
        (times_gradient[0], times_gradient[-1]), [initial_field],
        t_eval=times_gradient, rtol=1e-10, atol=1e-12)
    assert trajectory.success, trajectory.message
    field_path = trajectory.y[0]
    potential_path = gradient_potential(field_path, mu_demo)
    drift_path = mu_demo*field_path-beta_c*field_path**3
    balance = potential_path-potential_path[0] + cumulative_trapezoid(
        drift_path**2, times_gradient, initial=0)
    lyapunov_errors.append(np.max(np.abs(balance)))
    assert np.max(np.diff(potential_path)) < 1e-10
    axes[0].plot(times_gradient*kap, field_path, label=fr'$E(0)={initial_field}$')
    axes[1].plot(times_gradient*kap, potential_path-potential_path[0])
axes[0].set(xlabel=r'$\kappa t$', ylabel=r'$E(t)$', title='Притягання до одного з мінімумів')
axes[0].legend()
axes[1].set(xlabel=r'$\kappa t$', ylabel=r'$V(t)-V(0)$', title='Спадання функції Ляпунова')
print('max |V(t)-V(0)+integral(dE/dt)^2 dt| =', max(lyapunov_errors))
assert max(lyapunov_errors) < 1e-7
plt.show()
```

### Порушення симетрії: недосконала вилкова біфуркація та розгортка каспа

Мале стале зміщення $h$ порушує симетрію $E\mapsto-E$:

$$\dot E=\mu E-\beta E^3+h=-\partial_E V_h,\qquad
V_h=\frac{\beta E^4}{4}-\frac{\mu E^2}{2}-hE.$$

Тут $h$ має одиниці амплітуда/час. У моделі з фіксованою фазою воно може описувати слабкий когерентний затравний сигнал, спроєктований на збережену квадратуру; це не некогерентне накачування $D_0$. Це феноменологічний параметр розгортки, а не виведення загальної моделі лазера із зовнішнім введенням сигналу.

Виберімо масштаб амплітуди $E_*>0$, уведімо $x=E/E_*$ і $\tau=\beta E_*^2t$ та поділімо потенціал на $\beta E_*^4$. Тоді

$$U(x;a,b)=\frac{x^4}{4}+\frac{a x^2}{2}+bx,\qquad
\frac{dx}{d\tau}=-(x^3+ax+b),\quad
a=-\frac{\mu}{\beta E_*^2},\quad b=-\frac{h}{\beta E_*^3}.$$

Це та сама домовленість про знаки, що й у модулі 04. Симетрична вилкова біфуркація є перерізом $b=0$ **розгортки каспа з двома керувальними параметрами**. Для кубічного рівняння рівноваг $x^3+ax+b=0$ визначимо алгебраїчний дискримінант

$$\Delta=-4a^3-27b^2=\frac{4\mu^3-27\beta h^2}{\beta^3 E_*^6}.$$

За $\Delta>0$ існують три різні дійсні корені: два стійкі мінімуми та нестійкий максимум між ними. За $\Delta<0$ є один дійсний корінь — стійкий мінімум. За $\Delta=0$ корені зливаються: у типовій точці складки зустрічаються стійка й нестійка рівноваги; потрійний корінь у $(a,b)=(0,0)$ є точкою каспа. Негіперболічні корені потребують нелінійного аналізу. Типовий подвійний корінь на складці притягує з одного боку й відштовхує з іншого; потрійний корінь у точці каспа за фіксованих параметрів притягує алгебраїчно.

Умови $U_x=U_{xx}=0$ дають

$$a=-3x_f^2,\qquad b=2x_f^3,\qquad 4a^3+27b^2=0.$$

Еквівалентно, $\mu=3\beta E_f^2$, $h=-2\beta E_f^3$, тому область трьох рівноваг визначається $\mu>0$ та $|h|<2\mu^{3/2}/(3\sqrt{3\beta})$. Вираз $4a^3+27b^2$ також часто називають дискримінантом каспа; його знак протилежний знаку $\Delta$. За сталого $\mu>0$ зміна $h$ нахиляє потенціальні ями. Детермінований стан може залишатися в локальному мінімумі, доки той не зникне на складці. Однакова глибина ям ($h=0$) сама собою не примушує систему перемикатися. Тому повільні прямий і зворотний проходи можуть давати гістерезис; скінченна швидкість зміни параметрів і шум потребують подальшого динамічного аналізу.

**Дослідіть:** змініть `a_explore` і `b_explore` нижче, виконайте комірку знову та зіставте екстремуми потенціалу, переріз поверхні рівноваг і позначену точку керувальних параметрів. Спробуйте $(-1,0)$, $(-1,0.25)$, $(-1,0.5)$ і $(0.4,0.2)$. Зафарбовані кружечки/суцільні гілки позначають стійкі корені, хрестики/штрихові гілки — нестійкі. Перевірка складки використовує аналітичну параметризацію, уникаючи ненадійного підрахунку майже кратних чисельних коренів.


```python
# gradient_cusp: безрозмірні канонічні параметри; змініть їх і виконайте знову.
a_explore, b_explore = -1.0, 0.25
def cusp_potential(x, a, b):
    return x**4/4 + a*x**2/2 + b*x

roots = np.roots([1.0, 0.0, a_explore, b_explore])
real_roots = np.sort(roots.real[np.abs(roots.imag) < 1e-7])
delta = -4*a_explore**3-27*b_explore**2
print('Delta =', delta, '; дійсні корені (чисельно) =', real_roots)
if abs(delta) > 1e-6:
    assert len(real_roots) == (3 if delta > 0 else 1)
assert np.allclose(real_roots**3+a_explore*real_roots+b_explore, 0, atol=1e-7)

fig, axes = plt.subplots(1, 3, figsize=(15, 4.2), constrained_layout=True)
x_view = np.linspace(-2.0, 2.0, 700)
axes[0].plot(x_view, cusp_potential(x_view, a_explore, b_explore))
for root in real_roots:
    curvature = 3*root**2+a_explore
    marker = 'o' if curvature > 1e-6 else ('x' if curvature < -1e-6 else 's')
    axes[0].plot(root, cusp_potential(root, a_explore, b_explore),
                 marker=marker, color='C0' if curvature > 1e-6 else 'black', ms=7)
axes[0].set(xlabel=r'$x$', ylabel=r'$U$', title='Нахилений потенціал')

# Параметризуємо переріз рівноваг через x: b = -x^3-a*x.
curvature = 3*x_view**2+a_explore
b_section = -x_view**3-a_explore*x_view
axes[1].plot(np.where(curvature > 0, b_section, np.nan), x_view,
             color='C0', label='стійкі')
axes[1].plot(np.where(curvature < 0, b_section, np.nan), x_view,
             'k--', label='нестійкі')
axes[1].axvline(b_explore, color='C1', lw=1)
axes[1].set(xlabel=r'$b$', ylabel=r'$x_s$', xlim=(-0.8, 0.8),
            title=fr'Рівноваги за $a={a_explore:g}$')
axes[1].legend()

fold_x = np.linspace(-0.85, 0.85, 801)
fold_a, fold_b = -3*fold_x**2, 2*fold_x**3
assert np.allclose(fold_x**3+fold_a*fold_x+fold_b, 0)
assert np.allclose(3*fold_x**2+fold_a, 0)
assert np.allclose(4*fold_a**3+27*fold_b**2, 0)
a_region = np.linspace(-2.2, 0, 400)
b_limit = 2*(-a_region/3)**1.5
axes[2].fill_between(a_region, -b_limit, b_limit, color='C0', alpha=0.15,
                     label='3 рівноваги')
axes[2].plot(fold_a, fold_b, 'k-', label='множина складок')
axes[2].plot(a_explore, b_explore, 'o', color='C1', label='вибрані параметри')
axes[2].set(xlabel=r'$a$', ylabel=r'$b$', title='Площина параметрів каспа')
axes[2].legend(fontsize=9)
plt.show()
```

### Що випливає з потенціального ландшафту, а що потребує окремого обґрунтування

Для будь-якого гладкого **автономного скалярного** рівняння $\dot E=f(E)$ на інтервалі функція $V(E)=-\int^E f(u)\,du$ задає градієнтне представлення з одиничною рухливістю. Ця проста побудова не вимагає теплової рівноваги. Якщо керувальні параметри змінюються в часі, доданок $\partial_tV$ означає, що миттєвий потенціал може й не спадати: для нормальної форми зі зміщенням

$$\frac{dV_h}{dt}=-(\partial_E V_h)^2-\frac{\dot\mu}{2}E^2
+\frac{\dot\beta}{4}E^4-\dot h E.$$

У кількох вимірах евклідове представлення $\dot{\mathbf q}=\mathbf F=-\nabla V$ вимагає симетричного якобіана, $\partial_jF_i=\partial_iF_j$, і належної топології області (в однозв'язній області умова відсутності вихору є достатньою). Повна дійсна лазерна модель загалом не проходить цю перевірку: $\partial_P\dot E=g$, тоді як $\partial_E\dot P=gD$. Тому скалярний потенціал зведеної амплітуди не встановлює існування евклідового градієнтного потенціалу для $(E,P,D)$ і не виключає коливань та хаосу поза скалярним режимом.

Для системи $\dot{\mathbf q}=-\mathbf M(\mathbf q)\nabla V$ із симетричною додатно визначеною матрицею рухливості маємо $\dot V=-\nabla V^T\mathbf M\nabla V\le0$, але існування такого представлення потребує окремого обґрунтування. Сама функція Ляпунова не означає суто градієнтної динаміки: циркуляція може співіснувати зі спаданням функції. Для комплексної амплітуди $\mathsf E=X+iY$ спеціальне рівняння з дійсними коефіцієнтами $\dot{\mathsf E}=(\mu-\beta|\mathsf E|^2)\mathsf E$ є градієнтним у $(X,Y)$ з радіальним потенціалом $-\mu(X^2+Y^2)/2+\beta(X^2+Y^2)^2/4$. Вище порога воно має коло мінімумів із нейтральною фазою. Розстроювання або уявні коефіцієнти можуть породжувати обертання фази, тому цей окремий результат не можна автоматично переносити на загальні лазерні рівняння.
