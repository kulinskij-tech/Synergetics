## Необхідний вступ: градієнтний потік, метастабільність і гістерезис

В одновимірній градієнтній системі $\dot x=-\partial_xV(x;\mu)$ потенціал уздовж траєкторії спадає, адже $\dot V=-(\partial_xV)^2\le0$. Локальні мінімуми $V$ є стійкими рівновагами, а максимуми — нестійкими. Кілька мінімумів відповідають конкуруючим метастабільним макростанам, розділеним бар'єрами.

Гістерезис — це залежність стану від шляху зміни параметра: система може залишатися в локальному мінімумі, доки він не зникне на складці, а тоді стрибком перейти на іншу гілку. Потенціальний опис особливо сильний для справді градієнтної динаміки; для систем із циркуляцією в просторі станів одного скалярного ландшафту зазвичай недостатньо.


## Нормальна форма каспа

Канонічний потенціал

$$V(x;a,b)=\frac{x^4}{4}+\frac{a x^2}{2}+bx$$

породжує градієнтну динаміку $\dot x=-\partial_xV=-(x^3+ax+b)$. Рівноваги задовольняють $x^3+ax+b=0$ і є стійкими за умови $V''=3x^2+a>0$. На складках одночасно виконуються $V'=0$ та $V''=0$, звідки

$$a=-3x^2,\qquad b=2x^3,$$

а дискримінант каспа має вигляд $4a^3+27b^2=0$.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "axes.grid": True,
    "grid.alpha": 0.25,
    "font.size": 11,
})
rng = np.random.default_rng(20260904)

def real_equilibria(a, b, tol=1e-9):
    roots = np.roots([1.0, 0.0, a, b])
    roots = roots[np.abs(roots.imag) < tol].real
    return np.sort(roots)

a_values = [-1.4, -0.5, 0.4]
x = np.linspace(-2.0, 2.0, 800)
fig, ax = plt.subplots(1, 3, figsize=(13, 3.8), sharey=True)
for axis, a in zip(ax, a_values):
    for b in [-0.4, 0.0, 0.4]:
        V = x**4/4 + a*x**2/2 + b*x
        axis.plot(x, V, label=fr"$b={b}$")
    axis.set(title=fr"$a={a}$", xlabel="state x")
ax[0].set_ylabel("potential V")
ax[-1].legend()
plt.tight_layout()
```

```python
# Equilibrium surface projected onto the (b, x) plane for fixed a < 0.
a = -1.0
x_branch = np.linspace(-1.4, 1.4, 1000)
b_branch = -(x_branch**3 + a*x_branch)
stable = 3*x_branch**2 + a > 0

plt.plot(b_branch[stable], x_branch[stable], "C0.", ms=2, label="stable")
plt.plot(b_branch[~stable], x_branch[~stable], "C3.", ms=2, label="unstable")
plt.xlabel("control b")
plt.ylabel("equilibrium x")
plt.title("Cusp slice and hysteresis branches")
plt.legend()
plt.show()
```
