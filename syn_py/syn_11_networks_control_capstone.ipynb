## Prerequisite bridge: networks and diffusive coupling

A network consists of nodes with local states and edges that specify interactions. Its adjacency matrix $A_{ij}$ records connections, while the graph Laplacian $L=K-A$ compares each node with its neighbors. A standard continuous diffusive coupling is

$$\dot{\mathbf x}_i=\mathbf F(\mathbf x_i)-\epsilon\sum_jL_{ij}\mathbf H(\mathbf x_j).$$

The synchronized manifold has $\mathbf x_1=\cdots=\mathbf x_N$. Whether perturbations transverse to it decay depends jointly on local dynamics, coupling strength, and Laplacian eigenvalues. In discrete coupled maps, update order matters: all new states should be computed from the same old time layer unless an asynchronous model is intended.


## Coupled maps as a laboratory

A diffusive ring of logistic maps can be written

$$x_i^{n+1}=(1-\epsilon)f(x_i^n)+\frac{\epsilon}{2}\left[f(x_{i-1}^n)+f(x_{i+1}^n)\right],\quad f(x)=rx(1-x).$$

Local chaos competes with spatial coupling. Useful observables include spatial variance, neighbor correlation, laminar-domain lengths, and temporal Lyapunov growth. Calling a low-variance state “organized” is meaningful only relative to a defined observable and robustness test.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "axes.grid": True,
    "grid.alpha": 0.25,
    "font.size": 11,
})
rng = np.random.default_rng(20260904)

def coupled_logistic(epsilon, r=3.9, sites=120, steps=700, burn=200):
    x = rng.uniform(0.1, 0.9, sites)
    history = []
    for n in range(steps):
        fx = r*x*(1-x)
        x = ((1-epsilon)*fx
             + 0.5*epsilon*np.roll(fx, 1)
             + 0.5*epsilon*np.roll(fx, -1))
        if n >= burn:
            history.append(x.copy())
    return np.asarray(history)

eps_values = [0.0, 0.15, 0.35, 0.65]
fig, ax = plt.subplots(1, 4, figsize=(13, 4), sharey=True)
for axis, eps in zip(ax, eps_values):
    history = coupled_logistic(eps)
    axis.imshow(history, aspect="auto", origin="lower", cmap="viridis")
    axis.set(title=fr"$\epsilon={eps}$", xlabel="site")
ax[0].set_ylabel("time after transient")
plt.tight_layout()
```

## Small feedback and chaos control

The Ott–Grebogi–Yorke idea exploits recurrent visits of a chaotic trajectory near an unstable periodic orbit. A small, state-dependent change of an accessible control parameter can stabilize that orbit. The perturbation is small because chaos supplies the approach; feedback supplies the local contraction. Practical analysis must include parameter bounds, observation error, basin effects, and noise.
