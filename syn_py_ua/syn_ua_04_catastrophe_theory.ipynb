## Необхідний вступ: градієнтний потік, метастабільність і гістерезис

В одновимірній градієнтній системі $\dot x=-\partial_xV(x;\mu)$ потенціал уздовж траєкторії спадає, адже $\dot V=-(\partial_xV)^2\le0$. Локальні мінімуми $V$ є стійкими рівновагами, а максимуми — нестійкими. Кілька мінімумів відповідають конкуруючим метастабільним макростанам, розділеним бар'єрами.

Гістерезис — це залежність стану від шляху зміни параметра: система може залишатися в локальному мінімумі, доки він не зникне на складці, а тоді стрибком перейти на іншу гілку. Потенціальний опис особливо сильний для справді градієнтної динаміки; для систем із циркуляцією в просторі станів одного скалярного ландшафту зазвичай недостатньо.


## Звідки походить теорія катастроф: геометрія, що зберігається при збуреннях

Арнольд знайомить із предметом через два джерела: **особливості гладких відображень** і **біфуркації динамічних систем**. Гладка поверхня може проєктуватися на площину з гострим видимим контуром, хоча сама поверхня залишається гладкою. Головне питання — які риси зберігаються при малих збуреннях, а не точна форма окремого рисунка.

<img src="./figs/fig_whitneymaps.png" alt="Проєкції зі складкою та каспом Вітні за Арнольдом" width="500">

*Ілюстрація особливостей Вітні в Арнольда: проєкція сфери зі складкою (ліворуч) і гладка поверхня, проєкція якої має касп (праворуч). Джерело: надана викладачем книга Арнольда, друкована с. 5, рис. 1–2. Координати $(x_1,x_2)$ належать вихідній поверхні, $(y_1,y_2)$ — площині образу.*

Для гладкого відображення $F:\mathbb R^2\to\mathbb R^2$ точка є **регулярною**, якщо $\det DF\ne0$; локально відображення оборотне. В **особливій точці** ранг зменшується. Її образ є **критичним значенням**: розрізняйте точку на вихідній поверхні та її образ на площині.

Локально структурна стійкість означає, що достатньо мале гладке збурення має той самий тип після гладких оборотних замін координат у джерелі й образі; положення точки може змінитися. Схематично еквівалентні відображення задовольняють $\widetilde F=\psi\circ F\circ\phi^{-1}$. Близькість має контролювати відповідні похідні, а не лише значення функцій. Тут ідеться про локальні типи; глобальна стійкість потребує додаткових умов.

**Два різні значення стійкості:** складка або касп можуть бути структурно стійкими особливостями відображення; асимптотично стійка рівновага притягує сусідні траєкторії, які при цьому залишаються поблизу неї. Одне твердження не випливає з іншого. Рівновага може зберігатися при збуреннях і водночас відштовхувати траєкторії.


### Від багатовимірного аналізу до катастроф: схема переходу

Таблиця нижче адаптує таблицю 2.1 Ґілмора до знайомого багатовимірного аналізу. Нехай $V:\mathbb R^n\to\mathbb R$ — гладка функція; усі похідні обчислюємо в точці $p$. Позначимо $H=D^2V(p)$, $\ell=\dim\ker H$ (**коранг**) і $Q_m(\mathbf w)=\sum_{j=1}^m\sigma_jw_j^2$, $\sigma_j=\pm1$. Адитивні сталі опущено; координати локальні й гладко оборотні.

| Умови | Локальна форма $V-V(p)$ | Результат і його зміст |
|---|---|---|
| $\nabla V(p)\ne0$ | $y_1$ | Теорема про неявну/обернену функцію: випрямлення регулярної поверхні рівня; рівноваги $-\nabla V$ тут немає. |
| $\nabla V(p)=0$, $\det H\ne0$ | $Q_n(\mathbf y)$ | **Лема Морса:** невироджена критична точка локально квадратична; індекс дорівнює числу мінусів. |
| $\nabla V(p)=0$, $\operatorname{rank}H=n-\ell<n$ | $g(z_1,\ldots,z_\ell)+Q_{n-\ell}(\mathbf w)$ | **Лема розщеплення:** зберігаємо $\ell$ істотних вироджених координат; $g(0)=0$, $Dg(0)=D^2g(0)=0$. |
| Гладкі родини загального положення з $k$ керувальними параметрами, $k\le5$, за звичайної еквівалентності розгорток | $\mathcal C_\alpha(\mathbf z;\boldsymbol a)+Q_{n-\ell}(\mathbf w)$ | **Класифікація й загальність Тома–Мазера:** скінченний каталог локальних стійких розгорток; складка й касп — перші приклади. |

Тут **розгортка** додає незалежні керувальні члени, що описують близькі збурення виродженої функції. **Росток** означає локальну поведінку функції поблизу вибраної точки. Каталог стосується локальних гладких родин потенціалів загального положення; обмеження симетрії можуть змінювати допустимі розгортки. Відомі сім елементарних типів потребують щонайбільше чотирьох керувальних параметрів; каталог для п'яти параметрів ширший. Останній рядок не є класифікацією всіх глобальних динамічних систем.

**Чому це важливо для синергетики.** Поблизу переходу обґрунтоване підпорядкування зводить динаміку до кількох повільних параметрів порядку. Далі розщеплення відокремлює невироджені напрями потенціалу, а нормальні форми впорядковують істотну поведінку, що залишилася, у кілька стійких структурних моделей. Підпорядкування спирається на розділення часових масштабів; розщеплення є геометричним твердженням про функцію й саме собою не доводить швидкої релаксації. Різні фізичні системи можуть мати спільну локальну модель, зберігаючи різні коефіцієнти, області параметрів і спостережувані величини. Неградієнтні переходи потребують відповідних динамічних нормальних форм.


### Критичні точки Морса: стійкі елементи побудови

Для скалярного потенціалу $V(\mathbf q)$ критична точка $p$ задовольняє $\nabla V(p)=0$. Вона є **невиродженою**, якщо гессіан $H_{ij}=\partial_i\partial_jV(p)$ оборотний. **Індекс Морса** $r$ — число від'ємних власних значень гессіана. Лема Морса стверджує, що у відповідних гладких локальних координатах маємо точний вираз

$$V=V(p)-z_1^2-\cdots-z_r^2+z_{r+1}^2+\cdots+z_n^2.$$

Це більше, ніж обрив ряду Тейлора: поблизу невиродженої точки члени вищих порядків можна усунути заміною координат. Для $\dot{\mathbf q}=-\nabla V$ лінеаризація дорівнює $-H$:

| Тип у двох вимірах | Індекс Морса | Градієнтна динаміка |
|---|---|---|
| Мінімум, $+z_1^2+z_2^2$ | 0 | Атрактор |
| Сідло, $-z_1^2+z_2^2$ | 1 | Один напрям зростання й один згасання |
| Максимум, $-z_1^2-z_2^2$ | 2 | Репелер |

Оборотність гессіана дозволяє за теоремою про неявну функцію продовжити критичну точку при малих $C^2$-збуреннях; її індекс зберігається. Точка не обов'язково залишається в початку координат. Нелінійна заміна координат змінює також рухливість/метрику градієнта, тому лема не стверджує тотожності евклідових траєкторій чи часів релаксації.

**Рисунок 1.** Угорі — чаші, перевали й вершини; внизу — лінії рівня та стрілки руху вниз після додавання $\varepsilon(u+v/2)$. Змініть `morse_epsilon`: критична точка зміститься (зірочка), але її тип збережеться. Це оригінальні ілюстрації моделей.


```python
# morse_geometry: локальні критичні точки та малий нахил.
import numpy as np
import matplotlib.pyplot as plt

morse_epsilon = 0.25
mesh = np.linspace(-1.5, 1.5, 65)
UU, VV = np.meshgrid(mesh, mesh)
fig = plt.figure(figsize=(12, 7.2), constrained_layout=True)
morse_indices = []
for column, (su, sv, name) in enumerate([(1, 1, 'Мінімум'), (-1, 1, 'Сідло'), (-1, -1, 'Максимум')]):
    surface = (su*UU**2+sv*VV**2)/2
    axis = fig.add_subplot(2, 3, column+1, projection='3d')
    axis.plot_surface(UU, VV, surface, cmap='cividis', alpha=0.88,
                      rcount=30, ccount=30, linewidth=0)
    index = int(su < 0)+int(sv < 0)
    morse_indices.append(index)
    axis.set(xlabel=r'$u$', ylabel=r'$v$', title=f'{name}: індекс {index}')
    axis.text2D(0.90, 0.87, r'$V$', transform=axis.transAxes)
    axis.view_init(elev=27, azim=-55)
    axis = fig.add_subplot(2, 3, column+4)
    tilted = surface+morse_epsilon*(UU+VV/2)
    axis.contour(UU, VV, tilted, levels=12, colors='0.55', linewidths=0.7)
    arrows = np.linspace(-1.35, 1.35, 11)
    AU, AV = np.meshgrid(arrows, arrows)
    FU, FV = -su*AU-morse_epsilon, -sv*AV-morse_epsilon/2
    speed = np.hypot(FU, FV)
    axis.quiver(AU, AV, FU/(speed+1e-12), FV/(speed+1e-12),
                color='C0', scale=19, width=0.005)
    critical = np.array([-morse_epsilon/su, -morse_epsilon/(2*sv)])
    assert np.allclose(np.array([su, sv])*critical+[morse_epsilon, morse_epsilon/2], 0)
    assert np.count_nonzero(np.linalg.eigvalsh(np.diag([su, sv])) < 0) == index
    axis.plot(0, 0, 'ko', mfc='white', label='початкова точка')
    axis.plot(*critical, '*', color='C1', ms=13, label='зміщена точка')
    axis.set(xlabel=r'$u$', ylabel=r'$v$', aspect='equal',
             xlim=(-1.5, 1.5), ylim=(-1.5, 1.5))
    if column == 0:
        axis.legend(fontsize=8, loc='upper left')
assert morse_indices == [0, 1, 2]
print('Індекси Морса:', morse_indices)
plt.show()
```

### Складки й каспи Вітні: особливі відображення можуть бути стійкими

Вертикальна проєкція сфери $\pi(X,Y,Z)=(X,Y)$ має два прообрази всередині контуру, один на контурі й жодного зовні. На екваторі дотична площина містить напрям спостереження, тому ранг проєкції зменшується. Поблизу звичайної точки контуру її локальною формою є **складка**. Для гладких відображень загального положення між двовимірними поверхнями локальні типи особливостей — складки та каспи:

| Відображення $(u,v)\mapsto(u,y)$ | Особлива множина в джерелі | Критичні значення в образі |
|---|---|---|
| Складка: $y=v^2$ | $v=0$ | $y=0$ |
| Касп: $y=v^3+uv$ | $u=-3v^2$ | $(u,y)=(-3t^2,-2t^3)$ |

Справді, визначники якобіанів дорівнюють $2v$ і $3v^2+u$. Крива критичних значень каспа задовольняє $4u^3+27y^2=0$. Точка всередині клиноподібної області має три прообрази, зовні — один. На межі-складці два прообрази зливаються; у точці каспа зливаються всі три. Подвійний корінь разом із простим дають два різні прообрази в точці межі, відмінній від каспа.

**Рисунок 2.** Читайте зображення як вихідна поверхня → проєкція → крива критичних значень. Червоним позначено множину втрати рангу та її образ. Поверхня каспа гладка: особливість належить її проєкції. Це нові рисунки за явними рівняннями, а не відтворення скану Арнольда. Їхня стійкість стосується локальної геометрії при збуреннях загального положення, а не притягання траєкторій.


```python
# whitney_geometry: контур сфери, складка, поверхня каспа та її критичні значення.
fig = plt.figure(figsize=(12, 9), constrained_layout=True)
axis = fig.add_subplot(2, 2, 1, projection='3d')
theta = np.linspace(0, 2*np.pi, 80)
polar = np.linspace(0, np.pi, 25)
TT, PP = np.meshgrid(theta, polar)
axis.plot_wireframe(np.sin(PP)*np.cos(TT), np.sin(PP)*np.sin(TT), np.cos(PP),
                    rstride=3, cstride=7, color='0.6', linewidth=0.5)
ground = -1.4
axis.plot(np.cos(theta), np.sin(theta), np.zeros_like(theta), color='C3', lw=2)
axis.plot(np.cos(theta), np.sin(theta), ground*np.ones_like(theta), color='C3', lw=2)
RR, AA = np.meshgrid(np.linspace(0, 1, 10), theta)
axis.plot_surface(RR*np.cos(AA), RR*np.sin(AA), ground*np.ones_like(RR),
                  color='C0', alpha=0.10, linewidth=0)
for angle in np.linspace(0, 2*np.pi, 7)[:-1]:
    axis.plot([np.cos(angle)]*2, [np.sin(angle)]*2, [0, ground],
              '--', color='0.6', lw=0.7)
axis.set(xlabel=r'$X$', ylabel=r'$Y$', zlabel=r'$Z$', title='Сфера: складка вздовж контуру')
axis.view_init(elev=24, azim=-60)

axis = fig.add_subplot(2, 2, 2)
axis.axhspan(0, 1.4, color='C0', alpha=0.13)
axis.axhline(0, color='C3', lw=2)
axis.text(0, 0.75, '2 прообрази', ha='center')
axis.text(0, 0.08, '1 прообраз на складці', ha='center', va='bottom')
axis.text(0, -0.65, '0 прообразів', ha='center')
axis.set(xlim=(-1.3, 1.3), ylim=(-1.2, 1.4), xlabel=r'$u$', ylabel=r'$y$',
          title=r'Проєкція зі складкою: $(u,v)\mapsto(u,v^2)$')

axis = fig.add_subplot(2, 2, 3, projection='3d')
cu = np.linspace(-1.65, 0.6, 28)
cv = np.linspace(-1.1, 1.1, 42)
CU, CV = np.meshgrid(cu, cv)
CY = CV**3+CU*CV
axis.plot_wireframe(CU, CY, CV, rstride=3, cstride=3, color='C0',
                    alpha=0.5, linewidth=0.6)
parameter = np.linspace(-0.74, 0.74, 301)
fold_u, fold_y = -3*parameter**2, -2*parameter**3
axis.plot(fold_u, fold_y, parameter, color='C3', lw=2.5)
axis.plot(fold_u, fold_y, ground*np.ones_like(parameter), color='C3', lw=2)
for t in [-0.6, 0, 0.6]:
    axis.plot([-3*t*t]*2, [-2*t**3]*2, [t, ground], '--', color='0.5', lw=0.8)
axis.set(xlabel=r'$u$', ylabel=r'$y=v^3+uv$', zlabel=r'$v$',
          title='Гладка поверхня; особлива проєкція')
axis.view_init(elev=25, azim=-62)

axis = fig.add_subplot(2, 2, 4)
negative_u = np.linspace(-1.65, 0, 400)
limit_y = 2*(-negative_u/3)**1.5
axis.fill_between(negative_u, -limit_y, limit_y, color='C0', alpha=0.13)
axis.plot(fold_u, fold_y, color='C3', lw=2)
axis.plot([-1, -1], [0, 0.6], 'ko', ms=4)
axis.annotate('3 прообрази', (-1, 0), xytext=(-1.5, -0.13))
axis.annotate('1 прообраз', (-1, 0.6), xytext=(-0.75, 0.64))
axis.set(xlabel=r'$u$', ylabel=r'$y$', title='Крива критичних значень каспа',
          xlim=(-1.7, 0.45), ylim=(-0.9, 0.9))
assert np.allclose(parameter**3+fold_u*parameter, fold_y)
assert np.allclose(3*parameter**2+fold_u, 0)
assert np.allclose(4*fold_u**3+27*fold_y**2, 0)
for target_y, expected_count in [(0.0, 3), (0.6, 1)]:
    roots = np.roots([1, 0, -1, -target_y])
    assert np.count_nonzero(np.abs(roots.imag) < 1e-9) == expected_count
plt.show()
```

### Від теорії Морса до катастроф у параметричних родинах

Теорія Морса вивчає також множини підрівня $M_c=\{\mathbf q:V(\mathbf q)\le c\}$. Якщо проміжний шар рівнів компактний і не містить критичних точок, проходження цього шару не змінює топологію. У нашому прикладі $W(u,v)=(u^2-1)^2+v^2$ мінімуми $(\pm1,0)$ мають значення 0 та індекс 0, а сідло $(0,0)$ — значення 1 та індекс 1. За $0<c<1$ є дві компоненти; за $c>1$ вони з'єднуються. Це топологія **фіксованого** ландшафту при зміні рівня, а не динамічна біфуркація внаслідок зміни його керувальних параметрів.

Катастрофа стосується **родини**, у якій критична точка стає виродженою. Для потенціалу складки $G(x;\lambda)=x^3/3-\lambda x$ маємо $G_x=x^2-\lambda$: за $\lambda>0$ існують мінімум у $+\sqrt\lambda$ та максимум у $-\sqrt\lambda$; за нуля вони стикаються, причому $G_{xx}=0$; за $\lambda<0$ обидва зникають. Цей кубічний потенціал є локальною моделлю, а не глобальним потенціалом, що обмежує рух.

Для використаного нижче потенціалу четвертого степеня поверхня рівноваг є гладким графіком

$$b=-(x^3+ax).$$

Її проєкція на $(a,b)$ — саме відображення каспа Вітні, наведене вище, з $u=a$, $v=x$ та **$y=-b$**. Ранг проєкції зменшується там, де $V_{xx}=3x^2+a=0$. На звичайній складці $V_{xxx}=6x\ne0$; у точці каспа $x=a=b=0$ також $V_{xxx}=0$, але $V_{xxxx}=6\ne0$. Отже, типова складка потребує одного керувального параметра, касп — двох; невироджені напрями зберігають квадратичні члени, а вироджений напрям потребує вищих степенів. Касп відображення є стійким у родині з двома керувальними параметрами, хоча потенціал у точці каспа не є морсівським. Геометрія визначає можливі рівноваги; градієнтний закон виділяє притягальні мінімуми.

**Рисунок 3.** Ліворуч множини підрівня $W$ з'єднуються на сідловому значенні. Праворуч зміна $\lambda$ створює або знищує дві критичні точки $G$. Порівняйте ці дві різні зміни, перш ніж переходити до наявних обчислень для каспа.


```python
# morse_sublevels: топологія одного потенціалу та родина потенціалів.
from scipy.ndimage import label as connected_components
line = np.linspace(-1.7, 1.7, 501)
SU, SV = np.meshgrid(line, line)
WW = (SU**2-1)**2+SV**2
fig, axes = plt.subplots(1, 2, figsize=(12, 4.3), constrained_layout=True)
contours = axes[0].contour(SU, SV, WW, levels=[0.2, 0.8, 1.0, 1.3],
                           colors=['C0', 'C1', 'C3', 'black'])
axes[0].clabel(contours, fmt='%g', fontsize=9)
axes[0].plot([-1, 1], [0, 0], 'o', color='C0', label='мінімуми: W=0')
axes[0].plot(0, 0, 'x', color='C3', ms=8, label='сідло: W=1')
axes[0].set(xlabel=r'$u$', ylabel=r'$v$', aspect='equal', title='Множини підрівня з’єднуються на сідлі')
axes[0].legend(fontsize=8, loc='upper right')
counts = []
for level, expected in [(0.2, 2), (0.8, 2), (1.3, 1)]:
    count = connected_components(WW <= level)[1]
    counts.append(int(count))
    assert count == expected
assert np.allclose(np.linalg.eigvalsh(np.diag([8.0, 2.0])), [2, 8])
assert np.allclose(np.linalg.eigvalsh(np.diag([-4.0, 2.0])), [-4, 2])
print('Число компонент сітки на рівнях 0.2, 0.8, 1.3:', counts)

state = np.linspace(-1.3, 1.3, 600)
for control in [-0.4, 0.0, 0.4]:
    value = state**3/3-control*state
    axes[1].plot(state, value, label=fr'$\lambda={control:g}$')
    if control > 0:
        for root in [-np.sqrt(control), np.sqrt(control)]:
            assert np.isclose(root**2-control, 0)
            axes[1].plot(root, root**3/3-control*root,
                         'o' if root > 0 else 'x', color='black', ms=7)
axes[1].set(xlabel=r'$x$', ylabel=r'$G(x;\lambda)$', title='Складка: мінімум і максимум зливаються')
axes[1].legend()
plt.show()
```

### Як читати каталог елементарних катастроф Ґілмора

Ця стисла версія таблиці 2.2 Ґілмора містить росток $g$ та керувальне збурення $P$; потенціал дорівнює $g+P$ плюс відокремлені невироджені квадратичні члени. Число $k$ означає кількість істотних незалежних керувальних параметрів, а не змінних стану. Форми $A$ мають одну істотну змінну стану, форми $D$ та $E$ — дві. Верхні індекси зберігають дійсні знакові варіанти джерела.

| Тип | $k$ | Росток $g$ | Збурення $P$ |
|---|---:|---|---|
| $A_2$ (складка) | 1 | $x^3$ | $a_1x$ |
| $A_3^{\pm}$ (касп) | 2 | $\pm x^4$ | $a_1x+a_2x^2$ |
| $A_4$ (ластівчин хвіст) | 3 | $x^5$ | $\sum_{j=1}^{3}a_jx^j$ |
| $A_5^{\pm}$ (метелик) | 4 | $\pm x^6$ | $\sum_{j=1}^{4}a_jx^j$ |
| $A_6$ | 5 | $x^7$ | $\sum_{j=1}^{5}a_jx^j$ |
| $D_4^{-}$ (еліптична омбіліка) | 3 | $x^2y-y^3$ | $a_1x+a_2y+a_3y^2$ |
| $D_4^{+}$ (гіперболічна омбіліка) | 3 | $x^2y+y^3$ | $a_1x+a_2y+a_3y^2$ |
| $D_5$ (параболічна омбіліка) | 4 | $x^2y+y^4$ | $a_1x+a_2y+a_3x^2+a_4y^2$ |
| $D_6^{-}$ | 5 | $x^2y-y^5$ | $a_1x+a_2y+a_3x^2+a_4y^2+a_5y^3$ |
| $D_6^{+}$ | 5 | $x^2y+y^5$ | $a_1x+a_2y+a_3x^2+a_4y^2+a_5y^3$ |
| $E_6^{\pm}$ | 5 | $x^3\pm y^4$ | $a_1x+a_2y+a_3xy+a_4y^2+a_5xy^2$ |

Таблиця слугує орієнтиром для читання, а не матеріалом для заучування. Рядки $A_2$ та додатний $A_3$ — моделі, використані тут. Множення нашого потенціалу складки на 3 дає $x^3-3\lambda x$; множення потенціалу каспа на 4 дає $x^4+2ax^2+4bx$. Отже, коефіцієнти узгоджуються після додатного перемасштабування. Знаки залишаються важливими для мінімумів і градієнтної стійкості: $+x^4$ та $-x^4$ поводяться по-різному. Поява ростка в каталозі сама собою не гарантує притягальності рівноваги.

## Нормальна форма каспа

Канонічний потенціал

$$V(x;a,b)=\frac{x^4}{4}+\frac{a x^2}{2}+bx$$

породжує градієнтну динаміку $\dot x=-\partial_xV=-(x^3+ax+b)$. Рівноваги задовольняють $x^3+ax+b=0$ і є стійкими за умови $V''=3x^2+a>0$. На складках одночасно виконуються $V'=0$ та $V''=0$, звідки

$$a=-3x^2,\qquad b=2x^3,$$

а дискримінант каспа має вигляд $4a^3+27b^2=0$.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "axes.grid": True,
    "grid.alpha": 0.25,
    "font.size": 11,
})
rng = np.random.default_rng(20260904)

def real_equilibria(a, b, tol=1e-9):
    roots = np.roots([1.0, 0.0, a, b])
    roots = roots[np.abs(roots.imag) < tol].real
    return np.sort(roots)

a_values = [-1.4, -0.5, 0.4]
x = np.linspace(-2.0, 2.0, 800)
fig, ax = plt.subplots(1, 3, figsize=(13, 3.8), sharey=True)
for axis, a in zip(ax, a_values):
    for b in [-0.4, 0.0, 0.4]:
        V = x**4/4 + a*x**2/2 + b*x
        axis.plot(x, V, label=fr"$b={b}$")
    axis.set(title=fr"$a={a}$", xlabel="state x")
ax[0].set_ylabel("potential V")
ax[-1].legend()
plt.tight_layout()
```

```python
# Equilibrium surface projected onto the (b, x) plane for fixed a < 0.
a = -1.0
x_branch = np.linspace(-1.4, 1.4, 1000)
b_branch = -(x_branch**3 + a*x_branch)
stable = 3*x_branch**2 + a > 0

plt.plot(b_branch[stable], x_branch[stable], "C0.", ms=2, label="stable")
plt.plot(b_branch[~stable], x_branch[~stable], "C3.", ms=2, label="unstable")
plt.xlabel("control b")
plt.ylabel("equilibrium x")
plt.title("Cusp slice and hysteresis branches")
plt.legend()
plt.show()
```
