## Лазер як прототип синергетичної системи

Синергетика Гакена виросла з теорії лазера. Активне середовище містить багато мікроскопічних випромінювачів, чиї спонтанні фази спочатку не узгоджені. У резонаторі вимушене випромінювання та взаємний вплив поля й речовини створюють поріг: одна мода може набути стійкої макроскопічної амплітуди й фазової узгодженості. Цей порядок не задають ззовні для кожного атома — він постає як колективна нестійкість відкритої системи з підведенням і дисипацією енергії.

У дійсному одномодовому напівкласичному наближенні позначимо через $E$ амплітуду поля резонатора, через $P$ — колективну поляризацію, а через $D$ — інверсію населеностей:

$$\dot E=-\kappa E+gP,$$
$$\dot P=-\gamma_\perp P+gED,$$
$$\dot D=\gamma_\parallel(D_0-D)-2gEP.$$

Параметр $D_0$ задається накачуванням, тоді як $\kappa$, $\gamma_\perp$ і $\gamma_\parallel$ є швидкостями релаксації поля, поляризації та інверсії. Лінеаризація підсистеми $(E,P)$ поблизу безлазерного стану $(E,P,D)=(0,0,D_0)$ дає поріг

$$D_c=\frac{\kappa\gamma_\perp}{g^2}.$$

За $D_0<D_c$ флуктуації поля згасають. За $D_0>D_c$ безлазерний стан стає нестійким, а нелінійне виснаження інверсії насичує когерентне поле.


### Від квантових випромінювачів до колективних змінних лазера

Три наведені вище рівняння спираються на кілька фізичних наближень. Перш ніж переходити до числового розрахунку, простежимо їхнє походження. Наше завдання — зрозуміти, як мікроскопічні випромінювачі та спільне поле резонатора приводять до колективних змінних. Поки що ми не виключаємо швидкі змінні: принцип підпорядкування становить наступний крок редукції.

Розглянемо $N$ однакових ефективних дворівневих випромінювачів із верхнім і нижнім лазерними рівнями $|e\rangle$ та $|g\rangle$. Введемо оператори

$$\sigma_j^-=|g\rangle_j\langle e|,\qquad
\sigma_j^+=(\sigma_j^-)^\dagger,\qquad
\sigma_j^z=|e\rangle_j\langle e|-|g\rangle_j\langle g|.$$

Використовуємо три кутові частоти: $\omega_c$ — резонансна частота вибраної моди порожнього резонатора, а $\omega_a=(E_e-E_g)/\hbar$ — частота переходу кожного випромінювача. Опорна частота $\omega_r$ визначає оптичний фазовий множник $e^{-i\omega_r t}$, який виділяємо з оператора поля Гайзенберґа для означення його амплітуди. Цю частоту вибираємо довільно, зазвичай поблизу резонансних частот резонатора й атомів. Вибір $\omega_r=\omega_c$ дає нульову розстройку резонатора $\Delta_c=\omega_c-\omega_r=0$; ця розстройка та атомна розстройка $\Delta_a=\omega_a-\omega_r$ дорівнюють нулю за $\omega_r=\omega_c=\omega_a$. Усі три величини є кутовими частотами й вимірюються в радіанах за секунду.

Оператор пониження описує перехід атома з верхнього рівня на нижній; власні значення $\sigma_j^z$ дорівнюють $+1$ і $-1$. Для однієї моди резонатора оператор $a$ знищує фотон, причому $[a,a^\dagger]=1$. Після виділення опорного фазового множника додатночастотну частину оператора електричного поля подамо як

$$\widehat{\mathcal E}^{(+)}(\mathbf r,t)
=\mathcal E_{\rm vac}u(\mathbf r)a(t)e^{-i\omega_r t}.$$

Отже, **одномодовість** означає, що ми залишаємо один просторовий профіль $u(\mathbf r)$ та його залежну від часу амплітуду. Ця амплітуда не обов'язково дійсна. Припустимо також, що всі випромінювачі мають однакову частоту переходу й однакову дійсну сталу зв'язку $g$ із модою. Тим самим нехтуємо неоднорідним розширенням спектральної лінії та просторовою залежністю зв'язку.

**Представлення операторів.** Нехай $a_H(t)$ та $\sigma^-_{j,H}(t)$ — оператори Гайзенберґа з їхньою часовою залежністю на оптичній частоті. Уведемо амплітуди, з яких виділено опорне коливання:

$$a(t)=e^{i\omega_r t}a_H(t),\qquad
\widetilde\sigma^-_j(t)=e^{i\omega_r t}\sigma^-_{j,H}(t),\qquad
\widetilde\sigma^z_j(t)=\sigma^z_{j,H}(t).$$

Отже, $a_H(t)=a(t)e^{-i\omega_r t}$, що пояснює експоненту в наведеній вище формулі електричного поля. Наприклад, для вільної моди резонатора $\dot a_H=-i\omega_c a_H$, звідки $\dot a=-i(\omega_c-\omega_r)a$. Відповідне атомне рівняння містить $\omega_a-\omega_r$.

Еквівалентно можна вибрати представлення взаємодії з опорним гамільтоніаном

$$H_{\rm ref}=\hbar\omega_r\mathcal N,\qquad
\mathcal N=a^\dagger a+\frac12\sum_j(\sigma_j^z+1).$$

У наближенні обертової хвилі взаємодія зберігає $\mathcal N$. Тому перетворений гамільтоніан дорівнює $H_{\rm lab}-H_{\rm ref}$ з точністю до несуттєвої сталої, що дає розстройки в наведеному нижче гамільтоніані. Надалі тильди над атомними операторами опускаємо. Усталений вислів «обертова система відліку» означає тут обертання комплексних фаз у просторі операторів. Наближення обертової хвилі є окремим кроком: воно усуває контробертові доданки взаємодії.

В електродипольному наближенні та наближенні обертової хвилі гамільтоніан має вигляд

$$\frac{H}{\hbar}=\Delta_c a^\dagger a
+\frac{\Delta_a}{2}\sum_j\sigma_j^z
+g\sum_j(a^\dagger\sigma_j^-+a\sigma_j^+),$$

Тут $\Delta_c=\omega_c-\omega_r$, а $\Delta_a=\omega_a-\omega_r$. Доданок $a^\dagger\sigma_j^-$ описує народження фотона під час переходу випромінювача на нижній рівень; ермітово спряжений доданок описує поглинання. У наближенні обертової хвилі відкидаємо швидко осцилюючі доданки, які одночасно створюють або знищують і фотонне, і атомне збудження.

На відміну від безрозмірного числового прикладу, у цій мікроскопічній моделі спочатку використовуємо фізичний час: $g$, розстройки частот і швидкості релаксації мають розмірність оберненого часу. Вибір одиниці часу дає змогу перейти до безрозмірних параметрів.


### Накачування, втрати та квантове керувальне рівняння

Самого гамільтоніана недостатньо для опису працюючого лазера: накачування підводить енергію, а втрати її відводять. Найпростіша марковська модель взаємодії з резервуарами задається рівнянням

$$\dot\rho=-\frac{i}{\hbar}[H,\rho]+2\kappa\mathcal D[a]\rho
+\sum_j\left(w\mathcal D[\sigma_j^+]\rho
+\Gamma\mathcal D[\sigma_j^-]\rho
+\frac{\gamma_\phi}{2}\mathcal D[\sigma_j^z]\rho\right),$$

$$\mathcal D[L]\rho=L\rho L^\dagger-\frac12\{L^\dagger L,\rho\}.$$

Тут $\rho$ — оператор густини спільної системи атомів і поля. Дисипатор $\mathcal D[L]$ описує незворотний процес, пов'язаний з оператором $L$: $w$ — ефективна швидкість некогерентного накачування, $\Gamma$ — швидкість атомного розпаду через канали, відмінні від явно врахованої моди резонатора, а $\gamma_\phi$ характеризує втрату атомної фазової когерентності без зміни населеностей. Ефективне накачування моделює дію резервуара або допоміжних рівнів багаторівневого активного середовища, а не резонансне когерентне збудження ізольованого дворівневого атома.

Множник $2\kappa$ фіксує домовленість: $\kappa$ є швидкістю згасання **амплітуди поля**. Для порожнього резонатора

$$\frac{d}{dt}\langle a\rangle=-\kappa\langle a\rangle,
\qquad\frac{d}{dt}\langle a^\dagger a\rangle=-2\kappa\langle a^\dagger a\rangle.$$

Для випромінювача, не зв'язаного з полем резонатора, інверсія $z=\langle\sigma^z\rangle$ задовольняє рівняння

$$\dot z=w(1-z)-\Gamma(1+z)
=-(w+\Gamma)\left(z-\frac{w-\Gamma}{w+\Gamma}\right).$$

Отже, у цій моделі резервуарів

$$\gamma_\parallel=w+\Gamma,\qquad
\gamma_\perp=\frac{w+\Gamma}{2}+\gamma_\phi,\qquad
D_0=N\frac{w-\Gamma}{w+\Gamma}.$$

За відсутності поля резонатора накачування й атомний розпад спрямовують сумарну інверсію до $D_0$. Інверсія виникає за $w>\Gamma$. У докладніших багаторівневих моделях ефективні співвідношення можуть бути іншими. У феноменологічних розрахунках нижче ми змінюємо $D_0$, залишаючи швидкості релаксації сталими. Такий розрахунок не відповідає буквальній зміні лише $w$ у щойно записаній моделі резервуарів.


### Рівняння для середніх і проблема замикання

Введемо колективні оператори

$$S^-=\sum_j\sigma_j^-,\qquad S^+=(S^-)^\dagger,\qquad Z=\sum_j\sigma_j^z.$$

Середнє значення спостережуваної визначається як $\langle O\rangle=\operatorname{Tr}(\rho O)$. Його еволюцію знаходимо з рівняння

$$\frac{d}{dt}\langle O\rangle=\frac{i}{\hbar}\langle[H,O]\rangle
+\text{внески резервуарів}.$$

Використовуючи $[\sigma^+,\sigma^-]=\sigma^z$ та $[\sigma^z,\sigma^\pm]=\pm2\sigma^\pm$, дістаємо

$$\begin{aligned}
\frac{d}{dt}\langle a\rangle
&=-(\kappa+i\Delta_c)\langle a\rangle-ig\langle S^-\rangle,\\
\frac{d}{dt}\langle S^-\rangle
&=-(\gamma_\perp+i\Delta_a)\langle S^-\rangle+ig\langle aZ\rangle,\\
\frac{d}{dt}\langle Z\rangle
&=\gamma_\parallel(D_0-\langle Z\rangle)
+2ig(\langle a^\dagger S^-\rangle-\langle aS^+\rangle).
\end{aligned}$$

Наприклад, комутатор $[a^\dagger S^-+aS^+,S^-]=aZ$ породжує добуток поля та інверсії в рівнянні для поляризації. У межах прийнятої квантової моделі ці рівняння точні. Проте вони не замкнені: середні від добутків, як-от $\langle aZ\rangle$, загалом не дорівнюють добуткам середніх. Рівняння для цих кореляцій містили б наступні моменти.


### Напівкласична факторизація та нормування

Позначимо $\alpha=\langle a\rangle$, $s=\langle S^-\rangle$ і $D=\langle Z\rangle$. У напівкласичному наближенні середнього поля нехтуємо кореляційними внесками між атомними змінними та полем:

$$\langle aZ\rangle\simeq\alpha D,\qquad
\langle a^\dagger S^-\rangle\simeq\alpha^*s,\qquad
\langle aS^+\rangle\simeq\alpha s^*.$$

Наприклад, $\langle aZ\rangle=\alpha D+\langle(a-\alpha)(Z-D)\rangle$, і ми відкидаємо другий доданок. Тоді

$$\begin{aligned}
\dot\alpha&=-(\kappa+i\Delta_c)\alpha-igs,\\
\dot s&=-(\gamma_\perp+i\Delta_a)s+ig\alpha D,\\
\dot D&=\gamma_\parallel(D_0-D)+2ig(\alpha^*s-\alpha s^*).
\end{aligned}$$

Введемо нормовані комплексні амплітуди

$$\mathsf E=\sqrt2\,\alpha,\qquad\mathsf P=-i\sqrt2\,s.$$

Фазовий множник поглинає уявну одиницю в членах взаємодії, а числові множники забезпечують саме ті коефіцієнти, які використано в цьому модулі. Одержуємо комплексні рівняння Максвелла—Блоха:

$$\boxed{\begin{aligned}
\dot{\mathsf E}&=-(\kappa+i\Delta_c)\mathsf E+g\mathsf P,\\
\dot{\mathsf P}&=-(\gamma_\perp+i\Delta_a)\mathsf P+g\mathsf E D,\\
\dot D&=\gamma_\parallel(D_0-D)-2g\operatorname{Re}(\mathsf E^*\mathsf P).
\end{aligned}}$$

Поляризація описує колективну атомну когерентність, а не населеність збудженого рівня; населеності входять через $D$. На цьому етапі модель містить п'ять дійсних змінних: по дві компоненти поля й поляризації та одну інверсію. Факторизація не враховує квантові кореляції та динаміку флуктуацій; її точність не гарантована як завгодно близько до порога або за малої кількості випромінювачів і фотонів.


### Резонанс, фіксована фаза і три дійсні рівняння

Припустимо точний резонанс між модою резонатора й атомним переходом та виберемо систему відліку, що обертається з їхньою спільною частотою: $\Delta_c=\Delta_a=0$. Комплексні рівняння зберігають підпростір

$$\mathsf E=e^{i\theta}E,\qquad\mathsf P=e^{i\theta}P,\qquad E,P\in\mathbb R,$$

де $\theta$ стала. Вибравши цю спільну фазу за початок відліку, дістаємо саме

$$\boxed{\begin{aligned}
\dot E&=-\kappa E+gP,\\
\dot P&=-\gamma_\perp P+gED,\\
\dot D&=\gamma_\parallel(D_0-D)-2gEP.
\end{aligned}}$$

Це підпростір резонансної моделі з фіксованою фазою. Вибір одного спільного початку відліку фаз загалом не дозволяє одночасно зробити дійсними дві комплексні амплітуди з довільними фазами; за розстройки частот і для загальної динаміки відносної фази потрібні комплексні рівняння.

Баланс збуджень пояснює множник $2$. Когерентна складова числа фотонів дорівнює $n_{\rm coh}=|\alpha|^2=E^2/2$, а середня населеність верхнього рівня — $N_e=(N+D)/2$. Залишаючи лише внески взаємодії, маємо

$$\left.\dot n_{\rm coh}\right|_{\rm int}=gEP,\qquad
\left.\dot N_e\right|_{\rm int}=-gEP.$$

Один перехід на нижній рівень додає один фотон, але зменшує $D=N_e-N_g$ на два. На графіках нижче величина $E^2$ є нормованою інтенсивністю, пропорційною числу фотонів у когерентній складовій. У повному квантовому описі $\langle a^\dagger a\rangle$ не обов'язково дорівнює $|\langle a\rangle|^2$.


### Від колективної нестійкості до параметра порядку

Лінеаризація поблизу $(E,P,D)=(0,0,D_0)$ дає $\delta\dot D=-\gamma_\parallel\delta D$ та

$$\frac{d}{dt}\begin{pmatrix}\delta E\\\delta P\end{pmatrix}
=\begin{pmatrix}-\kappa&g\\gD_0&-\gamma_\perp\end{pmatrix}
\begin{pmatrix}\delta E\\\delta P\end{pmatrix}.$$

Два показники зростання визначаються рівнянням

$$\lambda^2+(\kappa+\gamma_\perp)\lambda
+\kappa\gamma_\perp-g^2D_0=0.$$

Слід матриці від'ємний, а її визначник змінює знак за $D_c=\kappa\gamma_\perp/g^2$. Отже, на порозі колективне збурення поля й поляризації переходить від згасання до зростання. За $D_0=D_c$ одне власне значення дорівнює нулю, а друге — $-(\kappa+\gamma_\perp)$; критичний власний вектор задовольняє $\delta P=(\kappa/g)\delta E$. Тому повільна змінна описує зв'язану моду поля й речовини, яку можна параметризувати амплітудою $E$.

Для ненульового стаціонарного поля перші два рівняння дають

$$P_s=\frac{\kappa}{g}E_s,\qquad D_s=D_c.$$

Із третього рівняння тоді випливає

$$E_s^2=\frac{\gamma_\parallel}{2\kappa}(D_0-D_c).$$

Поле виснажує інверсію доти, доки підсилення не зрівняється з втратами. Саме цю стаціонарну гілку зображено нижче; її стійкість далеко вище порога потребує окремого аналізу. Поле виникає завдяки колективній поляризації й водночас змінює середовище, що його породжує: у цьому полягає циклічна причинність у трактуванні Гакена.

Детерміністичні рівняння залишають стан із точно нульовими $E=P=0$ незмінним навіть вище порога. Мала початкова амплітуда в числовому прикладі задає збурення; фізичний механізм запуску забезпечують квантові флуктуації. У скінченному квантовому лазері без зовнішнього сигналу, що задає фазу, ансамблеве середнє $\langle a\rangle$ може залишатися нульовим навіть за ненульового числа фотонів. Ненульова напівкласична амплітуда описує вибрану фазу або окрему реалізацію на проміжках часу, коли дифузією фази можна знехтувати.

У цьому виведенні ми зберегли всі три змінні поля й середовища. Подальше наближення підпорядкування потребує додаткового обґрунтування через розділення часових масштабів: поблизу порога швидкість еволюції критичної колективної моди стає малою порівняно зі швидкостями релаксації некритичних мод. Саме по собі напівкласичне наближення ще не дозволяє покласти $\dot P$ або $\dot D$ рівними нулю.

**Додаткове читання.** G. J. de Valcárcel, E. Roldán, F. Prati, [*Semiclassical Theory of Amplification and Lasing*](https://arxiv.org/abs/quant-ph/0605084): виведення рівнянь Максвелла—Блоха та їхній зв'язок із багаторівневими активними середовищами. Інтерпретацію через нестійкість, параметри порядку та принцип підпорядкування розглянуто в основному джерелі курсу — *Synergetics* Гакена.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "axes.grid": True,
    "grid.alpha": 0.25,
    "font.size": 11,
})
rng = np.random.default_rng(20260904)

from scipy.integrate import solve_ivp

kap, gamma_perp, gamma_parallel, g = 1.0, 8.0, 0.7, 2.0
D_c = kap*gamma_perp/g**2

def laser(t, state, D0):
    E, P, D = state
    return (-kap*E + g*P,
            -gamma_perp*P + g*E*D,
            gamma_parallel*(D0-D) - 2*g*E*P)

t_eval = np.linspace(0, 60, 4000)
pumps = [0.7*D_c, 1.02*D_c, 1.6*D_c]

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
steady_intensity = []
for D0 in pumps:
    sol = solve_ivp(laser, (0, 60), [1e-4, 0.0, D0], args=(D0,), t_eval=t_eval,
                    rtol=1e-9, atol=1e-11)
    E, P, D = sol.y
    ax[0].plot(sol.t, E**2, label=fr"$D_0/D_c={D0/D_c:.2f}$")
    steady_intensity.append(np.mean(E[sol.t > 50]**2))
ax[0].set(xlabel="time", ylabel=r"intensity $E^2$", title="Laser threshold dynamics")
ax[0].legend()

pump_scan = np.linspace(0.4*D_c, 2.0*D_c, 60)
theory = gamma_parallel*np.maximum(pump_scan-D_c, 0)/(2*kap)
ax[1].plot(pump_scan/D_c, theory, lw=2)
ax[1].set(xlabel=r"pump $D_0/D_c$", ylabel=r"stationary $E^2$",
          title="Order parameter above threshold")
plt.tight_layout()
```

## Принцип підпорядкування і циклічна причинність

Якщо матеріальні змінні релаксують швидше за поле, покладемо $\dot P\approx0$ і $\dot D\approx0$:

$$P\approx\frac{g}{\gamma_\perp}ED,\qquad
D\approx\frac{D_0}{1+2g^2E^2/(\gamma_\perp\gamma_\parallel)}.$$

Підстановка в рівняння поля та розклад поблизу $E=0$ дають

$$\dot E\approx \mu E-\beta E^3,\qquad
\mu=\kappa\left(\frac{D_0}{D_c}-1\right),\quad \beta>0.$$

Когерентне поле $E$ є повільним параметром порядку. Швидкі атомні змінні наближено стають його функціями — вони **підпорядковуються** полю. Водночас саме атоми породжують поле через вимушену поляризацію, а поле узгоджує атоми. Для Гакена цей взаємний зв'язок був взірцем циклічної причинності: мікроскопічна активність створює макроскопічний параметр порядку, який у відповідь обмежує мікроскопічну динаміку.


```python
# Test the slaving relation on the above-threshold trajectory.
D0 = 1.6*D_c
sol = solve_ivp(laser, (0, 60), [1e-4, 0.0, D0], args=(D0,), t_eval=t_eval,
                rtol=1e-9, atol=1e-11)
E, P, D = sol.y
P_slaved = g*E*D/gamma_perp
D_slaved = D0/(1 + 2*g**2*E**2/(gamma_perp*gamma_parallel))
mask = sol.t > 10

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].plot(sol.t, P, label="full P")
ax[0].plot(sol.t, P_slaved, "--", label="slaved P(E,D)")
ax[0].set(xlabel="time", ylabel="polarization", title="Fast polarization follows the field")
ax[0].legend()
ax[1].plot(E[mask]**2, D[mask], ".", ms=1.5, label="full dynamics")
order = np.argsort(E[mask]**2)
ax[1].plot(E[mask][order]**2, D_slaved[mask][order], "k--", label="slaved inversion")
ax[1].set(xlabel=r"intensity $E^2$", ylabel="inversion D", title="Gain saturation")
ax[1].legend()
plt.tight_layout()
```
