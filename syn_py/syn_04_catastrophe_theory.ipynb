## Prerequisite bridge: gradient flow, metastability, and hysteresis

In a one-dimensional gradient system $\dot x=-\partial_xV(x;\mu)$, the potential decreases along trajectories because $\dot V=-(\partial_xV)^2\le0$. Local minima of $V$ are stable equilibria and local maxima are unstable. Multiple minima represent competing metastable macrostates separated by barriers.

Hysteresis is path dependence under parameter variation: a state may persist until its local minimum disappears at a fold, then jump to another branch. A potential is powerful when the dynamics is genuinely gradient-like; driven systems with circulation in state space generally require more than a scalar landscape.


## Where catastrophe theory comes from: robust geometry

Arnold introduces the subject through two sources: **singularities of smooth maps** and **bifurcations of dynamical systems**. A smooth surface can project onto a plane with a sharp outline even though the surface itself is smooth. The important question is which features survive small perturbations, rather than the exact shape of a particular drawing.

<img src="./figs/fig_whitneymaps.png" alt="Whitney fold and cusp projections from Arnold" width="500">

*Arnold’s illustration of Whitney singularities: a sphere projection with a fold (left), and a smooth surface whose projection has a cusp (right). Source: the instructor-supplied Arnold book, printed p. 5, Figs. 1–2. Coordinates $(x_1,x_2)$ belong to the source surface; $(y_1,y_2)$ to the target plane.*

For a smooth map $F:\mathbb R^2\to\mathbb R^2$, a point is **regular** if $\det DF\ne0$; locally the map is invertible. A **singular point** loses rank. Its image is a **critical value**: keep the point on the source surface distinct from its image in the target plane.

Locally, structural stability means that a sufficiently small smooth perturbation has the same type after smooth, invertible coordinate changes in source and target, allowing the point to move. Schematically, equivalent maps satisfy $\widetilde F=\psi\circ F\circ\phi^{-1}$. Closeness must control the relevant derivatives, not merely function values. Here we discuss local types; global stability has additional conditions.

**Two different meanings of stability:** a fold or cusp can be a structurally stable singularity of a map; an asymptotically stable equilibrium attracts nearby trajectories while keeping them nearby. One statement does not imply the other. An equilibrium can persist under perturbation and still be a repeller.


### A calculus-to-catastrophes roadmap

The following table adapts Gilmore's Table 2.1 to familiar multivariable calculus. Let $V:\mathbb R^n\to\mathbb R$ be smooth, with all derivatives evaluated at the point $p$. Write $H=D^2V(p)$, $\ell=\dim\ker H$ (the **corank**), and $Q_m(\mathbf w)=\sum_{j=1}^m\sigma_jw_j^2$, $\sigma_j=\pm1$. Additive constants are omitted; coordinates are local and smoothly invertible.

| Conditions | Local form of $V-V(p)$ | Result and interpretation |
|---|---|---|
| $\nabla V(p)\ne0$ | $y_1$ | Implicit/inverse-function theorem: straighten a regular level surface; no equilibrium of $-\nabla V$ here. |
| $\nabla V(p)=0$, $\det H\ne0$ | $Q_n(\mathbf y)$ | **Morse lemma:** a nondegenerate critical point is locally quadratic; its index counts minus signs. |
| $\nabla V(p)=0$, $\operatorname{rank}H=n-\ell<n$ | $g(z_1,\ldots,z_\ell)+Q_{n-\ell}(\mathbf w)$ | **Splitting lemma:** retain the $\ell$ essential degenerate coordinates; $g(0)=0$, $Dg(0)=D^2g(0)=0$. |
| Generic smooth $k$-control families, $k\le5$, under the usual equivalence of unfoldings | $\mathcal C_\alpha(\mathbf z;\boldsymbol a)+Q_{n-\ell}(\mathbf w)$ | **Thom–Mather classification/genericity:** a finite catalogue of local stable unfoldings; the fold and cusp are the first examples. |

Here an **unfolding** adds independent control terms that describe nearby perturbations of a degenerate function. A **germ** means the function's local behavior near the chosen point. The catalogue concerns local generic smooth potential families; symmetry constraints can change which unfoldings are allowed. The familiar seven elementary types use at most four controls; the five-control catalogue is larger. The last row is not a classification of every global dynamical system.

**Why this matters for synergetics.** Near a transition, justified slaving reduces the dynamics to a few slow order parameters. Splitting then separates nondegenerate directions of a potential, and normal forms organize the remaining essential behavior into a few robust structural models. Slaving uses time-scale separation; splitting is a geometric statement about a function and by itself proves no fast relaxation. Different physical systems can share a local model while retaining different coefficients, parameter ranges, and observables. Non-gradient transitions require the corresponding dynamical normal forms.


### Morse critical points: the robust building blocks

For a scalar potential $V(\mathbf q)$, a critical point $p$ satisfies $\nabla V(p)=0$. It is **nondegenerate** if the Hessian $H_{ij}=\partial_i\partial_jV(p)$ is invertible. The **Morse index** $r$ is the number of negative Hessian eigenvalues. The Morse lemma says that suitable smooth local coordinates give the exact expression

$$V=V(p)-z_1^2-\cdots-z_r^2+z_{r+1}^2+\cdots+z_n^2.$$

This is more than truncating a Taylor series: near a nondegenerate point, higher-order terms can be removed by a coordinate change. For $\dot{\mathbf q}=-\nabla V$, the linearization is $-H$:

| Two-dimensional type | Morse index | Gradient dynamics |
|---|---|---|
| Minimum, $+z_1^2+z_2^2$ | 0 | Attractor |
| Saddle, $-z_1^2+z_2^2$ | 1 | One growing and one decaying direction |
| Maximum, $-z_1^2-z_2^2$ | 2 | Repeller |

An invertible Hessian lets the implicit-function theorem continue the critical point under small $C^2$ perturbations; its index persists. The point need not remain at the origin. A nonlinear coordinate change also changes the gradient's mobility/metric, so the lemma does not claim identical Euclidean trajectories or relaxation times.

**Figure 1.** Bowls, passes, and hilltops above; level contours and downhill arrows after adding $\varepsilon(u+v/2)$ below. Change `morse_epsilon`: the critical point moves (star), but its type survives. These are original model illustrations.


```python
# morse_geometry: local critical points and a small tilt.
import numpy as np
import matplotlib.pyplot as plt

morse_epsilon = 0.25
mesh = np.linspace(-1.5, 1.5, 65)
UU, VV = np.meshgrid(mesh, mesh)
fig = plt.figure(figsize=(12, 7.2), constrained_layout=True)
morse_indices = []
for column, (su, sv, name) in enumerate([(1, 1, 'Minimum'), (-1, 1, 'Saddle'), (-1, -1, 'Maximum')]):
    surface = (su*UU**2+sv*VV**2)/2
    axis = fig.add_subplot(2, 3, column+1, projection='3d')
    axis.plot_surface(UU, VV, surface, cmap='cividis', alpha=0.88,
                      rcount=30, ccount=30, linewidth=0)
    index = int(su < 0)+int(sv < 0)
    morse_indices.append(index)
    axis.set(xlabel=r'$u$', ylabel=r'$v$', title=f'{name}: index {index}')
    axis.text2D(0.90, 0.87, r'$V$', transform=axis.transAxes)
    axis.view_init(elev=27, azim=-55)
    axis = fig.add_subplot(2, 3, column+4)
    tilted = surface+morse_epsilon*(UU+VV/2)
    axis.contour(UU, VV, tilted, levels=12, colors='0.55', linewidths=0.7)
    arrows = np.linspace(-1.35, 1.35, 11)
    AU, AV = np.meshgrid(arrows, arrows)
    FU, FV = -su*AU-morse_epsilon, -sv*AV-morse_epsilon/2
    speed = np.hypot(FU, FV)
    axis.quiver(AU, AV, FU/(speed+1e-12), FV/(speed+1e-12),
                color='C0', scale=19, width=0.005)
    critical = np.array([-morse_epsilon/su, -morse_epsilon/(2*sv)])
    assert np.allclose(np.array([su, sv])*critical+[morse_epsilon, morse_epsilon/2], 0)
    assert np.count_nonzero(np.linalg.eigvalsh(np.diag([su, sv])) < 0) == index
    axis.plot(0, 0, 'ko', mfc='white', label='original point')
    axis.plot(*critical, '*', color='C1', ms=13, label='shifted point')
    axis.set(xlabel=r'$u$', ylabel=r'$v$', aspect='equal',
             xlim=(-1.5, 1.5), ylim=(-1.5, 1.5))
    if column == 0:
        axis.legend(fontsize=8, loc='upper left')
assert morse_indices == [0, 1, 2]
print('Morse indices:', morse_indices)
plt.show()
```

### Whitney folds and cusps: singular maps can be robust

The sphere's vertical projection $\pi(X,Y,Z)=(X,Y)$ has two preimages inside its outline, one on the outline, and none outside. At the equator, the tangent plane contains the viewing direction, so the projection loses rank. Near an ordinary outline point its local form is a **fold**. For generic smooth maps between two-dimensional surfaces the local singular types are folds and cusps:

| Map $(u,v)\mapsto(u,y)$ | Singular set in the source | Critical values in the target |
|---|---|---|
| Fold: $y=v^2$ | $v=0$ | $y=0$ |
| Cusp: $y=v^3+uv$ | $u=-3v^2$ | $(u,y)=(-3t^2,-2t^3)$ |

Indeed, the Jacobian determinants are $2v$ and $3v^2+u$. The cusp's critical-value curve satisfies $4u^3+27y^2=0$. A point inside the wedge has three preimages; outside, one. On a fold boundary two preimages merge; at the cusp all three coalesce. A double root plus a simple root gives two distinct preimages at a noncusp boundary point.

**Figure 2.** Read source surface → projection → critical-value curve. Red marks the rank-loss set and its image. The cusp surface is smooth: the singularity belongs to its projection. These redrawings use explicit equations rather than reproducing Arnold's scan. Their robustness concerns local geometry under generic perturbations, not attraction of trajectories.


```python
# whitney_geometry: sphere outline, fold, cusp surface, and its critical values.
fig = plt.figure(figsize=(12, 9), constrained_layout=True)
axis = fig.add_subplot(2, 2, 1, projection='3d')
theta = np.linspace(0, 2*np.pi, 80)
polar = np.linspace(0, np.pi, 25)
TT, PP = np.meshgrid(theta, polar)
axis.plot_wireframe(np.sin(PP)*np.cos(TT), np.sin(PP)*np.sin(TT), np.cos(PP),
                    rstride=3, cstride=7, color='0.6', linewidth=0.5)
ground = -1.4
axis.plot(np.cos(theta), np.sin(theta), np.zeros_like(theta), color='C3', lw=2)
axis.plot(np.cos(theta), np.sin(theta), ground*np.ones_like(theta), color='C3', lw=2)
RR, AA = np.meshgrid(np.linspace(0, 1, 10), theta)
axis.plot_surface(RR*np.cos(AA), RR*np.sin(AA), ground*np.ones_like(RR),
                  color='C0', alpha=0.10, linewidth=0)
for angle in np.linspace(0, 2*np.pi, 7)[:-1]:
    axis.plot([np.cos(angle)]*2, [np.sin(angle)]*2, [0, ground],
              '--', color='0.6', lw=0.7)
axis.set(xlabel=r'$X$', ylabel=r'$Y$', zlabel=r'$Z$', title='Sphere: a fold along the outline')
axis.view_init(elev=24, azim=-60)

axis = fig.add_subplot(2, 2, 2)
axis.axhspan(0, 1.4, color='C0', alpha=0.13)
axis.axhline(0, color='C3', lw=2)
axis.text(0, 0.75, '2 preimages', ha='center')
axis.text(0, 0.08, '1 preimage on the fold', ha='center', va='bottom')
axis.text(0, -0.65, '0 preimages', ha='center')
axis.set(xlim=(-1.3, 1.3), ylim=(-1.2, 1.4), xlabel=r'$u$', ylabel=r'$y$',
          title=r'Fold projection: $(u,v)\mapsto(u,v^2)$')

axis = fig.add_subplot(2, 2, 3, projection='3d')
cu = np.linspace(-1.65, 0.6, 28)
cv = np.linspace(-1.1, 1.1, 42)
CU, CV = np.meshgrid(cu, cv)
CY = CV**3+CU*CV
axis.plot_wireframe(CU, CY, CV, rstride=3, cstride=3, color='C0',
                    alpha=0.5, linewidth=0.6)
parameter = np.linspace(-0.74, 0.74, 301)
fold_u, fold_y = -3*parameter**2, -2*parameter**3
axis.plot(fold_u, fold_y, parameter, color='C3', lw=2.5)
axis.plot(fold_u, fold_y, ground*np.ones_like(parameter), color='C3', lw=2)
for t in [-0.6, 0, 0.6]:
    axis.plot([-3*t*t]*2, [-2*t**3]*2, [t, ground], '--', color='0.5', lw=0.8)
axis.set(xlabel=r'$u$', ylabel=r'$y=v^3+uv$', zlabel=r'$v$',
          title='Smooth surface; singular projection')
axis.view_init(elev=25, azim=-62)

axis = fig.add_subplot(2, 2, 4)
negative_u = np.linspace(-1.65, 0, 400)
limit_y = 2*(-negative_u/3)**1.5
axis.fill_between(negative_u, -limit_y, limit_y, color='C0', alpha=0.13)
axis.plot(fold_u, fold_y, color='C3', lw=2)
axis.plot([-1, -1], [0, 0.6], 'ko', ms=4)
axis.annotate('3 preimages', (-1, 0), xytext=(-1.5, -0.13))
axis.annotate('1 preimage', (-1, 0.6), xytext=(-0.75, 0.64))
axis.set(xlabel=r'$u$', ylabel=r'$y$', title='Cusp critical-value curve',
          xlim=(-1.7, 0.45), ylim=(-0.9, 0.9))
assert np.allclose(parameter**3+fold_u*parameter, fold_y)
assert np.allclose(3*parameter**2+fold_u, 0)
assert np.allclose(4*fold_u**3+27*fold_y**2, 0)
for target_y, expected_count in [(0.0, 3), (0.6, 1)]:
    roots = np.roots([1, 0, -1, -target_y])
    assert np.count_nonzero(np.abs(roots.imag) < 1e-9) == expected_count
plt.show()
```

### From Morse theory to catastrophes in parameter families

Morse theory also studies sublevel sets $M_c=\{\mathbf q:V(\mathbf q)\le c\}$. If the intervening level strip is compact and contains no critical points, crossing that strip causes no change of topology. In our example $W(u,v)=(u^2-1)^2+v^2$, minima at $(\pm1,0)$ have value 0 and index 0; the saddle at $(0,0)$ has value 1 and index 1. For $0<c<1$ there are two components; for $c>1$ they join. This is topology of a **fixed** landscape as the level changes, not a dynamical bifurcation caused by changing its controls.

A catastrophe concerns a **family** in which a critical point becomes degenerate. For the fold potential $G(x;\lambda)=x^3/3-\lambda x$, $G_x=x^2-\lambda$: for $\lambda>0$ a minimum at $+\sqrt\lambda$ and maximum at $-\sqrt\lambda$ exist; at zero they collide with $G_{xx}=0$; for $\lambda<0$ neither remains. This cubic is a local model, not a globally confining potential.

For the quartic potential already used below, the equilibrium surface is the smooth graph

$$b=-(x^3+ax).$$

Its projection onto $(a,b)$ is precisely the Whitney cusp map above, with $u=a$, $v=x$, and **$y=-b$**. Projection rank is lost where $V_{xx}=3x^2+a=0$. At an ordinary fold $V_{xxx}=6x\ne0$; at the cusp $x=a=b=0$, also $V_{xxx}=0$ but $V_{xxxx}=6\ne0$. Thus a generic fold needs one control and a cusp two; nondegenerate directions retain quadratic terms while the degenerate direction needs higher powers. The map's cusp is robust as a two-control family, although its potential at the cusp is not Morse. Geometry identifies possible equilibria; the gradient law selects attracting minima.

**Figure 3.** Left: the sublevel sets of $W$ merge at the saddle value. Right: changing $\lambda$ creates or destroys two critical points of $G$. Compare these two different changes before continuing with the existing cusp calculations.


```python
# morse_sublevels: topology of one potential versus a family of potentials.
from scipy.ndimage import label as connected_components
line = np.linspace(-1.7, 1.7, 501)
SU, SV = np.meshgrid(line, line)
WW = (SU**2-1)**2+SV**2
fig, axes = plt.subplots(1, 2, figsize=(12, 4.3), constrained_layout=True)
contours = axes[0].contour(SU, SV, WW, levels=[0.2, 0.8, 1.0, 1.3],
                           colors=['C0', 'C1', 'C3', 'black'])
axes[0].clabel(contours, fmt='%g', fontsize=9)
axes[0].plot([-1, 1], [0, 0], 'o', color='C0', label='minima: W=0')
axes[0].plot(0, 0, 'x', color='C3', ms=8, label='saddle: W=1')
axes[0].set(xlabel=r'$u$', ylabel=r'$v$', aspect='equal', title='Sublevel sets join at a saddle')
axes[0].legend(fontsize=8, loc='upper right')
counts = []
for level, expected in [(0.2, 2), (0.8, 2), (1.3, 1)]:
    count = connected_components(WW <= level)[1]
    counts.append(int(count))
    assert count == expected
assert np.allclose(np.linalg.eigvalsh(np.diag([8.0, 2.0])), [2, 8])
assert np.allclose(np.linalg.eigvalsh(np.diag([-4.0, 2.0])), [-4, 2])
print('Grid component counts at levels 0.2, 0.8, 1.3:', counts)

state = np.linspace(-1.3, 1.3, 600)
for control in [-0.4, 0.0, 0.4]:
    value = state**3/3-control*state
    axes[1].plot(state, value, label=fr'$\lambda={control:g}$')
    if control > 0:
        for root in [-np.sqrt(control), np.sqrt(control)]:
            assert np.isclose(root**2-control, 0)
            axes[1].plot(root, root**3/3-control*root,
                         'o' if root > 0 else 'x', color='black', ms=7)
axes[1].set(xlabel=r'$x$', ylabel=r'$G(x;\lambda)$', title='Fold: a minimum and maximum collide')
axes[1].legend()
plt.show()
```

### Reading Gilmore's catalogue of elementary catastrophes

This compact version of Gilmore's Table 2.2 lists a germ $g$ and control perturbation $P$; the potential is $g+P$, plus any split-off nondegenerate quadratic terms. The number $k$ counts essential independent controls, not state variables. The $A$ forms have one essential state variable; the $D$ and $E$ forms have two. The superscripts retain the real sign alternatives shown in the source.

| Type | $k$ | Germ $g$ | Perturbation $P$ |
|---|---:|---|---|
| $A_2$ (fold) | 1 | $x^3$ | $a_1x$ |
| $A_3^{\pm}$ (cusp) | 2 | $\pm x^4$ | $a_1x+a_2x^2$ |
| $A_4$ (swallowtail) | 3 | $x^5$ | $\sum_{j=1}^{3}a_jx^j$ |
| $A_5^{\pm}$ (butterfly) | 4 | $\pm x^6$ | $\sum_{j=1}^{4}a_jx^j$ |
| $A_6$ | 5 | $x^7$ | $\sum_{j=1}^{5}a_jx^j$ |
| $D_4^{-}$ (elliptic umbilic) | 3 | $x^2y-y^3$ | $a_1x+a_2y+a_3y^2$ |
| $D_4^{+}$ (hyperbolic umbilic) | 3 | $x^2y+y^3$ | $a_1x+a_2y+a_3y^2$ |
| $D_5$ (parabolic umbilic) | 4 | $x^2y+y^4$ | $a_1x+a_2y+a_3x^2+a_4y^2$ |
| $D_6^{-}$ | 5 | $x^2y-y^5$ | $a_1x+a_2y+a_3x^2+a_4y^2+a_5y^3$ |
| $D_6^{+}$ | 5 | $x^2y+y^5$ | $a_1x+a_2y+a_3x^2+a_4y^2+a_5y^3$ |
| $E_6^{\pm}$ | 5 | $x^3\pm y^4$ | $a_1x+a_2y+a_3xy+a_4y^2+a_5xy^2$ |

The table is a reading map, not material to memorize. The $A_2$ and positive $A_3$ rows are the models used here. Multiplying our fold potential by 3 gives $x^3-3\lambda x$; multiplying the cusp potential by 4 gives $x^4+2ax^2+4bx$. Thus their coefficients agree after a positive rescaling. Signs still matter for minima and gradient stability: $+x^4$ and $-x^4$ behave differently. No equilibrium is guaranteed to be attracting merely because its germ appears in the catalogue.

## Cusp normal form

The canonical potential

$$V(x;a,b)=\frac{x^4}{4}+\frac{a x^2}{2}+bx$$

generates gradient dynamics $\dot x=-\partial_xV=-(x^3+ax+b)$. Equilibria satisfy $x^3+ax+b=0$ and are stable when $V''=3x^2+a>0$. Folds obey both $V'=0$ and $V''=0$, giving

$$a=-3x^2,\qquad b=2x^3,$$

and the cusp discriminant $4a^3+27b^2=0$.


```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    "figure.figsize": (7.2, 4.2),
    "axes.grid": True,
    "grid.alpha": 0.25,
    "font.size": 11,
})
rng = np.random.default_rng(20260904)

def real_equilibria(a, b, tol=1e-9):
    roots = np.roots([1.0, 0.0, a, b])
    roots = roots[np.abs(roots.imag) < tol].real
    return np.sort(roots)

a_values = [-1.4, -0.5, 0.4]
x = np.linspace(-2.0, 2.0, 800)
fig, ax = plt.subplots(1, 3, figsize=(13, 3.8), sharey=True)
for axis, a in zip(ax, a_values):
    for b in [-0.4, 0.0, 0.4]:
        V = x**4/4 + a*x**2/2 + b*x
        axis.plot(x, V, label=fr"$b={b}$")
    axis.set(title=fr"$a={a}$", xlabel="state x")
ax[0].set_ylabel("potential V")
ax[-1].legend()
plt.tight_layout()
```

```python
# Equilibrium surface projected onto the (b, x) plane for fixed a < 0.
a = -1.0
x_branch = np.linspace(-1.4, 1.4, 1000)
b_branch = -(x_branch**3 + a*x_branch)
stable = 3*x_branch**2 + a > 0

plt.plot(b_branch[stable], x_branch[stable], "C0.", ms=2, label="stable")
plt.plot(b_branch[~stable], x_branch[~stable], "C3.", ms=2, label="unstable")
plt.xlabel("control b")
plt.ylabel("equilibrium x")
plt.title("Cusp slice and hysteresis branches")
plt.legend()
plt.show()
```
